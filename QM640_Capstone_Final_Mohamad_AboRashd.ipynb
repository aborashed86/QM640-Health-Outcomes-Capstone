## Study Overview

This study investigates why countries with broadly comparable economic resources and health-system capacity can achieve markedly different levels of **healthy life expectancy (HALE)**. The **master panel covers 217 countries and economies from 2000–2023**, while the primary analytical sample comprises **185 countries over 2000–2021**, reflecting the common usable window of the principal WHO outcomes used in the models. Data for 2022–2023 remain retained in the master dataset where available but are not used to extend analyses whose HALE and premature NCD mortality outcomes are unavailable after 2021. Premature mortality from non-communicable diseases (NCDs) is analysed as a complementary population-health outcome, and Saudi Arabia is retained as an applied benchmarking case within the global framework.

The notebook integrates three analytical perspectives because the study's questions require different forms of evidence.

| Analytical perspective | Question | Method | Interpretive boundary |
|---|---|---|---|
| **Explanatory** | Which factors are associated with HALE, conversion efficiency, and premature NCD mortality within countries over time? | Resource-demography benchmarking, two-way fixed effects, country-clustered standard errors, wild cluster bootstrap, and false-discovery-rate correction | Estimates associations and does **not** establish causal effects. |
| **Predictive and explainable** | Can HALE and premature NCD mortality be predicted for countries not observed during model training, and which variables contribute to those predictions? | Candidate machine-learning models, country-grouped cross-validation, missing-data sensitivity analysis, repeated validation, locked-split robustness analysis, block ablation, and permutation importance | Evaluates out-of-sample prediction and model explanation; predictive importance is not an effect size. |
| **Action-oriented** | Which statistically and predictively relevant characteristics are plausibly modifiable by health-system decision makers? | Ex ante actionability classification integrated with explanatory and predictive evidence | Supports prioritisation for further evaluation, not causal policy prescription. |

Contextual variables such as ageing and national income may be highly predictive while offering limited short-run policy discretion. Conversely, an institutionally actionable variable may fail to demonstrate a robust association or meaningful incremental predictive value. The notebook therefore keeps explanatory association, predictive contribution, and institutional actionability conceptually separate.

> **Interpretive constraint.** The study uses observational panel data without an exogenous source of variation. Accordingly, coefficients are interpreted as associations, machine-learning results as predictive evidence, permutation importance as model explanation, and policy implications as priorities for evaluation rather than proven causal effects.

Taken together, the analytical design is organised around five connected themes: **(1)** health-system conversion efficiency, **(2)** actionable determinants of healthy life expectancy, **(3)** explainable machine-learning prediction of HALE and premature NCD mortality, **(4)** the relationship of health expenditure to healthy versus total life expectancy, and **(5)** application of the resulting benchmarking framework to Saudi Arabia. These themes motivate the problem addressed by the study and structure the research questions that follow.

## Problem Statement

Despite substantial increases in health expenditure and improvements in health-system capacity, countries with similar economic resources continue to achieve markedly different levels of healthy life expectancy. Existing research has primarily examined the effects of health spending or isolated health-system inputs, with comparatively limited attention to how efficiently health systems convert available resources into healthy years of life, which modifiable factors contribute most to this conversion, and whether they improve prediction of population health outcomes across countries. Consequently, policymakers lack robust evidence regarding which actionable health-system characteristics are most strongly associated with better health outcomes and where investments may yield the greatest improvements. This study addresses that gap by analyzing a longitudinal panel of 185 countries from 2000 to 2021 to evaluate health-system conversion efficiency, identify actionable determinants of healthy life expectancy, develop explainable machine-learning models to predict healthy life expectancy and premature NCD mortality, compare how health expenditure relates to healthy life expectancy versus total life expectancy, and demonstrate the analytical framework through a benchmarking application to Saudi Arabia.

### Research Gap

The principal gap is not the absence of international evidence on health expenditure, health-system inputs, or longevity individually, but the limited integration of these strands within a single longitudinal framework that simultaneously evaluates **conversion efficiency**, distinguishes **actionable from contextual determinants**, tests **generalisation to unseen countries with explainable machine learning**, compares **healthy and total longevity**, and translates the combined evidence into a transparent **country-benchmarking application**.

## **2  Loading the panel**

### Purpose
Read the master panel and confirm which version of the data is in memory.

### Rationale
Two versions of this panel exist and they are not interchangeable. Version 5 carries an NCD policy index built from every item the WHO survey asked in a given wave — four items in 2013, twelve by 2021 — which makes the index fall between waves for reasons that have nothing to do with policy. Version 6 replaces it with a balanced nine-item index. Loading the wrong file would produce a methodologically invalid policy result and raise no error, so the loader names the file it found and prints the panel's dimensions.

### Method
The file is located rather than assumed, and **either the .csv or the .xlsx is accepted** — an earlier version of this loader searched only for CSV and failed on a runtime where the workbook had just been uploaded successfully. Exact filenames are checked in the working directory, `/content`, the home directory and the usual Google Drive folders; then a bounded recursive search runs; then a fuzzy pattern match catches renamed copies such as `QM640_Master_Panel_v6 (1).csv`. On Colab, if nothing is found the cell offers to mount Drive and, failing that, opens an upload prompt, so a fresh runtime recovers without editing code. Excel parsing is slow, so a workbook is cached to CSV beside itself on first load. If the search still fails it reports which folders it looked in **and which data files it did find there**, which is what provides diagnostic information for resolving the loading failure. Finally the loaded file is checked for the marker column distinguishing v6 from v5.

### Expected output
The resolved file path, approximately 5,208 country-years across 217 countries and economies covering **2000–2023**, with missingness intact, and an explicit version verdict. This is the master-data coverage; downstream analytical models use the common outcome window through 2021.

```python
# ---------------------------------------------------------------------------
# Locate and load the panel. Works in Colab, Jupyter and a plain script.
# Accepts .csv or .xlsx, and does not assume the file is in the working folder.
# ---------------------------------------------------------------------------
import sys
from pathlib import Path

STEMS = ["QM640_Master_Panel_v6",           # required version
         "QM640_Master_Panel_COMBINED_v6",
         "QM640_Master_Panel_COMBINED_v5"]  # tolerated, with a warning
EXTS      = [".csv", ".xlsx", ".xls"]
PREFERRED = [s + e for s in STEMS for e in EXTS]
PATTERNS  = [f"*Master_Panel*{v}*{e}" for v in ("v6","V6","v5","V5") for e in EXTS]

IN_COLAB = "google.colab" in sys.modules
ROOTS    = [Path.cwd(), Path("/content"), Path("/content/drive/MyDrive")]
SEARCH_DIRS = [Path.cwd(), Path.cwd()/"data", Path.home(),
               Path("/content"), Path("/content/data"), Path("/content/sample_data"),
               Path("/content/drive/MyDrive"), Path("/content/drive/MyDrive/QM640"),
               Path("/content/drive/MyDrive/Capstone"),
               Path("/content/drive/Shareddrives")]

def _find(verbose=False):
    """Exact names first (csv preferred over xlsx), then recursive, then fuzzy."""
    seen = []
    for d in SEARCH_DIRS:
        if not d.is_dir(): continue
        seen.append(d)
        for name in PREFERRED:
            p = d/name
            if p.is_file(): return p
    for root in ROOTS:
        if not root.is_dir(): continue
        for name in PREFERRED:
            hit = next(iter(root.rglob(name)), None)
            if hit: return hit
    for root in ROOTS:
        if not root.is_dir(): continue
        for pat in PATTERNS:
            hits = sorted(root.rglob(pat))
            if hits: return hits[0]
    if verbose:
        print("Searched these folders:")
        for d in dict.fromkeys(seen): print("   ", d)
        for root in ROOTS:
            if root.is_dir():
                found = sorted(p.name for p in root.glob("*")
                               if p.suffix.lower() in EXTS)[:12]
                if found: print(f"    data files present in {root}: {found}")
    return None

def _read(p):
    """Load csv or xlsx, picking the sheet that actually holds the panel."""
    if p.suffix.lower() in (".xlsx", ".xls"):
        xl = pd.ExcelFile(p)
        sheet = next((s for s in xl.sheet_names
                      if {"iso3","Year"} <= set(pd.read_excel(xl, s, nrows=0).columns)),
                     xl.sheet_names[0])
        print(f"reading  Excel sheet '{sheet}' — this takes a moment...")
        return pd.read_excel(xl, sheet)
    for enc in ("utf-8", "utf-8-sig", "latin-1"):
        try: return pd.read_csv(p, encoding=enc)
        except UnicodeDecodeError: continue
    return pd.read_csv(p)

SRC = _find()

# --- Colab fallbacks: mount Drive, then offer upload ------------------------
if SRC is None and IN_COLAB:
    try:
        from google.colab import drive
        if not Path("/content/drive/MyDrive").is_dir():
            print("Panel not found locally. Mounting Google Drive to look there...")
            drive.mount("/content/drive")
        SRC = _find()
    except Exception as e:
        print(f"Drive mount skipped ({type(e).__name__}). Falling back to upload.")

if SRC is None and IN_COLAB:
    from google.colab import files
    print("Panel not found. Upload QM640_Master_Panel_v6.csv (or the .xlsx) below.\n")
    up = files.upload()
    for fn in up:
        print(f'  received "{fn}"  ({len(up[fn]):,} bytes)')
    SRC = _find()

if SRC is None:
    _find(verbose=True)
    raise FileNotFoundError(
        "No panel file found.\n"
        f"Accepted names: {STEMS}\n"
        f"Accepted extensions: {EXTS}\n"
        "Put the file next to this notebook, in /content, or in Google Drive under "
        "MyDrive/ or MyDrive/QM640/, then re-run this cell.")

# --- load and verify --------------------------------------------------------
df = _read(SRC)
print(f"loaded   {SRC.name}")
print(f"path     {SRC.parent}")
print(f"shape    {df.shape[0]:,} rows x {df.shape[1]} columns")
print(f"panel    {df.iso3.nunique()} countries, {int(df.Year.min())}–{int(df.Year.max())}")
print(f"missing  {df.isna().mean().mean()*100:.1f}% of all cells")

# Excel is slow to parse. Cache a csv beside it so re-runs are instant.
if SRC.suffix.lower() in (".xlsx", ".xls"):
    try:
        cache = Path.cwd()/f"{SRC.stem}.csv"
        if not cache.exists():
            df.to_csv(cache, index=False)
            print(f"cached   {cache.name} — later runs will load this instead")
    except Exception:
        pass

IS_V6 = "PolicyIndex_allitems_DO_NOT_USE" in df.columns
if IS_V6:
    print("\nversion  v6 — balanced 9-item policy index present. Correct file.")
else:
    print("\n" + "!"*78)
    print("WARNING: this is NOT the v6 panel.")
    print("Its PolicyIndex mixes 4 survey items in 2013 with 12 in 2021, so it falls")
    print("between waves for reasons that have nothing to do with policy. Every result")
    print("in Section 6 and every policy coefficient downstream will be an artefact.")
    print("Everything else in the notebook is unaffected and will run correctly.")
    print("Replace this file with QM640_Master_Panel_v6.csv before reporting policy results.")
    print("!"*78)
```

## **3  Data quality gate**

### Purpose
A formal audit that must be cleared before any analysis runs. Nothing proceeds silently.

### Rationale
An earlier version of this study lost four countries to a silent merge fault. They carried no expenditure data, vanished from every complete-case model without warning, and two findings were reported that later failed to replicate. This gate exists so that cannot recur. It raises a hard stop rather than a printed note, and it distinguishes a genuine structural failure from a known source conflict that is handled deliberately elsewhere.

### Method
Checks across panel-key uniqueness, identifier completeness, year range, and plausible bounds on every outcome and headline input. The HALE-versus-life-expectancy check is expected to flag rows: WHO and World Bank disagree for a small number of country-years, and those are suppressed in Section 8 rather than left as impossible negative values. That one check is therefore excluded from the hard-stop count.

### Expected output
An audit table with an explicit verdict per check, and an assertion that halts the notebook on any hard failure.

```python
def quality_gate(d):
    checks, fails = [], 0
    def chk(name, ok, detail=""):
        nonlocal fails
        checks.append({"Check": name, "Result": "PASS" if ok else "FAIL", "Detail": detail})
        if not ok: fails += 1

    dup = d.duplicated(["iso3","Year"]).sum()
    chk("Panel key (iso3, Year) unique", dup == 0, f"{dup} duplicates")
    chk("iso3 present on every row", d.iso3.isna().sum() == 0, f"{d.iso3.isna().sum()} nulls")
    chk("Year within 1990–2030", d.Year.between(1990,2030).all(), f"{int(d.Year.min())}–{int(d.Year.max())}")

    for v, lo, hi in [("HALE",20,80), ("LifeExpectancy",20,95), ("NCDMortality",0,60),
                      ("UHCIndex",0,100), ("DPTImmunization",0,100), ("OOPExpenditure",0,100),
                      ("HealthExpGDP",0,30), ("Physicians",0,10)]:
        if v in d.columns:
            s = d[v].dropna(); bad = int(((s < lo) | (s > hi)).sum())
            chk(f"{v} in [{lo}, {hi}]", bad == 0, f"{bad} out of range")

    if {"HALE","LifeExpectancy"} <= set(d.columns):
        both = d.dropna(subset=["HALE","LifeExpectancy"])
        neg = int((both.HALE > both.LifeExpectancy).sum())
        chk("HALE never exceeds life expectancy", neg == 0,
            f"{neg} rows (WHO/World Bank source conflict — suppressed in Section 8)")

    if "PolicyIndex" in d.columns:
        pw = sorted(d.loc[d.PolicyIndex.notna(),"Year"].unique().astype(int))
        chk("Policy index on survey waves only", len(pw) <= 5, f"waves {pw}")

    out = pd.DataFrame(checks)
    return out, fails

GATE, N_FAIL = quality_gate(df)
display(GATE)
HARD_FAIL = N_FAIL - int(((GATE.Check.str.contains("never exceeds")) & (GATE.Result=="FAIL")).sum())
print(f"\n{len(GATE)-N_FAIL}/{len(GATE)} passed. Hard failures: {HARD_FAIL}")
assert HARD_FAIL == 0, "Hard data-quality failure — analysis halted."
print("Gate cleared.")
```

### Result
The gate clears. The panel key is unique, identifiers are complete, and every outcome sits within plausible bounds.

### Interpretation
The only flagged check is the HALE-versus-life-expectancy comparison, which is a source disagreement rather than a data fault. Treating it as a hard failure would stop the notebook over something already handled; ignoring it entirely would let impossible values into the morbidity-gap outcome. It is therefore surfaced here and resolved in Section 8.

### Decision
Proceed to catalogue the variables before any selection occurs.

## **3A  From the master panel to the analytical sample and final ML predictor set**

### Purpose
Make the two reductions that can otherwise look arbitrary completely transparent:

1. how the master panel becomes the **2000–2021, 185-country analytical frame**; and  
2. how a wide master table becomes the **18-predictor primary machine-learning set**, with a **21-predictor imputation-sensitivity set**, used in RQ3.

### Important distinction: rows and columns are reduced for different reasons

The master workbook is a **data repository**, not a ready-made model matrix. Its columns include country identifiers, geographic and income classifications, outcomes, raw source measures, policy-survey components, transformed/derived fields, and candidate predictors. Therefore, the study does **not** treat every master-table column as an eligible predictor and then statistically delete approximately 100 variables.

The row/sample reduction and predictor reduction follow separate rules.

### A. Sample funnel

The loaded master panel is first audited in its full form. The primary analytical frame then applies two explicit requirements:

- calendar year must fall within **2000–2021**; and
- **HALE must be observed**, because HALE is the principal study outcome and defines the common primary analytical window.

No complete-case requirement is imposed on the predictors at this stage. This is important because later missing-data diagnostics show that complete-case deletion would disproportionately remove lower-income countries.

The code below prints the counts directly from the loaded dataset rather than hard-coding them. In the current v6 panel this should reproduce approximately **5,208 master country-years across 217 countries/economies** and **4,070 analytical country-years across 185 countries**.

### B. Predictor funnel

The predictor funnel starts with the **26 conceptually eligible candidate indicators** in Section 4. Section 9 then applies the prespecified redundancy/VIF screen. For the **primary RQ3 machine-learning analysis**, Section 19 now applies a stricter missing-data eligibility rule: a retained predictor must have **at least 70% observed data (no more than 30% missing)** in the analytical frame.

This replaces the earlier >55% observed rule as the **primary ML feature-entry criterion**. Predictors that fail the strict 30%-missing rule are excluded from the primary ML feature matrix, even if they remain scientifically important.

A separate **imputation sensitivity analysis** is added in Section 19B. It retains predictors with **more than 55% observed data** (including variables with roughly one-third missingness such as workforce/capacity measures when they meet that rule) and handles their missing values by **median imputation fitted only inside each training/CV fold**. This sensitivity analysis answers a different question: whether retaining these partially observed variables materially changes out-of-country predictive performance.

Thus:
- **Primary ML analysis:** ≤30% missing / ≥70% observed.
- **Sensitivity ML analysis:** >55% observed, with fold-contained median imputation.
- **Outcomes are never imputed.**
- **Explanatory models are unchanged** and continue to use specification-available observations without predictor imputation.

The final primary and sensitivity feature counts are generated dynamically in Section 19 rather than hard-coded here.

### Expected output
A sample-funnel table, a transparent classification of the master-table columns, and a check that the primary and sensitivity RQ3 feature sets are generated consistently with the documented coverage rules.

## **6  Reconstruction of the Policy Index**

### Purpose

Reconstruct the WHO NCD policy measure using a **balanced set of policy components** that is directly comparable across survey waves, and document the reconstruction so that another analyst can reproduce it from the raw policy variables in the master panel.

### Why a balanced index is required

The WHO NCD Country Capacity Survey (NCD-CCS) is repeated periodically, but the questionnaire is not identical in every wave. In the v6 master panel, only four of the policy components represented in the later questionnaire are observed in 2013, whereas twelve policy variables are represented by 2021. Several items were introduced after 2015. Consequently, an index calculated from *all items available in each wave* changes its measurement instrument over time: a score from one wave is not necessarily based on the same policy content as a score from another wave.

This creates a comparability problem. A decline in an all-items index could arise because additional or more demanding questions entered the questionnaire rather than because national NCD policy weakened. For longitudinal analysis, the construct must therefore be held constant across time.

The analytical solution is a **balanced nine-item index**. The index uses only policy components with observed data in **every analytical policy wave from 2015 through 2021 (2015, 2017, 2019, and 2021)**. The 2013 wave is not included in the balanced series because five of the nine common components are absent from that wave in the assembled panel. Three later-introduced variables are also excluded because they are not available across all four balanced waves.

### Reconstruction rule

For each country-wave, the nine harmonized binary policy indicators are coded on a common 0/1 scale and combined only when **all nine required components are observed**:

\[
\text{Policy Index}_{it}=100\times\frac{1}{9}\sum_{j=1}^{9}x_{ijt}
\]

where \(x_{ijt}\in\{0,1\}\) is the harmonized value for policy component \(j\), country \(i\), and survey wave \(t\). The resulting index ranges from **0 to 100**, with higher values representing a larger number of the selected policy/capacity components present.

Two transparent subindices are reconstructed using the same principle:

- **Commitment Index:** five policy, governance, and guideline components.
- **Delivery Index:** four service-delivery, technology, and medicine-availability components.

The master dataset already contains `PolicyIndex`, `CommitmentIndex`, and `DeliveryIndex`. Section 6 does **not** simply accept those derived fields. It rebuilds them from the underlying `POL_*` variables and verifies that the reconstructed values match the stored v6 indices before the variables are used downstream.

### **WHO policy components included in the balanced index**

The variable names below were verified directly in `QM640_Master_Panel_v6.xlsx`. The master panel contains the harmonized dataset variables but not the full questionnaire text. Because WHO wording and numbering can change slightly across survey rounds, the “WHO survey item” column gives concise, standardized wording corresponding to the NCD-CCS questionnaire/indicator definitions rather than implying that every wave used letter-for-letter identical phrasing.

| Domain / subindex | WHO survey item (concise wording) | Dataset variable name | Harmonized coding in v6 | Included in balanced index? |
|---|---|---|---|---|
| Prevention policy / Commitment | National policy to reduce population salt consumption | `POL_SaltPol` | 1 = policy present/implemented; 0 = not present; unavailable response = missing | **Yes** |

**Items deliberately excluded from the balanced index.** The v6 panel also contains `POL_SSBTAX`, `POL_TransFat`, and `POL_FOOD_TAX`. Including them would change the composition of the index over time. They are therefore excluded from the balanced longitudinal measure rather than treated as zeros before their introduction.

### Reproducibility and validation safeguards

The code cell below makes the reconstruction auditable. Before calculating any index it:

1. prints the four included survey years and the number of required items;
2. checks that **all nine required `POL_*` variables exist** and stops with an error if any are missing;
3. verifies that the harmonized observed values are binary (`0/1`);
4. confirms that every required item has observed data in every balanced survey wave;
5. reports item-level missingness percentages by survey wave;
6. reconstructs the overall, commitment, and delivery indices only for country-waves with all nine components observed; and
7. compares the reconstructed indices with the stored v6 indices and stops if a material discrepancy is detected.

These checks prevent the analysis from silently continuing with a renamed variable, an incomplete questionnaire wave, non-binary coding, or a different index definition.

### Source note

Policy-component definitions are mapped to the WHO NCD Country Capacity Survey documentation and questionnaire materials. WHO survey documentation also shows that questionnaire content has been revised across rounds, which is the reason a fixed-item longitudinal measure is required. Dataset variable names and wave availability are taken directly from the v6 master panel used in this capstone.

### Expected output

A component registry, the balanced waves and item count, a wave-by-wave missingness table, confirmation that all required variables and coding rules pass validation, a comparison of the reconstructed and stored indices, and the final size of the balanced policy panel.


```python
# ---------------------------------------------------------------------------
# SECTION 6 — Reconstruct and validate the balanced WHO NCD policy index
# ---------------------------------------------------------------------------

BALANCED_POLICY_YEARS = [2015, 2017, 2019, 2021]

# Five commitment/governance components and four delivery/capacity components.
COMMITMENT_POLICY_ITEMS = [
    "POL_SaltPol",
    "POL_Mkting",
    "POL_MultisecComm",
    "POL_targets",
    "POL_NCDGuide",
]

DELIVERY_POLICY_ITEMS = [
    "POL_riskstrat",
    "POL_CholMsmt",
    "POL_Statins",
    "POL_Insulin",
]

BALANCED_POLICY_ITEMS = COMMITMENT_POLICY_ITEMS + DELIVERY_POLICY_ITEMS

# Variables present in later questionnaires but deliberately excluded because
# they are not observed across every balanced wave.
LATER_POLICY_ITEMS_EXCLUDED = [
    "POL_SSBTAX",     # available in the assembled panel from 2017
    "POL_TransFat",   # available from 2019
    "POL_FOOD_TAX",   # available from 2021
]

print("Balanced policy-index years :", BALANCED_POLICY_YEARS)
print("Required balanced items      :", len(BALANCED_POLICY_ITEMS))
print("Commitment items             :", len(COMMITMENT_POLICY_ITEMS))
print("Delivery items               :", len(DELIVERY_POLICY_ITEMS))

# ---- 1. Required-column validation -----------------------------------------
missing_required = [v for v in BALANCED_POLICY_ITEMS if v not in df.columns]
if missing_required:
    raise KeyError(
        "Balanced policy index cannot be reconstructed. "
        f"Required policy variable(s) missing from the dataset: {missing_required}"
    )

print("All required policy variables present: YES")

# ---- 2. Restrict validation to the four comparable survey waves ------------
policy_wave = df[df["Year"].isin(BALANCED_POLICY_YEARS)].copy()

observed_years = sorted(policy_wave["Year"].dropna().astype(int).unique().tolist())
if observed_years != BALANCED_POLICY_YEARS:
    raise ValueError(
        "Balanced policy waves in the dataset do not match the expected set. "
        f"Expected {BALANCED_POLICY_YEARS}; found {observed_years}."
    )

# ---- 3. Verify harmonized 0/1 coding ----------------------------------------
bad_codes = {}
for var in BALANCED_POLICY_ITEMS:
    vals = set(policy_wave[var].dropna().unique().tolist())
    invalid = sorted(v for v in vals if v not in {0, 1, 0.0, 1.0})
    if invalid:
        bad_codes[var] = invalid

if bad_codes:
    raise ValueError(
        "Non-binary values detected in harmonized policy variables: "
        f"{bad_codes}. Expected only 0/1 or missing."
    )

print("Binary coding validation      : PASSED")

# ---- 4. Verify each item is represented in every balanced wave -------------
wave_sizes = policy_wave.groupby("Year").size()
availability_n = policy_wave.groupby("Year")[BALANCED_POLICY_ITEMS].count()

zero_availability = [
    (int(year), var)
    for year, row in availability_n.iterrows()
    for var in BALANCED_POLICY_ITEMS
    if int(row[var]) == 0
]
if zero_availability:
    raise ValueError(
        "At least one required policy item has no observations in a balanced wave: "
        f"{zero_availability}. The index would not be comparable across waves."
    )

print("Cross-wave availability check : PASSED")

# ---- 5. Report missingness transparently -----------------------------------
missing_pct = (
    100 * (1 - availability_n.div(wave_sizes, axis=0))
).round(1)

print("\nMissingness (%) for each required policy item by survey wave:")
display(missing_pct)

overall_missing = (
    policy_wave[BALANCED_POLICY_ITEMS].isna().mean().mul(100).round(1)
    .rename("Missing % (2015–2021 policy waves)")
    .to_frame()
)
display(overall_missing)

# ---- 6. Reconstruct indices only when all nine items are observed ----------
complete9 = policy_wave[BALANCED_POLICY_ITEMS].notna().all(axis=1)
rebuilt = policy_wave.loc[complete9, ["iso3", "Year"] + BALANCED_POLICY_ITEMS].copy()

rebuilt["PolicyIndex_rebuilt"] = (
    rebuilt[BALANCED_POLICY_ITEMS].mean(axis=1) * 100
)
rebuilt["CommitmentIndex_rebuilt"] = (
    rebuilt[COMMITMENT_POLICY_ITEMS].mean(axis=1) * 100
)
rebuilt["DeliveryIndex_rebuilt"] = (
    rebuilt[DELIVERY_POLICY_ITEMS].mean(axis=1) * 100
)

print(
    f"\nComplete 9-item country-waves : {len(rebuilt):,}\n"
    f"Countries represented         : {rebuilt['iso3'].nunique():,}\n"
    f"Included years                : "
    f"{sorted(rebuilt['Year'].astype(int).unique().tolist())}"
)

# ---- 7. Verify reconstruction against the stored v6 derived indices --------
stored_to_rebuilt = {
    "PolicyIndex": "PolicyIndex_rebuilt",
    "CommitmentIndex": "CommitmentIndex_rebuilt",
    "DeliveryIndex": "DeliveryIndex_rebuilt",
}

for stored, calc in stored_to_rebuilt.items():
    if stored not in df.columns:
        raise KeyError(
            f"Stored validation field '{stored}' is missing. "
            "The raw-item reconstruction exists, but the v6 consistency check cannot run."
        )

    stored_vals = policy_wave.loc[rebuilt.index, stored]
    overlap = stored_vals.notna()

    if not overlap.any():
        raise ValueError(
            f"No stored values are available for '{stored}' on the reconstructed rows."
        )

    max_abs_diff = (
        stored_vals.loc[overlap] - rebuilt.loc[overlap, calc]
    ).abs().max()

    print(
        f"{stored:<18} vs rebuilt: "
        f"n={int(overlap.sum()):,}, max |difference|={max_abs_diff:.10f}"
    )

    if max_abs_diff > 1e-8:
        raise AssertionError(
            f"{stored} does not reproduce from the documented component list. "
            f"Maximum absolute difference = {max_abs_diff:.10f}"
        )

print("Stored-index reproduction     : PASSED")

# ---- 8. Use the transparently reconstructed values downstream --------------
# Clear the balanced-wave derived values first so incomplete 9-item rows cannot
# accidentally retain a stale precomputed index.
balanced_mask = df["Year"].isin(BALANCED_POLICY_YEARS)
for col in ["PolicyIndex", "CommitmentIndex", "DeliveryIndex"]:
    df.loc[balanced_mask, col] = float("nan")

df.loc[rebuilt.index, "PolicyIndex"] = rebuilt["PolicyIndex_rebuilt"]
df.loc[rebuilt.index, "CommitmentIndex"] = rebuilt["CommitmentIndex_rebuilt"]
df.loc[rebuilt.index, "DeliveryIndex"] = rebuilt["DeliveryIndex_rebuilt"]

POLICY_PANEL = df.dropna(subset=["PolicyIndex"]).copy()

print(
    f"\nBalanced policy panel: {len(POLICY_PANEL):,} country-waves, "
    f"{POLICY_PANEL['iso3'].nunique():,} countries, "
    f"waves {sorted(POLICY_PANEL['Year'].astype(int).unique().tolist())}"
)

# ---- 9. Retain the original diagnostic comparison --------------------------
if "PolicyIndex_allitems_DO_NOT_USE" in df.columns:
    a = (
        df.dropna(subset=["PolicyIndex_allitems_DO_NOT_USE"])
          .groupby("Year")["PolicyIndex_allitems_DO_NOT_USE"]
          .agg(["count", "mean"])
    )
    b = (
        df.dropna(subset=["PolicyIndex"])
          .groupby("Year")["PolicyIndex"]
          .agg(["count", "mean"])
    )

    cmp = pd.concat(
        [
            a.rename(columns={"count": "n (all items)", "mean": "All-items index"}),
            b.rename(columns={"count": "n (balanced)", "mean": "Balanced 9-item index"}),
        ],
        axis=1,
    )
    display(cmp.round(1))

    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.plot(a.index, a["mean"], "o--", c="#b2182b", label="All items (not comparable)")
    ax.plot(b.index, b["mean"], "s-", c="#1b7837", lw=2.5, label="Balanced 9 items (used)")
    ax.set_ylabel("Mean policy index (0–100)")
    ax.set_xlabel("Survey wave")
    ax.set_title("Why the policy index was rebuilt", fontweight="bold")
    ax.legend()
    savefig("06_policy_index_rebuild")
```

## **7  Outliers — flagged, not deleted**

### Purpose
Identify extreme values and decide, explicitly, whether to remove them.

### Rationale
The extreme values in this panel are real countries, not typing errors. Luxembourg genuinely does spend more per head than anywhere else, and Lesotho's life expectancy genuinely did collapse during the HIV epidemic. Deleting them would remove precisely the variation the study exists to explain, and would quietly narrow the sample to a convenience set of median countries.

### Method
Interquartile-range flagging at a conservative multiplier, reported per variable with the most extreme cases named so a reader can judge whether each is plausible. No value is deleted; the flags are retained so that every model can be re-run without them as a robustness check.

### Expected output
A table of flagged variables with named extreme cases, and the share of rows carrying at least one flag.

```python
def iqr_flags(s, k=3.0):
    q1, q3 = s.quantile([.25,.75]); i = q3-q1
    return (s < q1-k*i) | (s > q3+k*i)

NUM = [c for c in ANALYSIS_COLS if pd.api.types.is_numeric_dtype(df[c])]
orows = []
for c in NUM:
    s = work[c].dropna()
    if len(s) < 50: continue
    f = iqr_flags(s)
    if f.sum():
        ex = work.loc[s[f].sort_values(ascending=False).index[:3], ["Country","Year",c]]
        orows.append({"Variable":c, "Outliers":int(f.sum()), "% of obs":round(f.mean()*100,1),
                      "Most extreme":", ".join(f"{r.Country} {int(r.Year)}" for r in ex.itertuples())})
OUTL = pd.DataFrame(orows).sort_values("Outliers", ascending=False)
display(OUTL.head(15))
print(f"\n{OUTL.Outliers.sum():,} flagged values across {len(OUTL)} variables. None deleted.")
work["_any_outlier"] = False
for c in NUM:
    if work[c].notna().sum() >= 50:
        work["_any_outlier"] |= iqr_flags(work[c]).fillna(False)
print(f"{work._any_outlier.sum():,} of {len(work):,} rows carry at least one flag "
      f"({work._any_outlier.mean()*100:.1f}%).")
```

### Result
Flags concentrate in the scale variables — expenditure, income, population density — where a right-skewed distribution guarantees them regardless of data quality.

### Interpretation
The flags behave exactly as expected and identify real economic dispersion rather than error. Removing Qatar for being wealthy or Afghanistan for being poor would destroy the between-country variation that Sections 12 to 14 depend on entirely.

### Decision
Retain all extremes. Suppress only the impossible values — the country-years where HALE exceeds life expectancy — which are a source conflict rather than an outlier.

## **8  Feature engineering**

### Purpose
Construct logs, lags, the between/within Mundlak split, and the derived morbidity-gap outcome.

### Rationale
Four transformations are used, each for a stated analytical reason rather than by convention. Logs are applied where a variable is right-skewed and its mechanism is proportional. Lags reflect that health-system changes may not affect population outcomes within the same calendar year. The Mundlak split separates a country's long-run level from its year-to-year deviation, making explicit whether information is primarily **between countries** or **within countries over time**. The morbidity gap captures the distinction between years lived and years lived in good health that is central to RQ4.

The between-versus-within variance decomposition is reported **as descriptive context only** to show where the information in each variable comes from. It is **not** used as a decision rule for including or excluding predictors. Variable selection is determined by the conceptual framework, the prespecified analytical blocks, and the research questions; the subsequent regression models then test the relevant associations under their stated specifications.

### Method
Log transforms are applied to skewed monetary and density variables; three-year lags are constructed for actionable levers; a country mean and deviation-from-country-mean pair is generated for each lever, risk factor, and control; and the morbidity gap is defined as life expectancy minus HALE, with impossible negative values set to missing. The variance decomposition table is retained to document whether the observed variation is predominantly between countries or within countries over time, not to screen variables out of later analyses.

> **Footnote.** The two-way fixed-effects regressions used later in the notebook identify coefficients from **within-country variation over time only**, after controlling for country fixed effects and common year effects. The descriptive between-versus-within variance table does not alter which variables enter those regressions.

### Expected output
Counts of each transformation type and a descriptive variance-decomposition table showing the share of observed variation that occurs within versus between countries.


```python
eng = work.sort_values(["iso3","Year"]).copy()

# (a) logs where skew is severe and the mechanism is proportional
for raw, new in [("HealthExpPC","logHealthExpPC"), ("GDPPerCapita","logGDPPerCapita"),
                 ("GovHealthExpPC","logGovHealthExpPC"), ("PopulationDensity","logPopDensity_ENG")]:
    if raw in eng.columns and new not in eng.columns:
        eng[new] = np.log1p(eng[raw].clip(lower=0))

# (b) lags — systems respond with delay
LAGGABLE = [v for v in ACTIONABLE if v in eng.columns]
for v in LAGGABLE:
    eng[f"{v}_lag3"] = eng.groupby("iso3")[v].shift(3)

# (c) Mundlak / correlated-random-effects split
#     _btw = a country's long-run average (cross-country comparison)
#     _wth = this year's deviation from that average (within-country movement)
MUNDLAK = [v for v in ACTIONABLE + RISK + CONTEXT if v in eng.columns]
for v in MUNDLAK:
    m = eng.groupby("iso3")[v].transform("mean")
    eng[f"{v}_btw"] = m
    eng[f"{v}_wth"] = eng[v] - m

# (d) the derived outcome
if {"LifeExpectancy","HALE"} <= set(eng.columns):
    eng["MorbidityGap"] = eng.LifeExpectancy - eng.HALE
    eng.loc[eng.MorbidityGap < 0, "MorbidityGap"] = np.nan   # source conflicts, not data

print(f"engineered frame: {eng.shape[0]:,} x {eng.shape[1]}")
print(f"  logs      {sum(c.startswith('log') for c in eng.columns)}")
print(f"  lags      {sum(c.endswith('_lag3') for c in eng.columns)}")
print(f"  Mundlak   {sum(c.endswith(('_btw','_wth')) for c in eng.columns)}")
```

```python
vd = []
for v in [x for x in ACTIONABLE + RISK + CONTEXT if x in eng.columns]:
    s = eng[["iso3", v]].dropna()
    if s.iso3.nunique() < 30: continue
    tot = s[v].var()
    wth = s.groupby("iso3")[v].transform(lambda z: z - z.mean()).var()
    vd.append({"Variable":v, "Within-country share of variance %": round(wth/tot*100,1),
               "Between-country share %": round((1-wth/tot)*100,1),
               "FE can see it": "yes" if wth/tot > .25 else ("partly" if wth/tot > .10 else "barely")})
VD = pd.DataFrame(vd).sort_values("Within-country share of variance %")
display(VD)

fig, ax = plt.subplots(figsize=(9, max(4, .32*len(VD))))
ax.barh(VD.Variable, VD["Within-country share of variance %"],
        color=np.where(VD["Within-country share of variance %"] > 25, "#1b7837",
               np.where(VD["Within-country share of variance %"] > 10, "#f4a582", "#b2182b")))
ax.axvline(25, ls="--", c="k", lw=1); ax.set_xlabel("% of variance that is within-country")
ax.set_title("What fixed effects can and cannot see", fontweight="bold")
savefig("08_variance_decomposition")
print("\nRed bars are variables a fixed-effects model is structurally near-blind to.")
print("For those, the between-country term recovered by the Mundlak split carries the signal.")
```

### Result
The variance decomposition shows that some predictors vary predominantly between countries, whereas others contain more within-country movement over time.

### Interpretation
This table is **contextual rather than selective**. It helps explain where the empirical information in each variable comes from and therefore why a within-country estimator may be more informative for some predictors than for others. A variable with little within-country movement provides limited information for a two-way fixed-effects coefficient, but that fact is not evidence that the variable is conceptually unimportant and is not a basis for removing it from the analysis.

### Decision
Retain variables according to the **conceptual framework and research questions**, subject only to the documented data-availability and specification requirements elsewhere in the notebook. Use the variance decomposition to interpret the information available to different estimators, while allowing the subsequent regressions to test the prespecified associations.

## **9  Multicollinearity screen**

### Purpose
Detect exact redundancy and inflated variance among the candidate predictors.

### Rationale
Two distinct problems are caught here. The first is exact redundancy — variables that are algebraic restatements of one another, such as a government share expressed once as a percentage and once as a fraction, or a density that is the difference of two logs already in the model. These make the design matrix singular and produce coefficients in the tens of thousands. The second is ordinary high correlation, which inflates standard errors without breaking anything visibly. An earlier iteration of this study read a variance inflation factor in the tens of millions from exactly the first problem.

### Method
Pairwise correlation screening at a near-unity threshold to catch identities first, then iterative variance-inflation screening. Variables that a hypothesis explicitly requires are protected from removal and reported with their inflation factor rather than dropped, and a floor prevents the loop from stripping the set down to the protected variables alone.

### Expected output
An identity report, a VIF table with protected variables marked, and a correlation heatmap of the retained set.

```python
SCREEN = [v for v in ACTIONABLE + RISK + CONTEXT + GOVERNANCE if v in eng.columns]
Xs = eng[SCREEN].dropna(thresh=int(.5*len(SCREEN)), axis=0)
Xs = Xs.fillna(Xs.median())

# (a) exact identities first — |r| > 0.999
ABSCORR = Xs.corr().abs()
np.fill_diagonal(ABSCORR.values, 0)
ident = [(a,b,round(ABSCORR.loc[a,b],5)) for a in ABSCORR.index for b in ABSCORR.columns
         if a < b and ABSCORR.loc[a,b] > .999]
if ident:
    print("EXACT REDUNDANCIES (one of each pair must be dropped):")
    for a,b,r in ident: print(f"  {a}  ==  {b}   r = {r}")
    drop_exact = {b for _,b,_ in ident}
else:
    print("No exact redundancies detected.")
    drop_exact = set()

# (b) iterative VIF, with protected variables and a floor
PROTECTED = set(REACH + CAPACITY + POLICY + ["logHealthExpPC","HealthExpGDP","Pop65Plus","logGDPPerCapita"])
DERIVED_SUFFIX = ("_lag3","_btw","_wth")
keep = [v for v in SCREEN if v not in drop_exact and not v.endswith(DERIVED_SUFFIX)]
MIN_RETAINED, VIF_MAX = 8, 10.0
removed = []
while len(keep) > MIN_RETAINED:
    Z = sm.add_constant(Xs[keep])
    vifs = pd.Series([variance_inflation_factor(Z.values, i) for i in range(1, Z.shape[1])], index=keep)
    cand = vifs.drop(labels=[v for v in keep if v in PROTECTED], errors="ignore")
    if cand.empty or cand.max() <= VIF_MAX: break
    worst = cand.idxmax(); removed.append((worst, round(cand.max(),2))); keep.remove(worst)

Zf = sm.add_constant(Xs[keep])
VIFT = pd.DataFrame({"Variable": keep,
                     "VIF": [variance_inflation_factor(Zf.values,i) for i in range(1,Zf.shape[1])]})
VIFT["Status"] = np.where(VIFT.Variable.isin(PROTECTED), "protected (kept by hypothesis)", "screened")
VIFT["Verdict"] = pd.cut(VIFT.VIF, [0,5,10,np.inf], labels=["fine","watch","high"])
display(VIFT.sort_values("VIF", ascending=False))
if removed:
    print("\nRemoved by VIF screen:")
    for v, s in removed: print(f"  {v:<28} VIF {s}")
print(f"\n{len(keep)} variables retained. Max VIF among unprotected: "
      f"{VIFT[VIFT.Status=='screened'].VIF.max():.2f}")
RETAINED = keep
```

```python
fig, ax = plt.subplots(figsize=(11,9))
cm = eng[[v for v in RETAINED if v in eng.columns]].corr()
sns.heatmap(cm, cmap="RdBu_r", center=0, vmin=-1, vmax=1, annot=False, square=True,
            cbar_kws={"label":"Pearson r"}, ax=ax)
ax.set_title("Correlation among retained predictors", fontweight="bold")
savefig("09_correlation_matrix")
hi = [(a,b,round(cm.loc[a,b],3)) for a in cm.index for b in cm.columns
      if a < b and abs(cm.loc[a,b]) > .7]
print(f"{len(hi)} pairs above |r| = 0.7:")
for a,b,r in sorted(hi, key=lambda t:-abs(t[2]))[:12]: print(f"  {a:<26} {b:<26} {r:+.3f}")
```

### Result
Exact identities are caught and removed, and the retained set clears the conventional inflation threshold among unprotected variables.

### Interpretation
Protecting hypothesis-mandated variables from automatic removal is deliberate. A screen that silently deletes the variable a research question is about will always produce a clean-looking model and an unanswerable question. Where a protected variable carries a high inflation factor, that is disclosed and its standard error is read accordingly.

### Decision
Carry the retained set into the predictive models. Inferential models use the specification each hypothesis requires, not the screened set.

## **10  Sample adequacy and minimum detectable effect**

### Purpose
Establish how large an association would have to be before this design could detect it.

### Rationale
The conventional answers to the sample-size question are wrong for panel data. Green's rule and the ten-observations-per-predictor heuristic both assume independent observations, and forty-eight annual observations from Norway are not forty-eight independent facts about health systems. Because standard errors are clustered by country, precision is governed by the number of **countries**, not the number of rows. Both conventional rules are reported and then explicitly set aside.

### Method
Cluster counts are checked against the thresholds that actually bind for cluster-robust inference and for the wild bootstrap. The minimum detectable effect is then computed per predictor as the effect size reachable at eighty per cent power given the realised clustered standard error and the available degrees of freedom.

### Expected output
An adequacy table with the binding constraint identified, and per-predictor minimum detectable effects for both outcomes.

```python
k = len(RETAINED)
N_ROWS = len(eng.dropna(subset=["HALE"]))
N_CL   = eng.dropna(subset=["HALE"]).iso3.nunique()

ADEQ = pd.DataFrame([
 ["Green (1991): 50 + 8k", 50+8*k, N_ROWS, "PASS", "Assumes independent rows — not applicable here"],
 ["10 observations per predictor", 10*k, N_ROWS, "PASS", "Same assumption — not applicable"],
 ["Clusters for cluster-robust SEs", "40–50", N_CL, "PASS" if N_CL>=50 else "REVIEW",
  "THIS is the binding constraint"],
 ["Clusters for wild bootstrap", "≥30", N_CL, "PASS" if N_CL>=30 else "REVIEW",
  "Bootstrap-t used where clusters are fewer"],
], columns=["Criterion","Required","Available","Verdict","Note"])
display(ADEQ)
```

```python
def mde_table(d, outcome, preds, alpha=.05, power=.80):
    rows = []
    for v in preds:
        s = d.dropna(subset=[outcome, v])
        G = s.iso3.nunique()
        if G < 20 or len(s) < 100: continue
        f = smf.ols(f"{outcome} ~ {v} + C(Year)", data=s)\
               .fit(cov_type="cluster", cov_kwds={"groups": s.iso3})
        se = f.bse[v]; sd = s[v].std()
        tc = stats.t.ppf(1-alpha/2, G-1); tp = stats.t.ppf(power, G-1)
        rows.append({"Predictor":v, "Clusters":G, "n":len(s),
                     "SE":se, "MDE (outcome units)": (tc+tp)*se,
                     "MDE per 1 SD of predictor": (tc+tp)*se*sd,
                     "Observed |b| per 1 SD": abs(f.params[v])*sd})
    t = pd.DataFrame(rows)
    t["Detectable?"] = np.where(t["Observed |b| per 1 SD"] >= t["MDE per 1 SD of predictor"],
                                "above MDE", "below MDE")
    return t.sort_values("MDE per 1 SD of predictor")

MDE_H = cached("mde_hale", lambda: mde_table(eng, "HALE",
        [v for v in ACTIONABLE if v in eng.columns]))
MDE_N = cached("mde_ncd", lambda: mde_table(eng, "NCDMortality",
        [v for v in ACTIONABLE if v in eng.columns]))
banner("MINIMUM DETECTABLE EFFECT — HALE (years)")
display(MDE_H.round(4))
banner("MINIMUM DETECTABLE EFFECT — NCD mortality (percentage points)")
display(MDE_N.round(4))
print(f"\nTypical MDE, HALE: {MDE_H['MDE per 1 SD of predictor'].median():.2f} years per SD")
print(f"Typical MDE, NCD : {MDE_N['MDE per 1 SD of predictor'].median():.2f} pp per SD")
print("Null results below these thresholds are inconclusive, not negative.")
```

### Result
The cluster count clears every threshold comfortably, and the minimum detectable effect is established for each lever against both outcomes.

### Interpretation
This is the honest version of a power calculation for a clustered design, and it changes how the nulls later in the notebook should be read. A coefficient smaller than the minimum detectable effect means the study could not have seen it, not that it is zero. Reporting such a result as evidence of no effect would overstate what the design supports.

### Decision
Report every null result against its minimum detectable effect rather than against zero.

### **11.1  Numeric-variable descriptive statistics and skewness diagnostic**

#### Purpose
Provide a reproducible distributional diagnostic for all numeric variables in the analytical data frame before modelling, and document whether a logarithmic transformation would improve symmetry.

#### Method
For each numeric variable, the table reports the mean, median, standard deviation, and sample skewness. Skewness is interpreted using the following prespecified rule: values between **−0.5 and 0.5** are classified as *roughly symmetric, no log*; values greater than **0.5 and up to 1.0** are classified as *moderately skewed, consider log*; and values **above 1.0** are classified as *highly skewed, log recommended*. Negative skewness below −0.5 is reported separately because a conventional natural-log transformation does not address left skew in the same way.

A second diagnostic is calculated for numeric variables whose observed values are strictly positive. Original skewness is compared with skewness after the natural-log transformation, and improvement is defined as a reduction in the **absolute magnitude of skewness**.

> **Transformation note.** The diagnostic flag is evidence for evaluating a transformation, not an automatic modelling rule. Log transformations are retained only when they reduce the absolute skewness and are substantively appropriate. Indices, percentages/proportions, binary indicators, calendar years, and other naturally bounded variables are not logged solely because of a skewness threshold. Existing transformations used elsewhere in the notebook remain unchanged.

#### Expected output
Two publication-ready tables: (1) descriptive statistics and skewness classifications for all numeric variables, and (2) original-versus-log skewness for all strictly positive numeric variables, including an explicit indicator of whether the log transformation improves symmetry.


> **Scope note.** The descriptive-statistics table is restricted to the study outcomes and the 26 conceptually eligible candidate indicators. Helper flags, audit fields, raw policy components, lagged variables, Mundlak between/within terms, carried values, and other intermediate engineering fields are intentionally excluded. The log-comparison table is restricted to transformations actually constructed in Section 8.


```python
# ---------------------------------------------------------------------------
# SECTION 11.1 — Descriptive statistics and skewness diagnostics
# Restricted to variables that are in scope for the study.
# ---------------------------------------------------------------------------
from scipy.stats import skew as scipy_skew

# Table 11.1A is intentionally restricted to:
#   (1) the study outcomes, and
#   (2) the 26 conceptually eligible candidate indicators from Section 4.
# Helper flags, audit fields, policy ingredients, lagged variables, Mundlak
# components, carried values, and other intermediate engineering fields are
# excluded from this descriptive table.
EDA_SCOPE_VARS = list(dict.fromkeys(
    [v for v in OUTCOMES.keys() if v in eng.columns] +
    [v for v in CANDIDATE_INDICATORS if v in eng.columns]
))

eda_numeric = eng[EDA_SCOPE_VARS].select_dtypes(include=[np.number]).copy()

def _sample_skew(series):
    """Bias-corrected sample skewness, omitting missing values."""
    x = pd.to_numeric(series, errors="coerce").dropna().to_numpy()
    if len(x) < 3 or np.nanstd(x, ddof=1) == 0:
        return np.nan
    return float(scipy_skew(x, bias=False, nan_policy="omit"))

def _skew_flag(sk):
    if pd.isna(sk):
        return "insufficient/constant"
    if -0.5 <= sk <= 0.5:
        return "roughly symmetric, no log"
    if 0.5 < sk <= 1.0:
        return "moderately skewed, consider log"
    if sk > 1.0:
        return "highly skewed, log recommended"
    return "negatively skewed, log not indicated"

# ---- Table 11.1A: analytical outcomes + candidate indicators ---------------
desc_rows = []
for var in eda_numeric.columns:
    s = pd.to_numeric(eda_numeric[var], errors="coerce").dropna()
    sk = _sample_skew(s)
    desc_rows.append({
        "Variable": var,
        "Mean": s.mean() if len(s) else np.nan,
        "Median": s.median() if len(s) else np.nan,
        "Standard deviation": s.std(ddof=1) if len(s) > 1 else np.nan,
        "Skewness": sk,
        "Suggested log-transformation flag": _skew_flag(sk),
    })

EDA_SKEWNESS = pd.DataFrame(desc_rows)

print("Table 11.1A. Descriptive statistics and skewness for study outcomes and 26 candidate indicators")
display(
    EDA_SKEWNESS.style
        .format({
            "Mean": "{:,.3f}",
            "Median": "{:,.3f}",
            "Standard deviation": "{:,.3f}",
            "Skewness": "{:,.3f}",
        }, na_rep="—")
        .hide(axis="index")
)

print(f"\nVariables shown in Table 11.1A: {len(EDA_SKEWNESS)}")
print("Helper, audit, lagged, between/within, carried, and raw policy-component fields are excluded.")

# ---- Table 11.1B: ONLY transformations actually constructed in Section 8 ---
# This is not a scan of every positive variable. It audits only the raw
# variables for which a log representation is actually constructed.
LOG_TRANSFORM_PAIRS = [
    ("HealthExpPC", "logHealthExpPC"),
    ("GDPPerCapita", "logGDPPerCapita"),
    ("GovHealthExpPC", "logGovHealthExpPC"),
    ("PopulationDensity", "logPopDensity_ENG"),
]

log_rows = []
for raw, logged in LOG_TRANSFORM_PAIRS:
    if raw not in eng.columns or logged not in eng.columns:
        continue

    s_raw = pd.to_numeric(eng[raw], errors="coerce").dropna()
    s_log = pd.to_numeric(eng.loc[s_raw.index, logged], errors="coerce").dropna()

    common = s_raw.index.intersection(s_log.index)
    s_raw = s_raw.loc[common]
    s_log = s_log.loc[common]

    original_skew = _sample_skew(s_raw)
    log_skew = _sample_skew(s_log)

    log_rows.append({
        "Raw variable": raw,
        "Analytical representation": logged,
        "Original skewness": original_skew,
        "Log skewness": log_skew,
        "Absolute skewness reduction": (
            abs(original_skew) - abs(log_skew)
            if pd.notna(original_skew) and pd.notna(log_skew)
            else np.nan
        ),
        "Log improved skewness?": (
            "Yes" if pd.notna(original_skew) and pd.notna(log_skew)
            and abs(log_skew) < abs(original_skew) else "No"
        ),
    })

EDA_LOG_COMPARISON = pd.DataFrame(log_rows)

print("\nTable 11.1B. Skewness check for log transformations actually constructed in Section 8")
display(
    EDA_LOG_COMPARISON.style
        .format({
            "Original skewness": "{:.3f}",
            "Log skewness": "{:.3f}",
            "Absolute skewness reduction": "{:.3f}",
        }, na_rep="—")
        .hide(axis="index")
)

print(
    "\nInterpretation: Table 11.1B is a transformation audit, not a feature-selection "
    "screen. Only log transformations actually constructed in Section 8 are shown."
)

assert set(EDA_SKEWNESS["Variable"]).issubset(set(EDA_SCOPE_VARS))
assert len(EDA_LOG_COMPARISON) <= len(LOG_TRANSFORM_PAIRS)
```

#### Interpretation

The first table is a descriptive screening device: it identifies variables whose empirical distributions warrant closer inspection. The second table provides the corresponding transformation diagnostic for strictly positive variables by showing whether a natural log moves the distribution closer to symmetry. A reduction in absolute skewness supports a log transformation where the variable's scale and substantive interpretation also make that transformation appropriate.

The skewness thresholds are therefore **diagnostic rather than mechanical selection criteria**. In particular, an index or bounded percentage may be skewed without being an appropriate candidate for logging. Conversely, a positive monetary or density measure may be logged when the transformation materially reduces skewness and is consistent with the proportional interpretation used in the analytical model.

**Decision.** Retain logs only where they improve distributional symmetry and are substantively defensible. Do not log indices or bounded variables solely because they cross a skewness threshold. This diagnostic documents the transformation rationale; it does not alter any previously specified models or reported results.


### Result
The outcomes are well behaved, the income gradient is large and persistent, and the return to spending flattens sharply at higher levels.

### Interpretation
That flattening is the substantive point. The relationship between money and healthy years is strongly concave, which means the same additional dollar buys very different amounts of health depending on where a country starts. A model linear in levels would misstate this badly; a model linear in logs captures it, and the residual from that model is what Part II calls efficiency.

### Decision
Use log spending throughout. Proceed to formalise the concave relationship as a production frontier.

# Part II — Theme 1 / RQ1: Health-System Conversion Efficiency


## **12  Building the production frontier**

### Purpose
Estimate expected healthy life expectancy conditional on national resources and demographic structure, quantify the residual as health-system conversion efficiency, and use the resulting benchmark to position Saudi Arabia within the global country distribution.

### Rationale
A health authority that is told 'spend more' has not been told anything it can act on, because the budget is usually the constraint rather than the choice. The more useful question is whether a country is getting as much health from its current budget as comparable countries do. Fitting an expected outcome from resources and demography alone, then reading the gap, converts a vague notion of performance into a measured quantity in years.

### Method
Ordinary least squares on country averages for the most recent five years, so a single transient year cannot define a country's position. The frontier holds only resource levels and demographic structure — spending per head, national income, ageing, urbanisation, fertility. Every health-system decision is deliberately held out, so that those decisions remain available in Section 14 as candidate explanations for the residual. Heteroscedasticity-robust standard errors throughout.

### Expected output
A fitted resource-demography benchmark, an efficiency score per country, the size of the residual relative to the design's detectable effect, leader and laggard tables, and the Saudi Arabia benchmarking output generated from the same global model.

```python
# The frontier holds LEVELS of resources and demographic structure only. Financing-mix
# variables — the health share of GDP, the government share, out-of-pocket reliance — are
# deliberately excluded, because those are allocation CHOICES. Putting a variable inside the
# frontier makes the residual orthogonal to it by construction, which would turn Section 14
# into a test that can only ever return zero.
FRONT_IN = [v for v in ["logHealthExpPC","logGDPPerCapita","Pop65Plus",
                        "UrbanPopulation","FertilityRate"]
            if v in eng.columns]
FINANCING_MIX = [v for v in ["HealthExpGDP","GovHealthExpShare","OOPExpenditure",
                             "ExternalHealthExpShare"] if v in eng.columns]
print(f"Frontier inputs      : {FRONT_IN}")
print(f"Held out for Sec. 14 : {FINANCING_MIX}")

fr = (eng[eng.Year.between(2017,2021)]
        .groupby(["iso3","Country","Region","IncomeGroup"], as_index=False)[["HALE"]+FRONT_IN]
        .mean())
fr = fr.dropna(subset=["HALE"] + FRONT_IN)
print(f"Frontier sample: {len(fr)} countries, 2017–2021 averages")

Xf = sm.add_constant(fr[FRONT_IN])
frontier = sm.OLS(fr.HALE, Xf).fit(cov_type="HC3")
print(frontier.summary().tables[1])

fr["Predicted"] = frontier.fittedvalues
fr["Efficiency"] = frontier.resid
R2 = frontier.rsquared

banner("H1a — DECOMPOSITION")
print(f"Variance in HALE explained by spending + demography : {R2*100:5.1f}%")
print(f"Variance remaining as efficiency                    : {(1-R2)*100:5.1f}%")
print(f"SD of the efficiency component                      : {fr.Efficiency.std():5.2f} years")
print(f"Spread, best to worst                               : {fr.Efficiency.max()-fr.Efficiency.min():5.1f} years")
mde_ref = MDE_H["MDE per 1 SD of predictor"].median()
print(f"Design MDE for reference                            : {mde_ref:5.2f} years")
H1a = (1-R2) > .15 and fr.Efficiency.std() > mde_ref
print(f"\nH1a₀ {'REJECTED' if H1a else 'not rejected'} — the efficiency component is "
      f"{'far larger than' if H1a else 'not clearly larger than'} what this design could mistake for noise.")
```

### Model Specification and Equation

**Model:** OLS resource-demography HALE benchmark with HC3 heteroscedasticity-robust standard errors.

**Purpose:** Estimate expected HALE from national resource and demographic conditions, then define the country-level residual as health-system conversion efficiency. This is an OLS benchmarking regression, not a stochastic-frontier estimator.

**Outcome (Y):** `HALE`, averaged by country over 2017–2021.

**Predictors (X):** `logHealthExpPC`, `logGDPPerCapita`, `Pop65Plus`, `UrbanPopulation`, and `FertilityRate`.

**General specification:**

\[
HALE_i=\beta_0+\beta_1\log(HealthExpPC_i)+\beta_2\log(GDPPC_i)+
\beta_3Pop65_i+\beta_4Urban_i+\beta_5Fertility_i+\varepsilon_i
\]

**Final fitted equation from the executed notebook:**

\[
\widehat{HALE}_i=
50.9280
-0.3667\,logHealthExpPC_i
+1.8762\,logGDPPerCapita_i
+0.1515\,Pop65Plus_i
+0.0214\,UrbanPopulation_i
-1.8585\,FertilityRate_i
\]

The intercept (50.9280) is the fitted HALE when all predictors equal zero on their coded scales; it is mainly a regression constant rather than a substantively realistic country profile. Each slope is the expected difference in HALE associated with a one-unit difference in that predictor, conditional on the other benchmark variables.

**Benchmark calculation:**

\[
Conversion\ Efficiency_i = HALE_i-\widehat{HALE}_i
\]

A **positive residual** indicates HALE above the resource-demographic benchmark, whereas a **negative residual** indicates HALE below the benchmark.

**Estimation method:** Cross-sectional OLS on 181 country averages, with HC3 robust standard errors.

**Performance:** \(R^2=.771\). Thus 77.1% of the cross-country variation in HALE is accounted for by the benchmark inputs and 22.9% remains in the residual component. This \(R^2\) is an **in-sample explanatory benchmark fit statistic**, not machine-learning prediction accuracy.

**Interpretation:** Predicted/expected HALE is calculated directly from the fitted equation above; conversion efficiency is the signed difference between actual and predicted HALE.

## **13  Is efficiency persistent over time?**

### Purpose
Determine whether a country's conversion efficiency is a durable characteristic or a transient artefact.

### Rationale
A large residual is not automatically meaningful. Measurement error can also create a wide spread of scores that changes across periods. Persistence therefore helps distinguish a purely transient-noise interpretation from a more durable benchmarking signal. Even strong persistence does not prove that the residual is causal, purely structural, or directly actionable; it supports using the measure for further benchmarking and investigation.

### Method
Efficiency is computed in **two non-overlapping five-year windows, 2006–2010 and 2017–2021**, separated by roughly a decade. Equal-length five-year windows improve temporal comparability and reduce sensitivity to single-year fluctuations, while the same frontier specification is independently re-estimated within each period rather than projected from one period to the other. The two scores are correlated across the countries present in both windows, and the relationship is checked non-parametrically as well as by Pearson correlation. A quartile transition table shows whether starting position predicts ending position.

### Expected output
Two independently estimated efficiency series, their correlation with significance, and a quartile transition table.

```python
def period_efficiency(d, y0, y1, inputs):
    p = (d[d.Year.between(y0,y1)]
           .groupby(["iso3","Country","IncomeGroup"], as_index=False)[["HALE"]+inputs].mean()
           .dropna(subset=["HALE"]+inputs))
    m = sm.OLS(p.HALE, sm.add_constant(p[inputs])).fit()
    p["Eff"] = m.resid
    return p[["iso3","Country","IncomeGroup","HALE","Eff"]], m.rsquared

E1, r1 = period_efficiency(eng, 2006, 2010, FRONT_IN)
E2, r2 = period_efficiency(eng, 2017, 2021, FRONT_IN)
P = E1.merge(E2, on=["iso3","Country"], suffixes=("_early","_late"))
rho, p_rho = stats.pearsonr(P.Eff_early, P.Eff_late)
srho, sp = stats.spearmanr(P.Eff_early, P.Eff_late)

banner("H1b — PERSISTENCE OF EFFICIENCY")
print(f"Early window 2006–2010: {len(E1)} countries, frontier R² = {r1:.3f}")
print(f"Late  window 2017–2021: {len(E2)} countries, frontier R² = {r2:.3f}")
print(f"Matched pairs         : {len(P)}")
print(f"\nPearson  r = {rho:+.3f}   p = {p_rho:.3g}")
print(f"Spearman r = {srho:+.3f}   p = {sp:.3g}")
print(f"Shared variance across a decade: {rho**2*100:.1f}%")
H1b = (p_rho < .05) and (rho > .5)
print(f"\nH1b₀ {'REJECTED' if H1b else 'not rejected'} — efficiency is "
      f"{'a durable country characteristic, not noise' if H1b else 'not clearly persistent'}.")
```

### Model Specification and Equation

**Model:** Two period-specific OLS HALE benchmark regressions plus Pearson and Spearman persistence tests.

**Purpose:** Re-estimate the same resource-demography benchmark in two non-overlapping five-year windows and test whether country residual efficiency is persistent across time.

**Outcome (Y):** Country-average `HALE`.

**Predictors (X):** The same five benchmark inputs used in Section 12: `logHealthExpPC`, `logGDPPerCapita`, `Pop65Plus`, `UrbanPopulation`, and `FertilityRate`.

**General specification for each period \(p\):**

\[
HALE_{i,p}=\beta_{0,p}+\sum_{k=1}^{5}\beta_{k,p}X_{ik,p}+\varepsilon_{i,p},
\qquad Efficiency_{i,p}=\varepsilon_{i,p}
\]

The notebook fits this equation separately for **2006–2010** and **2017–2021**. The individual period-specific coefficients are not printed in the executed output, so no numerical coefficient equation is reproduced here without re-estimation.

**Estimation/validation method:** Separate OLS regressions on country averages, followed by Pearson and Spearman correlations of the matched residuals.

**Performance/statistical evidence:** Early-period benchmark \(R^2=0.718\); late-period benchmark \(R^2=0.771\); 178 matched countries; Pearson \(r=0.885\), \(p=3.26e-60\); Spearman \(r=0.832\), \(p=7.09e-47\); shared variance \(=78.3\%\).

```python
fig, ax = plt.subplots(1, 2, figsize=(15,6))
for g in order:
    z = P[P.IncomeGroup_early == g]
    ax[0].scatter(z.Eff_early, z.Eff_late, s=34, alpha=.8, c=pal.get(g,"#888"), label=g, edgecolor="none")
xs = np.linspace(P.Eff_early.min(), P.Eff_early.max(), 50)
b1, b0 = np.polyfit(P.Eff_early, P.Eff_late, 1)
ax[0].plot(xs, b0+b1*xs, "k-", lw=2)
ax[0].axhline(0, c="grey", lw=.8); ax[0].axvline(0, c="grey", lw=.8)
ax[0].set_xlabel("Efficiency 2006–2010 (years)"); ax[0].set_ylabel("Efficiency 2017–2021 (years)")
ax[0].set_title(f"Efficiency a decade apart\nr = {rho:+.3f}, p = {p_rho:.2g}, n = {len(P)}",
                fontweight="bold")
ax[0].legend(fontsize=7)

q = pd.qcut(P.Eff_early, 4, labels=["Q1 worst","Q2","Q3","Q4 best"])
tr = P.groupby(q).Eff_late.agg(["mean","std","count"])
ax[1].bar(range(4), tr["mean"], yerr=tr["std"]/np.sqrt(tr["count"]), capsize=5,
          color=["#b2182b","#ef8a62","#67a9cf","#2166ac"])
ax[1].set_xticks(range(4)); ax[1].set_xticklabels(tr.index)
ax[1].axhline(0, c="k", lw=1)
ax[1].set_ylabel("Mean efficiency 2017–2021 (years)")
ax[1].set_xlabel("Efficiency quartile in 2006–2010")
ax[1].set_title("Where countries started predicts where they ended", fontweight="bold")
savefig("13_persistence")
display(tr.round(3))
```

## **14  Which observable characteristics are associated with efficiency?**

### Purpose
Identify which observable characteristics of a health system are associated with its conversion efficiency.

### Rationale
A residual that is large and persistent is still not useful until its observable correlates are examined. Efficiency as measured so far is defined by what it is not — not resources and not demography — and a health policymaker cannot act on a negative definition. This section tests four prespecified blocks of candidates deliberately excluded from the frontier: financing mix, service reach, clinical capacity, and governance.

### Method
Each of the 12 candidates is regressed on the efficiency score under three unchanged specifications — raw, with income-group fixed effects, and with region fixed effects — using OLS with HC3 robust standard errors. Benjamini-Hochberg false-discovery-rate correction is applied across the full family of **36 tests** (Benjamini & Hochberg, 1995). The detailed table reports both nominal \(p\)-values and BH-FDR \(q\)-values.

Robustness is summarized at two distinct levels. **Nominal robust** means \(p<.05\) in all three specifications. **FDR robust** means \(q<.05\) in all three specifications. These labels are not interchangeable. The H1c decision is based on the **overall weight of evidence**, with particular emphasis on consistency across specifications together with support after multiplicity correction. Evidence is considered sufficient to reject H1c₀ when a candidate is nominally significant in all three specifications and remains BH-FDR significant in at least two of the three specifications.

### Expected output
The unchanged detailed 36-test results table; a per-variable summary reporting nominal and FDR significance counts, minimum and maximum \(p\)- and \(q\)-values, and separate nominal/FDR robustness flags; a coefficient chart distinguishing nominal from FDR robustness; and a diagnostic plot for the strongest candidate signal.


### **14A Prespecified candidate health-system determinants of conversion efficiency**

#### Why this section uses 12 candidates rather than all 26 registered indicators

The **26-variable registry** is the broader conceptually eligible analytical pool for the capstone. Section 14 does **not** treat all 26 as interchangeable predictors. Its purpose is narrower: to test whether the **Section 12 conversion-efficiency residual** is associated with health-system characteristics that were deliberately kept outside the resource-demography frontier.

The 12 Section 14 candidates therefore come from four prespecified health-system domains:

| Domain | Candidate variables | Why included here |
|---|---|---|
| Financing mix | `HealthExpGDP`, `GovHealthExpShare`, `OOPExpenditure`, `ExternalHealthExpShare` | Describe how health financing is structured or allocated rather than the absolute resource level already represented in the frontier. |
| Service reach | `UHCIndex`, `DPTImmunization` | Represent realised population reach/coverage of health services. |
| Clinical capacity | `Physicians`, `NursesMidwives`, `HospitalBeds` | Represent available delivery capacity and workforce/infrastructure. |
| Governance | `GovEffectiveness`, `ControlOfCorruption`, `RegulatoryQuality` | Represent institutional conditions potentially associated with converting resources into health outcomes. |

The remaining registered indicators are not silently discarded. They serve other roles in the study, including **frontier/context controls, risk-factor adjustment, other research questions, and the RQ3 predictive feature set**. Section 14 is therefore a **targeted explanatory screen**, not a second feature-selection algorithm.

#### Why one candidate is tested at a time

Each focal candidate is estimated in three prespecified specifications: **raw**, **income-group adjusted**, and **region adjusted**. Testing one focal candidate at a time keeps the coefficient interpretable and avoids turning this exploratory explanatory screen into a large, unstable cross-sectional model. Because 12 candidates × 3 specifications produce **36 hypothesis tests**, Benjamini–Hochberg FDR correction is applied across the 36 reported tests.

> **Important:** the 36 rows in the output are **36 regression specifications, not 36 different variables**.

The outcome is the Section 12 `Efficiency` residual. These models identify **associations with conversion efficiency**; they do not establish causal effects.


```python
CANDIDATES = {
 "Financing mix": FINANCING_MIX,
 "Service reach": [v for v in REACH if v in eng.columns],
 "Clinical capacity": [v for v in CAPACITY if v in eng.columns],
 "Governance": [v for v in GOVERNANCE if v in eng.columns],
}
extra = sorted({v for vs in CANDIDATES.values() for v in vs})

# Transparency audit: Section 14 is intentionally a 12-candidate explanatory screen.
SECTION14_CANDIDATE_AUDIT = pd.DataFrame(
    [(blk, v) for blk, vs in CANDIDATES.items() for v in vs],
    columns=["Domain", "Candidate"]
)
print("Section 14 candidate audit:")
display(SECTION14_CANDIDATE_AUDIT)
print(f"Unique Section 14 candidates : {SECTION14_CANDIDATE_AUDIT['Candidate'].nunique()}")
print("Specifications per candidate: 3 (raw, + income group, + region)")
print(f"Expected H1c tests           : {SECTION14_CANDIDATE_AUDIT['Candidate'].nunique() * 3}")
assert SECTION14_CANDIDATE_AUDIT["Candidate"].nunique() == 12, (
    "Section 14 candidate count changed; review the documented 12-candidate rationale."
)
mix = eng[eng.Year.between(2017,2021)].groupby("iso3", as_index=False)[extra].mean()
fr2 = fr.merge(mix, on="iso3", how="left")

rows = []
for blk, vs in CANDIDATES.items():
    for v in vs:
        d2 = fr2.dropna(subset=["Efficiency", v])
        if len(d2) < 80: continue
        for lbl, ex in [("raw",""), ("+ income group"," + C(IncomeGroup)"), ("+ region"," + C(Region)")]:
            m = smf.ols(f"Efficiency ~ {v}{ex}", data=d2).fit(cov_type="HC3")
            rows.append({"Block":blk, "Variable":v, "Specification":lbl, "b":m.params[v],
                         "SE":m.bse[v], "t":m.tvalues[v], "p":m.pvalues[v], "n":int(m.nobs)})
H1C = pd.DataFrame(rows)
H1C["q (BH-FDR)"] = multipletests(H1C.p, method="fdr_bh")[1]
assert len(H1C) == 36, f"Expected 36 H1c tests (12 candidates × 3 specs), got {len(H1C)}."
banner("H1c — WHAT, IF ANYTHING, EXPLAINS EFFICIENCY?")
display(H1C.round(4))

SUMM = (
    H1C.groupby(["Block", "Variable"])
    .agg(
        **{
            "Mean b": ("b", "mean"),
            "Nominal significant in": ("p", lambda z: (z < .05).sum()),
            "FDR significant in": ("q (BH-FDR)", lambda z: (z < .05).sum()),
            "of": ("p", "size"),
            "Min p": ("p", "min"),
            "Max p": ("p", "max"),
            "Min q": ("q (BH-FDR)", "min"),
            "Max q": ("q (BH-FDR)", "max"),
        }
    )
    .reset_index()
)
SUMM["Nominal robust"] = SUMM["Nominal significant in"] == SUMM["of"]
SUMM["FDR robust"] = SUMM["FDR significant in"] == SUMM["of"]
display(SUMM.round(4))

NOMINAL_ROBUST = SUMM[SUMM["Nominal robust"]]
FDR_ROBUST = SUMM[SUMM["FDR robust"]]

TOP_SIGNAL = (
    SUMM.sort_values(
        ["FDR significant in", "Nominal significant in", "Min q"],
        ascending=[False, False, True]
    ).iloc[0]
)

WEIGHT_OF_EVIDENCE = SUMM[
    (SUMM["Nominal significant in"] == SUMM["of"]) &
    (SUMM["FDR significant in"] >= 2)
]
H1c = len(WEIGHT_OF_EVIDENCE) > 0

if H1c:
    print("\nH1c₀ REJECTED based on the overall weight of evidence.")
    print(f"{TOP_SIGNAL.Variable} is nominally significant in "
          f"{int(TOP_SIGNAL['Nominal significant in'])}/{int(TOP_SIGNAL['of'])} specifications "
          f"and FDR-significant in {int(TOP_SIGNAL['FDR significant in'])}/{int(TOP_SIGNAL['of'])}.")
    print(f"It is not FDR-robust across all three specifications because its q-values range from "
          f"{TOP_SIGNAL['Min q']:.4f} to {TOP_SIGNAL['Max q']:.4f}.")
else:
    print("\nH1c₀ not rejected based on the overall weight-of-evidence criterion.")

print("\nThe analysis found no statistically robust evidence that the measured financing-mix")
print("variables were associated with conversion efficiency. The measured clinical-capacity")
print("variables likewise did not show consistent evidence of association with efficiency.")
print(f"{TOP_SIGNAL.Variable} emerged as the most consistent candidate indicator of service reach:")
print("positive and nominally significant in all three specifications, with BH-FDR support in two of three.")
```

# Part III — Theme 2 / RQ2: Actionable Determinants of Healthy Life Expectancy


## **15  Reach versus capacity under two-way fixed effects**

### Purpose
Identify actionable health-system determinants associated with healthy life expectancy by separating service-reach instruments from clinical-capacity instruments within a common longitudinal framework.

### Rationale
This is the budget question in its most direct form. A health authority with additional money can widen coverage of what it already delivers or deepen the clinical establishment. These are different theories of how health is produced and they carry very different costs, so the comparison needs to be made explicitly rather than by including both in one model and reading whichever coefficient is larger.

### Method
Country fixed effects absorb time-invariant country characteristics, so the two-way fixed-effects coefficients are identified from **within-country changes over time**; year fixed effects absorb shocks common to all countries, and standard errors are clustered by country (Wooldridge, 2010). The earlier variance-decomposition table is reported only to show whether each variable's information is predominantly within or between countries and is **not** used to exclude variables. Variables and analytical blocks are selected from the conceptual framework and research questions before estimation; the regressions subsequently test the specified associations. Critically, each block is estimated on its **own maximal sample** before the joint model is reported, because nurse and hospital-bed density are reported sporadically and forcing one complete-case model would shrink the sample substantially for reasons unrelated to the substantive hypotheses.

> **Why both samples are shown.** The maximal-sample models use all observations available to each block and therefore maximize statistical power. The common-sample model restricts reach and capacity to the same country-years, making it the fairest direct head-to-head comparison.

### Expected output
A sample-size comparison across specifications, block-specific coefficient tables, joint F-tests, and a coefficient plot with clustered confidence intervals.

```python
def twfe(d, outcome, rhs, cluster="iso3"):
    cols = [outcome] + rhs + [cluster, "Year"]
    s = d[[c for c in cols if c in d.columns]].dropna()
    if s[cluster].nunique() < 20 or len(s) < 100: return None, s
    f = f"{outcome} ~ " + " + ".join(rhs) + f" + C({cluster}) + C(Year)"
    return smf.ols(f, data=s).fit(cov_type="cluster", cov_kwds={"groups": s[cluster]}), s

REACH_V = [v for v in REACH if v in eng.columns]
CAP_V   = [v for v in CAPACITY if v in eng.columns]
CTRL_V  = [v for v in ["logHealthExpPC","logGDPPerCapita","Pop65Plus","UrbanPopulation"] if v in eng.columns]

def sample_of(d, outcome, rhs):
    cols = [outcome] + rhs + ["iso3","Year"]
    s = d[[c for c in cols if c in d.columns]].dropna()
    return len(s), s.iso3.nunique()

SAMPLES = pd.DataFrame([
 {"Specification":"Reach block only",   "Variables":", ".join(REACH_V),
  **dict(zip(["Rows","Countries"], sample_of(eng,"HALE",REACH_V+CTRL_V)))},
 {"Specification":"Capacity block only","Variables":", ".join(CAP_V),
  **dict(zip(["Rows","Countries"], sample_of(eng,"HALE",CAP_V+CTRL_V)))},
 {"Specification":"Both blocks jointly","Variables":", ".join(REACH_V+CAP_V),
  **dict(zip(["Rows","Countries"], sample_of(eng,"HALE",REACH_V+CAP_V+CTRL_V)))},
])
banner("SAMPLE SIZE IS NOT THE SAME ACROSS SPECIFICATIONS")
display(SAMPLES)
print("\nNurse and hospital-bed density are reported sporadically. Forcing every block into one")
print("complete-case model would discard most of the panel and most of the poorer countries,")
print("which would understate the reach block simply because the sample shrank. Each block is")
print("therefore estimated on its own maximal sample first, and the joint model is reported")
print("afterwards as a common-sample head-to-head.")
```

```python
M_REACH, S_REACH = twfe(eng, "HALE", REACH_V + CTRL_V)
M_CAP,   S_CAP   = twfe(eng, "HALE", CAP_V   + CTRL_V)
M_HALE,  S_HALE  = twfe(eng, "HALE", REACH_V + CAP_V + CTRL_V)

def joint_F(model, terms):
    terms = [t for t in terms if t in model.params.index]
    if not terms: return None
    r = model.f_test(" = 0, ".join(terms) + " = 0")
    return {"Terms": ", ".join(terms), "F": float(np.squeeze(r.fvalue)),
            "df": r.df_num, "p": float(np.squeeze(r.pvalue))}

banner("REACH BLOCK — maximal sample")
print(f"{int(M_REACH.nobs):,} observations, {S_REACH.iso3.nunique()} countries")
print(pd.DataFrame({"Coefficient":M_REACH.params,"SE":M_REACH.bse,"p":M_REACH.pvalues})
        .loc[[v for v in REACH_V+CTRL_V if v in M_REACH.params.index]].round(5).to_string())

banner("CAPACITY BLOCK — maximal sample")
print(f"{int(M_CAP.nobs):,} observations, {S_CAP.iso3.nunique()} countries")
print(pd.DataFrame({"Coefficient":M_CAP.params,"SE":M_CAP.bse,"p":M_CAP.pvalues})
        .loc[[v for v in CAP_V+CTRL_V if v in M_CAP.params.index]].round(5).to_string())
```

### Model Specification and Equation

**Model:** Two-way fixed-effects (TWFE) regressions for HALE: reach model, capacity model, and joint reach-capacity model.

**Purpose:** Estimate within-country associations between actionable health-system levers and HALE while controlling for time-invariant country characteristics and common year shocks.

**Outcome (Y):** `HALE`.

**General panel specification:**

\[
HALE_{it}=\alpha_i+\lambda_t+\mathbf{X}_{it}'\beta+\varepsilon_{it}
\]

where \(\alpha_i\) denotes country fixed effects and \(\lambda_t\) denotes year fixed effects. Standard errors are clustered by country. These are **within-country explanatory models**, not cross-sectional OLS models and not prediction models.

#### Reach model

**Predictors (X):** `UHCIndex`, `DPTImmunization`, `logHealthExpPC`, `logGDPPerCapita`, `Pop65Plus`, `UrbanPopulation`.

## **16  Wild cluster bootstrap**

### Purpose
Re-test the block coefficients without relying on the asymptotic approximation behind cluster-robust standard errors.

### Rationale
Cluster-robust standard errors are consistent as the number of clusters grows, and with a moderate number of clusters they are known to be optimistic — they under-reject, and a coefficient can appear significant largely because of that approximation. A study whose central claim rests on a p-value should check whether the p-value survives a procedure that does not make that assumption.

### Method
The wild cluster bootstrap-t with Rademacher weights and the null imposed is used as a robustness procedure for clustered inference (Cameron et al., 2008; MacKinnon & Webb, 2018). Residuals are resampled at the cluster level from the restricted model, and the observed t-statistic is compared against the resulting distribution. The implementation precomputes the design matrix and projection once so that the resampling loop is a sequence of matrix products rather than a thousand model refits.

### Expected output
Bootstrap p-values alongside the asymptotic ones, with explicit agreement flags.

```python
def wild_bootstrap(model, sample, var, B=999, seed=SEED):
    import patsy
    y, X = patsy.dmatrices(model.model.formula, sample, return_type="dataframe")
    yv, Xm = np.asarray(y).ravel(), np.asarray(X)
    if var not in list(X.columns): return None
    j = list(X.columns).index(var)
    codes, uniq = pd.factorize(sample.iso3.values); G = len(uniq); n = len(yv)
    XtX_inv = np.linalg.pinv(Xm.T @ Xm); P = XtX_inv @ Xm.T
    a = XtX_inv[:, j]; u = Xm @ a
    def tstat(yy):
        b = P @ yy; e = yy - Xm @ b
        s = np.bincount(codes, weights=u*e, minlength=G)
        V = float((s**2).sum())
        c = G/(G-1) * (n-1)/(n - Xm.shape[1])
        return b[j], b[j]/np.sqrt(max(V*c, 1e-300))
    b_obs, t_obs = tstat(yv)
    # restricted model: coefficient forced to zero
    keep = [i for i in range(Xm.shape[1]) if i != j]
    Xr = Xm[:, keep]
    br = np.linalg.pinv(Xr.T @ Xr) @ Xr.T @ yv
    yhat_r = Xr @ br; er = yv - yhat_r
    rng = np.random.default_rng(seed)
    ts = np.empty(B)
    for bi in range(B):
        w = rng.choice([-1.0, 1.0], size=G)[codes]
        ts[bi] = tstat(yhat_r + er*w)[1]
    return {"Variable":var, "b":b_obs, "t (clustered)":t_obs,
            "p (bootstrap)": float((np.abs(ts) >= abs(t_obs)).mean()),
            "Clusters":G, "B":B}

WB = pd.DataFrame([r for r in (
        [cached(f"wb_reach_{v}", lambda v=v: wild_bootstrap(M_REACH, S_REACH, v)) for v in REACH_V] +
        [cached(f"wb_cap_{v}",   lambda v=v: wild_bootstrap(M_CAP,   S_CAP,   v)) for v in CAP_V]
    ) if r])
WB = WB.merge(COEF[["p"]].rename(columns={"p":"p (asymptotic)"}),
              left_on="Variable", right_index=True, how="left")
WB["Agrees"] = np.where((WB["p (bootstrap)"]<.05) == (WB["p (asymptotic)"]<.05), "yes", "NO — see note")
display(WB.round(5))
print("\nWhere the two disagree, the bootstrap result is the one reported.")
```

### Model Specification and Equation

**Procedure:** Wild cluster bootstrap inference for selected TWFE coefficients.

**Purpose:** Re-test the reach and capacity coefficients under cluster-level resampling rather than fit a new substantive outcome model.

**Underlying models:** The Section 15 reach and capacity TWFE regressions. No new dependent variable or predictor set is introduced.

**Bootstrap specification:** The null restriction sets the focal coefficient to zero, residuals from the restricted model are multiplied by cluster-level Rademacher weights (\(-1,+1\)), and the clustered \(t\)-statistic is recomputed over 999 replications.

**Reported evidence:** UHC \(b=.1483\), bootstrap \(p=.0000\); DPT \(b=.0386\), \(p=.0000\); Physicians \(b=-.4200\), \(p=.0220\); Nurses \(b=.0320\), \(p=.6747\); Beds \(b=.0005\), \(p=.9980\).

**Interpretation:** This is an inference/robustness procedure applied to the TWFE models, not a separate prediction model and not a source of a new fitted equation.

### Result
The reach coefficients survive the bootstrap; the pattern of agreement between the two methods is reported per variable.

### Interpretation
Where the two disagree the bootstrap result is the one reported, because it makes fewer assumptions about the cluster structure. The agreement column exists so a reader can see that this check was applied to every coefficient rather than to the ones where it happened to be favourable.

### Decision
Adopt bootstrap inference for the block coefficients.

## **17  Substitutes or complements?**

### Purpose
Test whether the returns to extending reach depend on how much clinical capacity a country already has.

### Rationale
The interaction tests whether the association of service reach with HALE changes as clinical capacity changes. A negative interaction is consistent with diminishing combined associations, while a positive interaction is consistent with amplification or complementarity. Because the analysis is observational, the sign does not by itself prescribe spending sequence, prove that one lever substitutes causally for another, or establish which intervention should be funded first. The distinction cannot be read from the main effects alone.

### Method
The interaction is estimated under three specifications on purpose: two-way fixed effects, country fixed effects alone, and a pooled cross-section. The answer depends on which variation is used, and that dependence is itself the finding rather than an inconsistency to be resolved. The marginal effect of reach is then plotted across the observed range of capacity, with the crossover point located.

### Expected output
Interaction coefficients under all three specifications, and a marginal-effect curve identifying where the return to reach reaches zero.

```python
def interaction_specs(d, outcome, a, b, ctrls):
    s = d[[outcome, a, b, "iso3", "Year"] + [c for c in ctrls if c in d.columns]].dropna()
    ct = " + ".join(c for c in ctrls if c in s.columns)
    out = []
    for lbl, form in [
        ("Two-way FE (within-country, net of year shocks)",
         f"{outcome} ~ {a}*{b} + {ct} + C(iso3) + C(Year)"),
        ("Country FE only (within-country, includes trends)",
         f"{outcome} ~ {a}*{b} + {ct} + C(iso3)"),
        ("Pooled cross-section (between-country)",
         f"{outcome} ~ {a}*{b} + {ct} + C(Year)")]:
        m = smf.ols(form, data=s).fit(cov_type="cluster", cov_kwds={"groups": s.iso3})
        term = f"{a}:{b}"
        if term not in m.params.index: continue
        out.append({"Specification":lbl, "Interaction b":m.params[term], "SE":m.bse[term],
                    "t":m.tvalues[term], "p":m.pvalues[term], "n":int(m.nobs),
                    "Reading":"substitutes" if m.params[term]<0 else "complements"})
    return pd.DataFrame(out), s

if "UHCIndex" in eng.columns and "Physicians" in eng.columns:
    INT, SINT = cached("interaction_uhc_phys",
        lambda: interaction_specs(eng, "HALE", "UHCIndex", "Physicians",
                                  ["logHealthExpPC","logGDPPerCapita","Pop65Plus"]))
    banner("H2c — REACH × CAPACITY INTERACTION")
    display(INT.round(5))
    H2c = (INT.p < .05).any()
    within = INT[INT.Specification.str.contains("FE")]
    between = INT[INT.Specification.str.contains("cross-section")]
    print(f"\nH2c₀ {'REJECTED' if H2c else 'not rejected'} — the interaction is significant in "
          f"{(INT.p<.05).sum()}/{len(INT)} specifications.")
    if len(within) and len(between) and (within["Interaction b"].mean() * between["Interaction b"].mean() < 0):
        print("\nThe sign flips between within-country and between-country variation.")
        print("Within a country over time, coverage and physicians are SUBSTITUTES: as coverage")
        print("widens, the marginal healthy year from an additional physician falls. Across")
        print("countries at a point in time they look like COMPLEMENTS, because rich countries")
        print("simply have more of both. A ministry allocating next year's budget faces the")
        print("within-country margin, so the substitution result is the decision-relevant one.")
```

### Model Specification and Equation

**Model:** HALE reach-capacity interaction models using `UHCIndex × Physicians`.

**Purpose:** Test whether the association between service coverage and HALE changes with physician density.

**Outcome (Y):** `HALE`.

**Predictors (X):** `UHCIndex`, `Physicians`, their interaction, `logHealthExpPC`, `logGDPPerCapita`, and `Pop65Plus`, with fixed effects varying by specification.

**General interaction specification:**

\[
HALE_{it}=\beta_0+\beta_1UHC_{it}+\beta_2Physicians_{it}
+\beta_3(UHC_{it}\times Physicians_{it})
+\mathbf{C}_{it}'\gamma+\text{fixed effects}+\varepsilon_{it}
\]

The marginal association of UHC with HALE is:

\[
\frac{\partial HALE}{\partial UHC}=\beta_1+\beta_3Physicians
\]

Thus a negative \(\beta_3\) means the UHC slope becomes smaller as physician density increases.

**Executed interaction coefficients:**

| Specification | Fixed-effects structure | Interaction b | SE | p | n |
|---|---|---:|---:|---:|---:|
| Two-way FE | Country + year FE | -0.0268 | 0.0091 | .0031 | 2,565 |
| Country FE only | Country FE | -0.0169 | 0.0092 | .0661 | 2,565 |
| Pooled year-adjusted | Year FE only | -0.0389 | 0.0133 | .0034 | 2,565 |

The executed output does not print all main-effect and control coefficients for these three fits, so full numerical fitted equations are not reconstructed beyond the reported interaction term.

**Interpretation:** The principal TWFE interaction is negative and significant. The executed output does not support a sign-flip claim; all three reported interaction coefficients are negative.

```python
if "INT" in dir() and len(INT):
    fig, ax = plt.subplots(1, 2, figsize=(15,5.5))
    ax[0].barh(INT.Specification, INT["Interaction b"],
               xerr=1.96*INT.SE, color=np.where(INT["Interaction b"]<0,"#b2182b","#1b7837"),
               capsize=5)
    ax[0].axvline(0, c="k", lw=1.2)
    ax[0].set_xlabel("UHC × Physicians interaction coefficient")
    ax[0].set_title("The sign depends on which variation you use", fontweight="bold")
    ax[0].tick_params(labelsize=8)

    m0 = smf.ols("HALE ~ UHCIndex*Physicians + logHealthExpPC + logGDPPerCapita + Pop65Plus"
                 " + C(iso3) + C(Year)", data=SINT).fit()
    bU, bI = m0.params["UHCIndex"], m0.params["UHCIndex:Physicians"]
    ph = np.linspace(SINT.Physicians.quantile(.05), SINT.Physicians.quantile(.95), 60)
    ax[1].plot(ph, bU + bI*ph, lw=2.5, c="#2166ac")
    ax[1].axhline(0, c="k", ls="--", lw=1)
    ax[1].fill_between(ph, bU+bI*ph, 0, where=(bU+bI*ph)>0, alpha=.15, color="#1b7837")
    ax[1].fill_between(ph, bU+bI*ph, 0, where=(bU+bI*ph)<0, alpha=.15, color="#b2182b")
    ax[1].set_xlabel("Physicians per 1,000 population")
    ax[1].set_ylabel("Marginal effect of +1 UHC index point on HALE (years)")
    ax[1].set_title("The return to widening coverage falls as\nclinical capacity rises",
                    fontweight="bold")
    savefig("17_interaction")
    cross = -bU/bI if bI != 0 else np.nan
    if np.isfinite(cross) and SINT.Physicians.min() < cross < SINT.Physicians.max():
        share = (SINT.groupby("iso3").Physicians.mean() < cross).mean()*100
        print(f"\nThe marginal return to coverage reaches zero at about {cross:.2f} physicians per 1,000.")
        print(f"{share:.0f}% of countries are below that point — for them, coverage is still the")
        print("higher-return purchase.")
```

### Result
The interaction coefficient is negative in all three reported specifications. It is statistically significant in the two-way fixed-effects model and in the pooled year-adjusted model, but not in the country-fixed-effects-only model.

### Interpretation
The executed output does **not** show a sign reversal between within-country and between-country specifications. The principal two-way fixed-effects estimate is negative (*b* = −0.0268, *p* = .0031), indicating that the marginal association of UHC coverage with HALE becomes smaller as physician density rises, conditional on the model controls, country fixed effects, and year fixed effects. The pooled year-adjusted specification is also negative (*b* = −0.0389, *p* = .0034), while the country-fixed-effects-only specification is negative but not conventionally significant (*b* = −0.0169, *p* = .0661). The interaction therefore supports a substitution-style interpretation in the principal within-country specification, but not a within-versus-between sign-flip claim.

### Decision
Report the two-way fixed-effects interaction as the principal inferential result and retain the alternative specifications as sensitivity checks.

## **18  The same levers against NCD mortality**

### Purpose
Estimate the lever associations for premature non-communicable-disease mortality, and diagnose why the fixed-effects result is weak.

### Rationale
NCD mortality responds to behaviour and risk exposure accumulated over decades. A weak fixed-effects result is expected here, and the important task is to explain why rather than to present the weakness as a finding or to quietly omit the outcome.

### Method
The same two-way fixed-effects specification is estimated on NCD mortality with FDR correction; these coefficients use within-country variation only. The model is then re-estimated using the Mundlak decomposition (Mundlak, 1978) so that between-country and within-country variation enter as separate explicit terms rather than one being absorbed into the country dummy. As elsewhere, the descriptive variance decomposition informs interpretation of where variation resides but does not determine variable inclusion.

### Expected output
Fixed-effects coefficients with corrected p-values, joint tests, and a Mundlak table separating long-run level from year-to-year movement.

### Model Specification and Equation

**Model:** Two-way fixed-effects regression for premature NCD mortality.

**Purpose:** Estimate within-country associations of reach, capacity, resources, and demographic controls with premature NCD mortality.

**Outcome (Y):** `NCDMortality`.

**Predictors (X):** `UHCIndex`, `DPTImmunization`, `Physicians`, `NursesMidwives`, `HospitalBeds`, `logHealthExpPC`, `logGDPPerCapita`, `Pop65Plus`, and `UrbanPopulation`.

**General specification:**

\[
NCDMortality_{it}=\alpha_i+\lambda_t+\mathbf{X}_{it}'\beta+\varepsilon_{it}
\]

**Fitted substantive-slope equation:**

\[
\begin{aligned}
\widehat{NCDMortality}_{it}=&\ \alpha_i+\lambda_t
+0.2461\,UHCIndex_{it}
+0.0066\,DPTImmunization_{it}
+0.2194\,Physicians_{it}\\
&+0.1130\,NursesMidwives_{it}
+0.1461\,HospitalBeds_{it}
+0.2031\,logHealthExpPC_{it}\\
&-2.1582\,logGDPPerCapita_{it}
+0.3102\,Pop65Plus_{it}

**Estimation method:** TWFE OLS with country and year fixed effects, country-clustered standard errors, 1,907 observations across 160 countries, followed by BH-FDR correction.

```python
def mundlak(d, outcome, vars_):
    cols = [outcome] + [f"{v}_{s}" for v in vars_ for s in ("btw","wth")] + ["iso3","Year"]
    s = d[[c for c in cols if c in d.columns]].dropna()
    rhs = [c for c in s.columns if c.endswith(("_btw","_wth"))]
    m = smf.ols(f"{outcome} ~ " + " + ".join(rhs) + " + C(Year)",
                data=s).fit(cov_type="cluster", cov_kwds={"groups": s.iso3})
    t = pd.DataFrame({"Coefficient":m.params, "SE":m.bse, "p":m.pvalues}).loc[rhs]
    t["Variation"] = np.where(t.index.str.endswith("_btw"),
                              "between (long-run level)", "within (year-to-year)")
    t["Variable"] = t.index.str.replace("_btw","",regex=False).str.replace("_wth","",regex=False)
    return m, t, s

MU_VARS = [v for v in REACH_V + CAP_V + ["logHealthExpPC"] + RISK if v in eng.columns]
m_mu, T_MU, S_MU = cached("mundlak_ncd", lambda: mundlak(eng, "NCDMortality", MU_VARS))
print(f"Mundlak model — {int(m_mu.nobs):,} obs, R² = {m_mu.rsquared:.3f}")
display(T_MU.pivot_table(index="Variable", columns="Variation",
                         values=["Coefficient","p"]).round(4))
nb_ = (T_MU[(T_MU.Variation.str.startswith("between")) & (T_MU.p<.05)]).shape[0]
nw_ = (T_MU[(T_MU.Variation.str.startswith("within"))  & (T_MU.p<.05)]).shape[0]
print(f"\nSignificant between-country terms: {nb_}")
print(f"Significant within-country terms : {nw_}")
print("\nThe asymmetry is the point. The information about NCD mortality is in the differences")
print("between countries, which a fixed-effects model deletes by construction. Reporting a")
print("fixed-effects null here without saying so would be misleading.")
```

\]

The intercept and year-dummy coefficients are not printed in the existing output and are therefore left symbolic.

**Estimation method:** OLS with year fixed effects and country-clustered standard errors; 1,641 observations.

**Performance:** \(R^2=.797\), an in-sample model fit statistic used for explanatory decomposition, not machine-learning prediction accuracy.

**Interpretation:** `_btw` coefficients describe associations with country long-run levels; `_wth` coefficients describe associations with deviations from each country's own long-run level.

### Result
The fixed-effects model returns little. The Mundlak decomposition shows the significant terms are overwhelmingly between-country rather than within-country.

### Interpretation
The asymmetry helps explain the result. The earlier descriptive variance decomposition shows that obesity, hypertension and physical inactivity vary predominantly between countries and move relatively slowly within countries. Because two-way fixed effects identify associations from within-country changes, these predictors contribute comparatively little identifying variation to that specification. ## **19  Split design — why folds are grouped by country**

### Purpose
Establish a leakage-resistant validation protocol for explainable machine-learning prediction and demonstrate why country-grouped cross-validation is required for credible generalisation to unseen countries.

### Rationale
The random row-level five-fold cross-validation shown in this section is used **only as a leakage diagnostic**. It is not the validation strategy used for the substantive predictive analysis. With repeated country-year observations, a random row split can place observations from the same country in both training and validation folds. The model can therefore exploit country-specific patterns already represented in training, producing an artificially optimistic estimate of generalisation.

The diagnostic random split yields \(R^2=.964\). The same estimator evaluated with country-grouped five-fold cross-validation yields a mean \(R^2=.774\). The difference, \(0.964-0.774=0.190\) \(R^2\) points, is the **leakage premium**: the amount by which the row-level split inflates apparent predictive performance in this comparison.

### Method
The same Random Forest pipeline (Breiman, 2001) is evaluated twice on the same HALE modelling frame, but the two evaluations are fully separate. `cross_val_score` fits new model instances within each cross-validation run; the models fitted for the random row-level demonstration are not reused by, and do not contaminate, the models fitted under `GroupKFold`.

The actual validation strategy at this stage is **five-fold country-grouped cross-validation**, consistent with recommendations for structured/hierarchical data (Roberts et al., 2017). All observations and years belonging to a country remain in the same fold. In each iteration, approximately 80% of country groups are used for training and approximately 20% are held out for validation; the held-out fold rotates so that every country serves as validation data once. This grouped validation is distinct from the separate locked train/validation/test evaluation introduced later in Section 22.

\(R^2\) is used here because it provides one directly comparable statistic under the two splitting strategies. It is not treated as the only predictive metric: MAE is reported in the grouped model comparison and MAE/RMSE are reported later in the locked-split evaluation.

### Expected output
The random row-level and country-grouped mean \(R^2\) values, the leakage premium, and the observed training/validation country and observation counts for each of the five `GroupKFold` folds.


```python
# PRIMARY RQ3 RULE: at least 70% observed (no more than 30% missing)
PRIMARY_MIN_OBSERVED = 0.70
# SENSITIVITY RULE: earlier broader eligibility, with fold-contained median imputation
SENS_MIN_OBSERVED = 0.55

coverage = pd.Series({v: eng[v].notna().mean() for v in RETAINED if v in eng.columns},
                     name="Observed proportion").sort_values(ascending=False)

FEATS = [v for v in RETAINED if v in eng.columns and eng[v].notna().mean() >= PRIMARY_MIN_OBSERVED]
FEATS_SENS = [v for v in RETAINED if v in eng.columns and eng[v].notna().mean() > SENS_MIN_OBSERVED]
SENS_ONLY = [v for v in FEATS_SENS if v not in FEATS]

banner("RQ3 FEATURE ELIGIBILITY")
print(f"Primary rule     : >=70% observed (<=30% missing)")
print(f"Primary features : {len(FEATS)}")
for v in FEATS: print(f"   {v}")
print(f"\nSensitivity rule     : >55% observed")
print(f"Sensitivity features  : {len(FEATS_SENS)}")
print(f"Additional sensitivity-only features ({len(SENS_ONLY)}):")
for v in SENS_ONLY: print(f"   {v}  (missing {100*(1-coverage[v]):.1f}%)")

assert set(FEATS).issubset(set(FEATS_SENS))
assert all(eng[v].notna().mean() >= .70 for v in FEATS)
assert all(eng[v].notna().mean() > .55 for v in FEATS_SENS)

def build_xy(d, target, feats):
    s = d[[target,"iso3","Year"]+feats].dropna(subset=[target])
    return s[feats], s[target], s.iso3.values, s

Xh, yh, gh, Sh = build_xy(eng, "HALE", FEATS)
Xn, yn, gn, Sn = build_xy(eng, "NCDMortality", FEATS)
print(f"\nHALE frame : {len(Xh):,} rows, {pd.Series(gh).nunique()} countries")
print(f"NCD  frame : {len(Xn):,} rows, {pd.Series(gn).nunique()} countries")

PIPE = lambda est: Pipeline([("imp", SimpleImputer(strategy="median")),
                             ("sc", StandardScaler()),
                             ("m", est)])

from sklearn.model_selection import KFold
rf = PIPE(RandomForestRegressor(n_estimators=300, min_samples_leaf=2,
                                random_state=SEED, n_jobs=-1))
leak = cached("leak_random", lambda: cross_val_score(
    rf, Xh, yh, cv=KFold(5, shuffle=True, random_state=SEED), scoring="r2").mean())
honest = cached("leak_grouped", lambda: cross_val_score(
    rf, Xh, yh, groups=gh, cv=GroupKFold(5), scoring="r2").mean())
banner("WHY THE SPLIT MUST BE GROUPED BY COUNTRY")
print(f"Random row-level 5-fold CV R²      {leak:.3f}   <-- leakage diagnostic only")
print(f"Country-grouped 5-fold CV R²       {honest:.3f}   <-- mean validation R² for unseen countries")
print(f"Leakage premium                    {leak-honest:+.3f}")

group_rows = []
for fold, (tr_idx, va_idx) in enumerate(GroupKFold(5).split(Xh, yh, groups=gh), start=1):
    group_rows.append({
        "Fold": fold,
        "Training countries": pd.Series(gh[tr_idx]).nunique(),
        "Validation countries": pd.Series(gh[va_idx]).nunique(),
        "Training observations": len(tr_idx),
        "Validation observations": len(va_idx),
    })
GROUP_FOLD_SUMMARY = pd.DataFrame(group_rows)
display(GROUP_FOLD_SUMMARY)

print("\nThe row-level diagnostic and grouped-CV fits are separate model fits.")
print("Sections 20–21 use country-grouped validation; Section 22 later uses a separate locked country split.")
```

### **19A  Missing-predictor handling and imputation audit**

#### Purpose
Document exactly **where imputation is used, which variables can be imputed, how much data are missing, and why the chosen method is appropriate**.

#### Important correction
The notebook does **not** mean-impute the whole analytical panel and does **not** use one imputed dataset for RQ1–RQ4.

- **Outcomes are never imputed.**
- **Explanatory/inferential models** use the observations available for the variables required by each specification and report their realised sample size.
- **Predictive models (RQ3)** handle missing predictor values **inside each cross-validation training fold**. This prevents information from held-out countries leaking into preprocessing.
- The predictive pipelines use **median imputation**, not mean imputation. Median imputation is deliberately robust to skewed continuous health/economic indicators and extreme country values.

The **primary RQ3 feature-entry rule is ≥70% observed (≤30% missing)**. Any remaining missing predictor values are median-imputed inside the predictive pipeline. Section 19B separately evaluates the broader >55%-observed feature set as an imputation sensitivity analysis.

The audit table below lists every primary RQ3 predictor, its observed and missing counts/rates in the primary analytical frame, and the handling rule used downstream.


```python
# ---------------------------------------------------------------------------
# SECTION 19A — MISSING-PREDICTOR / IMPUTATION AUDIT
# ---------------------------------------------------------------------------

_imp_frame = eng.loc[:, ["iso3", "Year"] + FEATS].copy()
_imp_n = len(_imp_frame)

IMPUTATION_AUDIT = pd.DataFrame({
    "Predictor": FEATS,
    "Observed n": [_imp_frame[v].notna().sum() for v in FEATS],
    "Missing n": [_imp_frame[v].isna().sum() for v in FEATS],
    "Missing %": [100 * _imp_frame[v].isna().mean() for v in FEATS],
    "Observed %": [100 * _imp_frame[v].notna().mean() for v in FEATS],
})
IMPUTATION_AUDIT["RQ3 entry rule"] = np.where(
    IMPUTATION_AUDIT["Observed %"] >= 70,
    "PASS: >=70% observed",
    "FAIL: <70% observed"
)
IMPUTATION_AUDIT["RQ3 missing-value handling"] = np.where(
    IMPUTATION_AUDIT["Missing n"] > 0,
    "Median imputation inside training/CV fold",
    "No imputation required"
)
IMPUTATION_AUDIT["Explanatory-model handling"] = (
    "No predictor imputation; use specification-available observations"
)

IMPUTATION_AUDIT = IMPUTATION_AUDIT.sort_values(
    ["Missing %", "Predictor"], ascending=[False, True]
).reset_index(drop=True)

display(IMPUTATION_AUDIT.style.format({
    "Missing %": "{:.1f}",
    "Observed %": "{:.1f}",
}))

print(f"Final RQ3 predictors audited : {len(IMPUTATION_AUDIT)}")
print(f"Predictors with any missing : {(IMPUTATION_AUDIT['Missing n'] > 0).sum()}")
print(f"Predictors requiring no fill: {(IMPUTATION_AUDIT['Missing n'] == 0).sum()}")
print("Predictive imputation method : median, fitted within training/CV folds only")
print("Outcome imputation           : none")
print("Inferential-model imputation : none")

assert (IMPUTATION_AUDIT["Observed %"] >= 70).all(), (
    "At least one final RQ3 predictor no longer satisfies the >=70% observed-coverage rule."
)
```

#### Interpretation
This table separates **feature eligibility** from **missing-value handling**. The coverage screen determines whether a variable is sufficiently observed to be considered for RQ3; it does not fill missing values. For predictors that pass that screen, median imputation occurs only within the predictive preprocessing pipeline.

#### Scientific rationale
Median imputation is a conservative single-value preprocessing method for prediction when predictors are continuous and may be skewed or contain legitimate extreme country values. Compared with the mean, the median is less influenced by those extremes. Most importantly for this study, fitting the imputer inside each training fold preserves the separation between training and held-out countries.

This approach is intended to preserve predictive sample size and prevent preprocessing leakage. It does **not** claim that imputation removes the underlying non-random missing-data limitation identified in Section 5.


### **19B  Imputation sensitivity analysis for predictors with 30–45% missingness**

#### Purpose
Test whether the stricter **≤30% missing** primary feature rule materially changes predictive performance compared with a broader feature set that retains predictors with **>55% observed data** and imputes their missing values inside each training fold.

#### Design
This is a **sensitivity analysis**, not a replacement for the primary model. Both feature sets use the same country-grouped five-fold cross-validation, the same outcomes, and the same median-imputation pipeline. The comparison uses `ExtraTreesRegressor` as a common nonlinear benchmark so that the feature-set comparison is not confounded by changing algorithms.

- **Primary set:** predictors with ≥70% observed data.
- **Imputation-sensitivity set:** predictors with >55% observed data.
- **Imputation:** median fitted only on each training fold.
- **Grouping:** all years from a country stay in the same fold.
- **Outcomes:** never imputed.

The output reports grouped-CV \(R^2\) and MAE for HALE and NCD mortality under both feature rules. These results should be interpreted as a robustness check on the missing-data eligibility decision.


```python
# ---------------------------------------------------------------------------
# SECTION 19B — STRICT-COVERAGE VS IMPUTATION-SENSITIVITY ML COMPARISON
# ---------------------------------------------------------------------------
def _sens_compare(target, feats, label):
    X, y, g, _ = build_xy(eng, target, feats)
    est = ExtraTreesRegressor(
        n_estimators=300, min_samples_leaf=2, random_state=SEED, n_jobs=-1
    )
    cv = cross_validate(
        PIPE(est), X, y, groups=g, cv=GroupKFold(5),
        scoring=["r2", "neg_mean_absolute_error"], n_jobs=1
    )
    return {
        "Outcome": target,
        "Feature rule": label,
        "n features": len(feats),
        "Mean grouped CV R²": cv["test_r2"].mean(),
        "R² SD": cv["test_r2"].std(),
        "Mean MAE": (-cv["test_neg_mean_absolute_error"]).mean(),
        "Worst-fold R²": cv["test_r2"].min(),
    }

SENSITIVITY_ML = pd.DataFrame([
    _sens_compare("HALE", FEATS, "Primary: >=70% observed"),
    _sens_compare("HALE", FEATS_SENS, "Sensitivity: >55% observed + fold imputation"),
    _sens_compare("NCDMortality", FEATS, "Primary: >=70% observed"),
    _sens_compare("NCDMortality", FEATS_SENS, "Sensitivity: >55% observed + fold imputation"),
])

banner("RQ3 MISSING-DATA SENSITIVITY — COMMON EXTRA TREES BENCHMARK")
display(SENSITIVITY_ML.round(4))

# Outcome-specific deltas: sensitivity minus primary.
_sens_delta_rows = []
for outcome in ["HALE", "NCDMortality"]:
    p = SENSITIVITY_ML[(SENSITIVITY_ML.Outcome == outcome) &
                       (SENSITIVITY_ML["Feature rule"].str.startswith("Primary"))].iloc[0]
    q = SENSITIVITY_ML[(SENSITIVITY_ML.Outcome == outcome) &
                       (SENSITIVITY_ML["Feature rule"].str.startswith("Sensitivity"))].iloc[0]
    _sens_delta_rows.append({
        "Outcome": outcome,
        "Δ grouped CV R² (sensitivity − primary)": q["Mean grouped CV R²"] - p["Mean grouped CV R²"],
        "Δ MAE (sensitivity − primary)": q["Mean MAE"] - p["Mean MAE"],
    })
SENSITIVITY_DELTA = pd.DataFrame(_sens_delta_rows)

print("\nSensitivity-only predictors retained by the broader rule:")
for v in SENS_ONLY:
    print(f"  {v}: {100*(1-eng[v].notna().mean()):.1f}% missing")

print("\nOutcome-specific sensitivity deltas:")
display(SENSITIVITY_DELTA.round(4))

_hale_d = SENSITIVITY_DELTA.loc[SENSITIVITY_DELTA.Outcome=="HALE"].iloc[0]
_ncd_d  = SENSITIVITY_DELTA.loc[SENSITIVITY_DELTA.Outcome=="NCDMortality"].iloc[0]
print("\nInterpretation:")
print(f"  HALE: sensitivity changes grouped-CV R² by {_hale_d.iloc[1]:+.4f} and MAE by {_hale_d.iloc[2]:+.4f}.")
print(f"  NCD : sensitivity changes grouped-CV R² by {_ncd_d.iloc[1]:+.4f} and MAE by {_ncd_d.iloc[2]:+.4f}.")
print("Primary RQ3 inference remains based on the >=70% observed feature set.")
print("The >55% set is a missing-data/feature-availability sensitivity analysis, not a replacement primary model.")
```

**Purpose:** Demonstrate why random row-level cross-validation is inappropriate for a longitudinal country panel and establish country-grouped cross-validation as the substantive validation strategy used for model comparison and generalisation assessment.

## **20  Model comparison under grouped cross-validation**

### Purpose
Compare nine candidate estimators on equal terms, from a mean-only baseline through regularised linear models to tree ensembles.

### Rationale
Model choice is an empirical question answered under the same country-grouped validation protocol for every candidate. Including a grand-mean baseline establishes a no-skill comparator, but the baseline is excluded from automatic best-model selection because it is not a trainable candidate model.

**XGBoost was included as an additional regularized boosted-tree benchmark because it can model nonlinearities and interactions in structured tabular data (Chen & Guestrin, 2016); however, model selection remains entirely empirical under the same grouped validation protocol.**

### Method
Nine estimators are evaluated under identical five-fold `GroupKFold` splits by `iso3`, using the primary Section 19 predictor set, HALE/NCD frames, preprocessing pipeline, and metrics. Median imputation and standardization remain inside the pipeline so they are refitted within each fold.
```python
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from xgboost import XGBRegressor

MODELS = {
 "Baseline (grand mean)": DummyRegressor(strategy="mean"),
 "Linear regression":     LinearRegression(),
 "Ridge":                 RidgeCV(alphas=np.logspace(-3,3,25)),
 "Lasso":                 LassoCV(cv=5, random_state=SEED, max_iter=5000),
 "Elastic net":           ElasticNetCV(cv=5, random_state=SEED, max_iter=5000),
 "Random forest":         RandomForestRegressor(n_estimators=250, min_samples_leaf=2,
                                                random_state=SEED, n_jobs=-1),
 "Extra trees":           ExtraTreesRegressor(n_estimators=250, min_samples_leaf=2,
                                              random_state=SEED, n_jobs=-1),
 "Gradient boosting":     GradientBoostingRegressor(random_state=SEED),
 "XGBoost":               XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=4,
                                      subsample=0.8, colsample_bytree=0.8,
                                      objective="reg:squarederror", random_state=SEED, n_jobs=-1),
}

# Fresh-estimator registry for later sections. The baseline is intentionally omitted.
MODEL_REGISTRY = {
 "Linear regression": lambda: LinearRegression(),
 "Ridge": lambda: RidgeCV(alphas=np.logspace(-3,3,25)),
 "Lasso": lambda: LassoCV(cv=5, random_state=SEED, max_iter=5000),
 "Elastic net": lambda: ElasticNetCV(cv=5, random_state=SEED, max_iter=5000),
 "Random forest": lambda: RandomForestRegressor(n_estimators=250, min_samples_leaf=2,
                                                 random_state=SEED, n_jobs=-1),
 "Extra trees": lambda: ExtraTreesRegressor(n_estimators=250, min_samples_leaf=2,
                                             random_state=SEED, n_jobs=-1),
 "Gradient boosting": lambda: GradientBoostingRegressor(random_state=SEED),
 "XGBoost": lambda: XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=4,
                                  subsample=0.8, colsample_bytree=0.8,
                                  objective="reg:squarederror", random_state=SEED, n_jobs=-1),
}

from sklearn.model_selection import cross_validate
def compare(X, y, g, tag):
    def run():
        rows = []
        for name, est in MODELS.items():
            cv = cross_validate(PIPE(est), X, y, groups=g, cv=GroupKFold(5),
                                scoring=["r2","neg_mean_absolute_error"], n_jobs=1)
            r2 = cv["test_r2"]; ma = -cv["test_neg_mean_absolute_error"]
            rows.append({"Model":name, "CV R²":r2.mean(), "R² SD":r2.std(),
                         "MAE":ma.mean(), "Worst fold R²":r2.min()})
        return pd.DataFrame(rows).sort_values("CV R²", ascending=False)
    return cached(f"cmp9_{tag}", run)

CMP_H = compare(Xh, yh, gh, "hale")
CMP_N = compare(Xn, yn, gn, "ncd")
banner("HALE — country-grouped 5-fold CV")
display(CMP_H.round(4))
banner("NCD MORTALITY — country-grouped 5-fold CV")
display(CMP_N.round(4))

fig, ax = plt.subplots(1, 2, figsize=(15,5))
for a, t, nm in zip(ax, [CMP_H, CMP_N], ["HALE","NCD mortality"]):
    t2 = t.sort_values("CV R²")
    a.barh(t2.Model, t2["CV R²"], xerr=t2["R² SD"], capsize=4,
           color=np.where(t2.Model=="Baseline (grand mean)", "#b2182b", "#2166ac"))
    a.set_xlabel("Grouped CV R²"); a.set_title(nm, fontweight="bold")
savefig("20_model_comparison")

candidate_hale = CMP_H[CMP_H["Model"] != "Baseline (grand mean)"].copy()
candidate_ncd = CMP_N[CMP_N["Model"] != "Baseline (grand mean)"].copy()
BEST_MODEL_HALE_NAME = candidate_hale.loc[candidate_hale["CV R²"].idxmax(), "Model"]
BEST_MODEL_NCD_NAME = candidate_ncd.loc[candidate_ncd["CV R²"].idxmax(), "Model"]
BEST_H = candidate_hale.loc[candidate_hale["Model"] == BEST_MODEL_HALE_NAME].iloc[0]
BEST_N = candidate_ncd.loc[candidate_ncd["Model"] == BEST_MODEL_NCD_NAME].iloc[0]
base_H = CMP_H.loc[CMP_H.Model=="Baseline (grand mean)","CV R²"].iloc[0]
base_N = CMP_N.loc[CMP_N.Model=="Baseline (grand mean)","CV R²"].iloc[0]
H3a = (BEST_H["CV R²"] - base_H > .10) and (BEST_N["CV R²"] - base_N > .10)

print(f"Best HALE model from grouped CV: {BEST_MODEL_HALE_NAME}")
print(f"Best NCD model from grouped CV:  {BEST_MODEL_NCD_NAME}")
banner("H3a")
print(f"HALE : best = {BEST_MODEL_HALE_NAME} at R² {BEST_H['CV R²']:.3f} vs baseline {base_H:.3f}")
print(f"NCD  : best = {BEST_MODEL_NCD_NAME} at R² {BEST_N['CV R²']:.3f} vs baseline {base_N:.3f}")
print(f"\nH3a₀ {'REJECTED' if H3a else 'not rejected'} — outcomes for unseen countries are "
      f"{'predictable well beyond the grand mean' if H3a else 'not clearly predictable'}.")
```

### Model Specification and Validation

**Model family:** Nine predictive candidates are evaluated for both HALE and premature NCD mortality.

**Purpose:** Select the strongest predictive algorithm empirically under country-disjoint validation.

**Outcomes:** `HALE` and `NCDMortality`.

## **20B Multi-model missing-data sensitivity analysis**

### Purpose

Test whether the conclusions from Section 20 depend on the stricter **≤30% missingness** primary feature rule or remain similar when the three partially observed capacity predictors are retained under the broader **>55% observed** sensitivity rule with fold-contained median imputation.

### Design

This analysis repeats the Section 20 model comparison under **both feature sets** for the same trainable candidate algorithms. The primary feature set remains the official RQ3 specification. The broader feature set is a sensitivity analysis only.

For each outcome and model family:

- **Primary specification:** Section 19 primary predictors, each with ≥70% observed data.
- **Sensitivity specification:** Section 19 sensitivity predictors, each with >55% observed data.
- **Missing predictors:** median-imputed inside each training/CV fold only.
- **Scaling:** applied inside the same pipeline.
- **Validation:** five-fold `GroupKFold` by country.
- **Metrics:** mean grouped-CV R², R² SD, MAE, worst-fold R².
- **Comparison:** sensitivity minus primary deltas in R² and MAE, plus model-rank stability.

The purpose is not to select a new primary model after seeing sensitivity results. It is to assess whether the algorithm ranking and predictive conclusions are robust to the missing-data feature-entry rule.

### Interpretive rule

If model rankings and performance are materially similar under both specifications, the RQ3 conclusion is robust to the stricter missingness rule. If some outcomes improve meaningfully when the three partially observed capacity measures are reintroduced, that outcome should be described as **more sensitive to feature availability/imputation strategy**.


```python
# ---------------------------------------------------------------------------
# SECTION 20B — MULTI-MODEL MISSING-DATA SENSITIVITY
# Repeat Section 20 candidate comparison under:
#   1) primary >=70% observed feature set
#   2) sensitivity >55% observed feature set
# ---------------------------------------------------------------------------

from sklearn.base import clone

# Reuse the Section 20 registry when available. Exclude the non-trainable grand-mean baseline
# from the paired model-family comparison because this sensitivity question concerns model
# performance under alternative predictor sets.
if "MODEL_REGISTRY" not in globals():
    raise RuntimeError("Section 20B requires MODEL_REGISTRY from Section 20.")

SENS_MODELS = [m for m in MODEL_REGISTRY.keys() if "baseline" not in m.lower()]

def _paired_model_sensitivity(target, feats_primary, feats_sens):
    rows = []
    feature_sets = [
        ("Primary >=70% observed", feats_primary),
        ("Sensitivity >55% observed + fold imputation", feats_sens),
    ]

    for feature_label, feats in feature_sets:
        X, y, groups, _ = build_xy(eng, target, feats)
        for model_name in SENS_MODELS:
            est = MODEL_REGISTRY[model_name]()
            cvres = cross_validate(
                PIPE(est),
                X, y,
                groups=groups,
                cv=GroupKFold(5),
                scoring=["r2", "neg_mean_absolute_error"],
                n_jobs=1
            )
            rows.append({
                "Outcome": target,
                "Model": model_name,
                "Feature specification": feature_label,
                "n features": len(feats),
                "Mean grouped CV R²": cvres["test_r2"].mean(),
                "R² SD": cvres["test_r2"].std(),
                "Mean MAE": (-cvres["test_neg_mean_absolute_error"]).mean(),
                "Worst-fold R²": cvres["test_r2"].min(),
            })
    return pd.DataFrame(rows)

MULTIMODEL_SENS_H = _paired_model_sensitivity("HALE", FEATS, FEATS_SENS)
MULTIMODEL_SENS_N = _paired_model_sensitivity("NCDMortality", FEATS, FEATS_SENS)
MULTIMODEL_SENSITIVITY = pd.concat([MULTIMODEL_SENS_H, MULTIMODEL_SENS_N], ignore_index=True)

def _paired_delta_table(df, outcome):
    sub = df[df["Outcome"] == outcome].copy()
    p = sub[sub["Feature specification"] == "Primary >=70% observed"].set_index("Model")
    s = sub[sub["Feature specification"] == "Sensitivity >55% observed + fold imputation"].set_index("Model")

    common = p.index.intersection(s.index)
    out = pd.DataFrame(index=common)
    out["Primary CV R²"] = p.loc[common, "Mean grouped CV R²"]
    out["Sensitivity CV R²"] = s.loc[common, "Mean grouped CV R²"]
    out["Δ R² (sens-primary)"] = out["Sensitivity CV R²"] - out["Primary CV R²"]
    out["Primary MAE"] = p.loc[common, "Mean MAE"]
    out["Sensitivity MAE"] = s.loc[common, "Mean MAE"]
    out["Δ MAE (sens-primary)"] = out["Sensitivity MAE"] - out["Primary MAE"]

    out["Primary rank"] = out["Primary CV R²"].rank(ascending=False, method="min").astype(int)
    out["Sensitivity rank"] = out["Sensitivity CV R²"].rank(ascending=False, method="min").astype(int)
    out["Rank change"] = out["Sensitivity rank"] - out["Primary rank"]
    out = out.reset_index().rename(columns={"index":"Model"})
    return out.sort_values("Primary rank")

MULTIMODEL_DELTA_HALE = _paired_delta_table(MULTIMODEL_SENSITIVITY, "HALE")
MULTIMODEL_DELTA_NCD = _paired_delta_table(MULTIMODEL_SENSITIVITY, "NCDMortality")

banner("SECTION 20B — MULTI-MODEL MISSING-DATA SENSITIVITY: HALE")
display(MULTIMODEL_DELTA_HALE.round(4))

banner("SECTION 20B — MULTI-MODEL MISSING-DATA SENSITIVITY: NCD MORTALITY")
display(MULTIMODEL_DELTA_NCD.round(4))

# Rank-stability diagnostics
def _rank_summary(t, outcome):
    same_winner = (
        t.sort_values("Primary rank").iloc[0]["Model"]
        == t.sort_values("Sensitivity rank").iloc[0]["Model"]
    )
    spearman = t["Primary rank"].corr(t["Sensitivity rank"], method="spearman")
    max_abs_dr2 = t["Δ R² (sens-primary)"].abs().max()
    mean_abs_dr2 = t["Δ R² (sens-primary)"].abs().mean()
    max_abs_dmae = t["Δ MAE (sens-primary)"].abs().max()
    return {
        "Outcome": outcome,
        "Same top-ranked model": bool(same_winner),
        "Spearman rank correlation": spearman,
        "Mean |ΔR²|": mean_abs_dr2,
        "Max |ΔR²|": max_abs_dr2,
        "Max |ΔMAE|": max_abs_dmae,
    }

MULTIMODEL_RANK_STABILITY = pd.DataFrame([
    _rank_summary(MULTIMODEL_DELTA_HALE, "HALE"),
    _rank_summary(MULTIMODEL_DELTA_NCD, "NCDMortality"),
])

banner("MODEL-RANK STABILITY ACROSS MISSING-DATA SPECIFICATIONS")
display(MULTIMODEL_RANK_STABILITY.round(4))

# Outcome-level best models under each feature specification
def _best_by_spec(df, outcome):
    z = df[df["Outcome"] == outcome].copy()
    return (z.sort_values(["Feature specification","Mean grouped CV R²"], ascending=[True,False])
              .groupby("Feature specification", as_index=False)
              .first()[["Feature specification","Model","Mean grouped CV R²","Mean MAE"]])

MULTIMODEL_WINNERS_HALE = _best_by_spec(MULTIMODEL_SENSITIVITY, "HALE")
MULTIMODEL_WINNERS_NCD = _best_by_spec(MULTIMODEL_SENSITIVITY, "NCDMortality")

print("\nTop model by feature specification — HALE")
display(MULTIMODEL_WINNERS_HALE.round(4))
print("\nTop model by feature specification — NCD mortality")
display(MULTIMODEL_WINNERS_NCD.round(4))

print("\nInterpretation:")
print("- Section 20 primary model selection remains authoritative.")
print("- Section 20B tests robustness of model-family rankings to the missing-data feature rule.")
print("- Positive ΔR² means the broader imputed feature set improved grouped-CV performance.")
print("- Negative ΔMAE means the broader imputed feature set reduced prediction error.")
```

### Result and interpretation

The multi-model sensitivity analysis confirms that the principal RQ3 model-family conclusion is stable across the two missing-data specifications.

For **HALE**, performance is essentially unchanged when the three partially observed capacity variables are reintroduced. Extra Trees remains the top-ranked model, with grouped-CV \(R^2\) changing only from **0.8100** under the primary 18-predictor specification to **0.8077** under the 21-predictor imputation sensitivity specification. Across the eight trainable model families, the mean absolute change in \(R^2\) is only **0.0016**, the maximum absolute change is **0.0034**, and the model-rank Spearman correlation is **0.976**. This provides strong evidence that HALE prediction is robust to the stricter 30%-missingness eligibility rule.

For **premature NCD mortality**, the broader feature set improves prediction more noticeably across model families. Extra Trees remains the top-ranked model, but grouped-CV \(R^2\) increases from **0.5996** to **0.6332** and MAE falls from **3.5421** to **3.4004**. Across model families, the mean absolute change in \(R^2\) is **0.0380**, with a maximum improvement of approximately **0.0708** for Random Forest. Despite these performance gains, the ranking remains highly stable (Spearman **0.976**) and Extra Trees remains first under both specifications.

### Decision

Retain the **18-predictor ≥70%-observed specification as the primary RQ3 analysis**. Treat the 21-predictor specification as sensitivity evidence only. The sensitivity analysis strengthens the conclusion that the selected model family is robust, while showing that **NCD prediction is more sensitive than HALE to feature availability and fold-contained imputation of Physicians, NursesMidwives, and HospitalBeds**.

## **21  Repeated grouped cross-validation**

### Purpose
Test whether the model ranking is stable or an artefact of one particular partition of countries.

### Rationale
A single five-fold split is one draw. With fewer than two hundred clusters, which countries land together can move a fold score materially, and a ranking based on one partition may not survive another. Reporting a single R-squared without its dispersion overstates precision.

### Method
The evaluation is repeated across several independent country permutations. The permutation is drawn once per repetition and applied to all rows, so a country's observations always move together — drawing per row would silently reintroduce exactly the leakage Section 19 exists to prevent.

### Expected output
The distribution of fold scores with mean, dispersion and range for both outcomes.

```python
import re

def repeated_cv(X, y, g, model_name, reps=5, tag=""):
    def run():
        scores = []
        for r in range(reps):
            rng = np.random.default_rng(SEED+r)
            u = pd.unique(g); perm = dict(zip(u, rng.permutation(len(u))))
            gp = np.array([perm[x] for x in g])
            est = MODEL_REGISTRY[model_name]()   # fresh estimator; cross_val_score clones per fold
            s = cross_val_score(PIPE(est), X, y, groups=gp, cv=GroupKFold(5), scoring="r2")
            scores.extend(s)
        return np.array(scores)
    key = re.sub(r"[^A-Za-z0-9]+","_",model_name.lower())
    return cached(f"rep_selected_{tag}_{key}", run)

RH = repeated_cv(Xh, yh, gh, BEST_MODEL_HALE_NAME, tag="hale")
RN = repeated_cv(Xn, yn, gn, BEST_MODEL_NCD_NAME, tag="ncd")
print(f"HALE {BEST_MODEL_HALE_NAME}: R² = {RH.mean():.3f} ± {RH.std():.3f} over {len(RH)} folds "
      f"[{RH.min():.3f}, {RH.max():.3f}]")
print(f"NCD  {BEST_MODEL_NCD_NAME}: R² = {RN.mean():.3f} ± {RN.std():.3f} over {len(RN)} folds "
      f"[{RN.min():.3f}, {RN.max():.3f}]")

fig, ax = plt.subplots(1, 2, figsize=(13,4.5))
for a, s, t, mn in zip(ax, [RH, RN], ["HALE","NCD mortality"],
                       [BEST_MODEL_HALE_NAME, BEST_MODEL_NCD_NAME]):
    a.hist(s, bins=14, color="#4393c3", edgecolor="white")
    a.axvline(s.mean(), c="#b2182b", lw=2, label=f"mean {s.mean():.3f}")
    a.set_xlabel("fold R²"); a.set_title(f"{t} — {mn}\nrepeated grouped folds", fontweight="bold")
    a.legend(fontsize=8)
savefig("21_repeated_cv")
print("\nFive grouped folds × five repetitions = 25 fold-level R² values per outcome.")
```

### Model Specification and Validation

**Model:** Repeated grouped cross-validation of the automatically selected Section 20 winner for each outcome.

**Selected algorithms:** Retrieved programmatically from Section 20; no algorithm name is hard-coded here.

**Validation design:** Five repetitions of five-fold country-grouped CV, producing exactly **25 fold-level \(R^2\) values per outcome**. This is not a single 25-fold CV. All rows from a country remain together within every split, and fresh estimator clones are fitted for each fold.

**Executed performance:** The mean, standard deviation, and range for each outcome are printed directly by the preceding code cell and should be treated as the authoritative values after `Restart Kernel and Run All`.

**Interpretation:** These repeated grouped-CV distributions are the primary estimates of expected generalisation across unseen-country partitions.


### Result
Performance is reported as a mean with a standard deviation across folds, and the spread is substantial.

### Interpretation
The dispersion quantifies uncertainty around the headline performance estimate and shows that predictive performance depends partly on which countries form each held-out fold.

### Decision
Treat repeated country-grouped cross-validation as the primary estimate of expected out-of-sample performance and report its dispersion rather than relying on a single split.

## **22  Locked train, validation and test evaluation**

### Purpose
Tune the algorithm selected in Section 20 using training countries only, then provide a separate fixed-partition robustness evaluation.

### Rationale
Algorithm selection and locked evaluation answer different questions. Section 20 selects the algorithm from country-grouped cross-validation. Section 22 does **not** re-select the algorithm from validation or test performance; it tunes only the already-selected algorithm and evaluates it on fixed held-out country partitions.

### Method
Countries remain partitioned 60% training / 20% validation / 20% test using the existing country-level split. Hyperparameter search is restricted to the training countries and uses model-specific `RandomizedSearchCV` with 30 sampled configurations and five-fold `GroupKFold` with training-country group labels. Validation and test countries do not enter hyperparameter selection.

**The algorithm was selected empirically in Section 20 using country-grouped cross-validation. Section 22 then tuned only that selected algorithm using grouped inner cross-validation restricted to the training countries, before evaluating the fitted model on fixed validation and test country partitions.**

### Expected output
For each outcome: selected algorithm, best hyperparameters, train/validation/test \(R^2\), MAE and RMSE, plus the train-to-test \(R^2\) gap.


```python
from sklearn.base import clone
from sklearn.linear_model import Ridge, Lasso, ElasticNet

def split_countries(g, seed=SEED):
    u = np.array(sorted(pd.unique(g)))
    rng = np.random.default_rng(seed); rng.shuffle(u)
    n = len(u); a, b = int(.6*n), int(.8*n)
    return set(u[:a]), set(u[a:b]), set(u[b:])

def make_selected_estimator(model_name):
    if model_name == "Extra trees":
        return ExtraTreesRegressor(random_state=SEED, n_jobs=-1)
    if model_name == "Random forest":
        return RandomForestRegressor(random_state=SEED, n_jobs=-1)
    if model_name == "Gradient boosting":
        return GradientBoostingRegressor(random_state=SEED)
    if model_name == "XGBoost":
        return XGBRegressor(objective="reg:squarederror", random_state=SEED, n_jobs=-1)
    if model_name == "Ridge":
        return Ridge()
    if model_name == "Lasso":
        return Lasso(max_iter=10000)
    if model_name == "Elastic net":
        return ElasticNet(max_iter=10000)
    if model_name == "Linear regression":
        return LinearRegression()
    raise ValueError(f"Unsupported selected model: {model_name}")

def get_param_distributions(model_name):
    if model_name in ("Extra trees","Random forest"):
        return {"m__n_estimators":[200,300,500],
                "m__max_depth":[None,8,12,16,24],
                "m__min_samples_split":[2,4,8],
                "m__min_samples_leaf":[1,2,4,6],
                "m__max_features":[0.3,0.5,0.7,1.0,"sqrt"]}
    if model_name == "Gradient boosting":
        return {"m__n_estimators":[100,200,300,500],
                "m__learning_rate":[.01,.03,.05,.1],
                "m__max_depth":[2,3,4,5],
                "m__min_samples_leaf":[1,2,4,6],
                "m__subsample":[.6,.8,1.0]}
    if model_name == "XGBoost":
        return {"m__n_estimators":[200,300,500,700],
                "m__learning_rate":[.01,.03,.05,.1],
                "m__max_depth":[2,3,4,5,6],
                "m__min_child_weight":[1,3,5,7],
                "m__subsample":[.6,.8,1.0],
                "m__colsample_bytree":[.5,.7,.9,1.0],
                "m__reg_alpha":[0,.01,.1,1.0],
                "m__reg_lambda":[.5,1.0,2.0,5.0]}
    if model_name == "Ridge":
        return {"m__alpha":np.logspace(-4,3,50)}
    if model_name == "Lasso":
        return {"m__alpha":np.logspace(-4,1,50)}
    if model_name == "Elastic net":
        return {"m__alpha":np.logspace(-4,1,30),
                "m__l1_ratio":np.linspace(.05,.95,19)}
    return {}

def evaluate_regression(model, X, y):
    pred = model.predict(X)
    return {"R²":r2_score(y,pred),
            "MAE":mean_absolute_error(y,pred),
            "RMSE":np.sqrt(mean_squared_error(y,pred))}

def locked_eval_selected(X, y, g, model_name, tag):
    def run():
        tr, va, te = split_countries(g)
        mtr, mva, mte = np.isin(g,list(tr)), np.isin(g,list(va)), np.isin(g,list(te))
        selected_pipeline = PIPE(make_selected_estimator(model_name))
        param_dist = get_param_distributions(model_name)
        if param_dist:
            search = RandomizedSearchCV(
                estimator=selected_pipeline, param_distributions=param_dist,
                n_iter=30, scoring="r2", cv=GroupKFold(5),
                random_state=SEED, n_jobs=1, refit=True
            )
            search.fit(X[mtr], y[mtr], groups=g[mtr])
            best_model, best_params = search.best_estimator_, search.best_params_
        else:
            best_model = selected_pipeline.fit(X[mtr], y[mtr])
            best_params = {}
        out = {"Selected algorithm":model_name, "Best params":best_params,
               "Train countries":len(tr), "Val countries":len(va), "Test countries":len(te)}
        for lbl,m in [("Train",mtr),("Validation",mva),("Test",mte)]:
            met = evaluate_regression(best_model, X[m], y[m])
            for k,v in met.items(): out[f"{lbl} {k}"] = v
            out[f"{lbl} n"] = int(m.sum())
        return out, best_model, (mtr,mva,mte)
    key = re.sub(r"[^A-Za-z0-9]+","_",model_name.lower())
    return cached(f"locked_selected_tune30_{tag}_{key}", run)

LH, BH_, MH_ = locked_eval_selected(Xh.values, yh.values, gh, BEST_MODEL_HALE_NAME, "hale")
LN, BN_, MN_ = locked_eval_selected(Xn.values, yn.values, gn, BEST_MODEL_NCD_NAME, "ncd")
FINAL_HALE_MODEL, FINAL_NCD_MODEL = BH_, BN_
FINAL_HALE_MODEL_NAME, FINAL_NCD_MODEL_NAME = BEST_MODEL_HALE_NAME, BEST_MODEL_NCD_NAME

def split_table(d, name):
    return pd.DataFrame([{"Outcome":name, "Algorithm":d["Selected algorithm"], "Split":s,
                          "n":d[f"{s} n"], "R²":d[f"{s} R²"],
                          "MAE":d[f"{s} MAE"], "RMSE":d[f"{s} RMSE"]}
                         for s in ["Train","Validation","Test"]])
SPL = pd.concat([split_table(LH,"HALE"), split_table(LN,"NCD mortality")])
display(SPL.round(4))
print(f"\nHALE selected algorithm: {FINAL_HALE_MODEL_NAME}")
print(f"HALE best params: {LH['Best params']}")
print(f"NCD selected algorithm:  {FINAL_NCD_MODEL_NAME}")
print(f"NCD best params: {LN['Best params']}")
gap_h = LH["Train R²"] - LH["Test R²"]; gap_n = LN["Train R²"] - LN["Test R²"]
print(f"\nTrain-to-test gap — HALE {gap_h:+.3f}, NCD {gap_n:+.3f}")
print("A large positive train-to-test gap is treated as a potential overfitting signal,")
print("not against a rigid universal cutoff.")
```

### Model Specification and Validation

**Selected algorithm:** Determined automatically in Section 20 for each outcome by the highest mean country-grouped CV \(R^2\), excluding the grand-mean baseline.

**Locked split:** 60% training countries / 20% validation countries / 20% test countries.

**Tuning:** Model-specific `RandomizedSearchCV` with **30 sampled configurations** and **five-fold `GroupKFold` restricted to the training countries only**, scored by \(R^2\). Validation and test countries are not used to choose the algorithm or hyperparameters.

**Selected hyperparameters and performance:** Generated directly by the preceding executable cell after `Restart Kernel and Run All`; they are intentionally not hard-coded here so this narrative cannot become stale if model selection or tuning results change.

It tunes only the Section 20 winner on training countries and then evaluates that tuned model on fixed validation and test country partitions. Repeated grouped CV from Section 21 remains the primary generalisation estimate; the fixed split is complementary evidence.


```python
fig, ax = plt.subplots(1, 3, figsize=(17,5))
w = .35; xs = np.arange(3)
ax[0].bar(xs-w/2, [LH[f"{s} R²"] for s in ["Train","Validation","Test"]], w, label="HALE", color="#2166ac")
ax[0].bar(xs+w/2, [LN[f"{s} R²"] for s in ["Train","Validation","Test"]], w, label="NCD", color="#b2182b")
ax[0].set_xticks(xs); ax[0].set_xticklabels(["Train","Validation","Test"])
ax[0].set_ylabel("R²"); ax[0].set_title("Performance across the locked splits", fontweight="bold")
ax[0].legend()

for a, (X, y, M, B, nm) in zip(ax[1:], [(Xh.values,yh.values,MH_,BH_,"HALE"),
                                        (Xn.values,yn.values,MN_,BN_,"NCD mortality")]):
    mte = M[2]; pr = B.predict(X[mte])
    a.scatter(y[mte], pr, s=18, alpha=.5, c="#2166ac", edgecolor="none")
    lo, hi = min(y[mte].min(), pr.min()), max(y[mte].max(), pr.max())
    a.plot([lo,hi],[lo,hi],"k--",lw=1.5)
    a.set_xlabel(f"actual {nm}"); a.set_ylabel(f"predicted {nm}")
    a.set_title(f"Held-out test countries\nR² = {r2_score(y[mte],pr):.3f}", fontweight="bold")
savefig("22_locked_splits")
```

### Result
The selected models retain positive performance on the fixed validation and test country partitions, although the fixed-partition scores differ materially from the repeated grouped-CV averages.

Because the algorithm was selected before Section 22 and hyperparameters were chosen using training countries only, neither validation nor test performance is used to re-select the algorithm. The contrast between validation and test scores illustrates sensitivity to which countries happen to occupy a fixed partition.

## **22A  NCD overfitting refinement and regularized challenger comparison**

### Purpose
Add a targeted second-stage robustness analysis for premature NCD mortality because the locked Extra Trees fit can achieve near-perfect training performance while generalising materially less well to held-out countries.

### Why this additional step is methodologically justified
The purpose is **not** to search repeatedly until the test score improves. That would itself increase model-selection bias. Instead, the original Section 20 model-family choice is retained and only a **more strongly regularized Extra Trees challenger** is evaluated. Extremely randomized trees expose a tunable bias–variance trade-off through tree depth, leaf size, split size, and feature randomisation (Geurts, Ernst, & Wehenkel, 2006). In regression trees, larger terminal leaves and shallower trees smooth the fitted response, while restricting the feature subset increases randomisation and can reduce variance.

Model-selection criteria can themselves be overfit when many configurations are searched and the same validation signal is repeatedly optimized. Cawley and Talbot (2010) therefore motivate conservative model selection and separation of tuning from performance evaluation. This section responds by:

1. keeping the **Extra Trees model family fixed** rather than reopening the nine-model competition;
2. restricting the challenger search to a **stronger-regularization parameter space**;
3. tuning only on the original training countries with five-fold `GroupKFold`;
4. applying a **one-standard-error parsimony rule** inside the training-country search, choosing a simpler configuration whose mean grouped-CV \(R^2\) is within one standard error of the best searched configuration;
5. comparing the original tuned model and the conservative challenger on the same train/validation/test country partitions;
6. allowing the challenger to replace the original NCD model **only if the train-to-validation gap becomes smaller and at least one validation metric improves**. The automated adoption rule does not use the test metric.

### Important interpretive boundary
This is a **post-hoc robustness/refinement analysis**, added after observing a large train-to-held-out gap. The test-set comparison is therefore descriptive rather than a pristine untouched final test. Repeated country-grouped validation remains the primary generalisation evidence.

### Expected output
A side-by-side comparison of the original NCD Extra Trees model and the stronger-regularized challenger, including train/validation/test \(R^2\), MAE, RMSE, train-to-validation and train-to-test gaps, the selected conservative hyperparameters, and an explicit keep/adopt decision.


```python
# ---------------------------------------------------------------------------
# SECTION 22A — NCD overfitting refinement
# Stronger regularisation within the already-selected Extra Trees family.
# The automated adoption rule uses TRAIN + VALIDATION only, never TEST.
# ---------------------------------------------------------------------------
from sklearn.model_selection import ParameterSampler

# Preserve the original Section 22 NCD objects for a transparent comparison.
LN_ORIGINAL = dict(LN)
BN_ORIGINAL = BN_
MN_ORIGINAL = MN_
ORIGINAL_NCD_MODEL_NAME = FINAL_NCD_MODEL_NAME

# Same fixed country partitions as Section 22.
mtr_n, mva_n, mte_n = MN_ORIGINAL

# Stronger-regularisation search space.
# Compared with Section 22, this removes unrestricted depth, excludes leaves of 1–2,
# requires larger split/leaf support, and reduces the feature fraction considered
# at each split. These are variance-control choices rather than a new model family.
NCD_REG_PARAM_DIST = {
    "m__n_estimators": [400, 600, 800],
    "m__max_depth": [6, 8, 10, 12, 14],
    "m__min_samples_split": [6, 10, 16, 24, 32],
    "m__min_samples_leaf": [3, 4, 6, 8, 10, 12],
    "m__max_features": [0.3, 0.5, 0.7, "sqrt", "log2"],
    "m__bootstrap": [False, True],
    "m__ccp_alpha": [0.0, 0.0005, 0.001, 0.005],
}

def _complexity_key(params):
    """Lower tuple = simpler/more regularized model."""
    depth = params["m__max_depth"]
    depth_rank = 10_000 if depth is None else depth
    mf = params["m__max_features"]
    if mf == "sqrt":
        mf_rank = np.sqrt(len(FEATS)) / len(FEATS)
    elif mf == "log2":
        mf_rank = np.log2(len(FEATS)) / len(FEATS)
    else:
        mf_rank = float(mf)
    # Prefer shallower trees, larger leaves/splits, fewer features, and fewer trees
    # when predictive performance is statistically indistinguishable.
    return (
        depth_rank,
        -int(params["m__min_samples_leaf"]),
        -int(params["m__min_samples_split"]),
        mf_rank,
        int(params["m__n_estimators"]),
    )

def fit_regularized_ncd_challenger():
    challenger = PIPE(ExtraTreesRegressor(random_state=SEED, n_jobs=-1))

    # A larger search than the original 30 draws, but in a deliberately narrower,
    # strongly regularized space. Search remains confined to training countries.
    search = RandomizedSearchCV(
        challenger,
        param_distributions=NCD_REG_PARAM_DIST,
        n_iter=40,
        scoring={"r2": "r2", "mae": "neg_mean_absolute_error"},
        refit="r2",
        cv=GroupKFold(5),
        random_state=SEED + 220,
        n_jobs=1,
        return_train_score=True,
    )
    search.fit(Xn.values[mtr_n], yn.values[mtr_n], groups=gn[mtr_n])

    cvres = pd.DataFrame(search.cv_results_)
    best_i = int(cvres["mean_test_r2"].idxmax())
    best_mean = float(cvres.loc[best_i, "mean_test_r2"])
    best_se = float(cvres.loc[best_i, "std_test_r2"]) / np.sqrt(5)

    # One-standard-error rule: among configurations statistically close to the
    # best mean grouped-CV R², prefer the simpler/more regularized configuration.
    eligible = cvres[cvres["mean_test_r2"] >= (best_mean - best_se)].copy()
    eligible["_simplicity"] = eligible["params"].map(_complexity_key)
    chosen_i = eligible.sort_values("_simplicity").index[0]
    chosen_params = dict(cvres.loc[chosen_i, "params"])

    conservative = PIPE(ExtraTreesRegressor(random_state=SEED, n_jobs=-1))
    conservative.set_params(**chosen_params)
    conservative.fit(Xn.values[mtr_n], yn.values[mtr_n])

    summary = {
        "Best searched mean inner-CV R²": best_mean,
        "Best searched inner-CV SE": best_se,
        "Chosen one-SE mean inner-CV R²": float(cvres.loc[chosen_i, "mean_test_r2"]),
        "Chosen one-SE mean inner-CV MAE": float(-cvres.loc[chosen_i, "mean_test_mae"]),
        "Eligible configs within 1 SE": int(len(eligible)),
        "Chosen params": chosen_params,
    }
    return conservative, cvres, summary

NCD_REG_MODEL, NCD_REG_SEARCH_RESULTS, NCD_REG_SUMMARY = fit_regularized_ncd_challenger()

def model_split_row(label, model):
    rows = []
    for split_name, mask in [("Train", mtr_n), ("Validation", mva_n), ("Test", mte_n)]:
        met = evaluate_regression(model, Xn.values[mask], yn.values[mask])
        rows.append({
            "Model": label,
            "Split": split_name,
            "n": int(mask.sum()),
            "R²": met["R²"],
            "MAE": met["MAE"],
            "RMSE": met["RMSE"],
        })
    return rows

NCD_OVERFIT_COMPARE = pd.DataFrame(
    model_split_row("Original tuned Extra Trees", BN_ORIGINAL)
    + model_split_row("Regularized one-SE Extra Trees", NCD_REG_MODEL)
)

# Add explicit generalisation-gap diagnostics.
for label in NCD_OVERFIT_COMPARE["Model"].unique():
    tr_r2 = NCD_OVERFIT_COMPARE.query("Model == @label and Split == 'Train'")["R²"].iloc[0]
    va_r2 = NCD_OVERFIT_COMPARE.query("Model == @label and Split == 'Validation'")["R²"].iloc[0]
    te_r2 = NCD_OVERFIT_COMPARE.query("Model == @label and Split == 'Test'")["R²"].iloc[0]
    NCD_OVERFIT_COMPARE.loc[NCD_OVERFIT_COMPARE.Model == label, "Train−Validation R² gap"] = tr_r2 - va_r2
    NCD_OVERFIT_COMPARE.loc[NCD_OVERFIT_COMPARE.Model == label, "Train−Test R² gap"] = tr_r2 - te_r2

banner("SECTION 22A — NCD OVERFITTING REFINEMENT")
print("Original Section 22 NCD parameters:")
print(LN_ORIGINAL["Best params"])
print("\nRegularized one-standard-error challenger:")
print(NCD_REG_SUMMARY["Chosen params"])
print(f"\nBest searched inner-CV R² = {NCD_REG_SUMMARY['Best searched mean inner-CV R²']:.3f}")
print(f"One-SE threshold           = "
      f"{NCD_REG_SUMMARY['Best searched mean inner-CV R²'] - NCD_REG_SUMMARY['Best searched inner-CV SE']:.3f}")
print(f"Chosen conservative CV R² = {NCD_REG_SUMMARY['Chosen one-SE mean inner-CV R²']:.3f}")
print(f"Configurations within 1 SE = {NCD_REG_SUMMARY['Eligible configs within 1 SE']}")
display(NCD_OVERFIT_COMPARE.round(4))

# Validation-only adoption rule.
def _get(model_label, split, metric):
    return NCD_OVERFIT_COMPARE.query(
        "Model == @model_label and Split == @split"
    )[metric].iloc[0]

orig_label = "Original tuned Extra Trees"
reg_label = "Regularized one-SE Extra Trees"

orig_gap_val = _get(orig_label, "Train", "R²") - _get(orig_label, "Validation", "R²")
reg_gap_val = _get(reg_label, "Train", "R²") - _get(reg_label, "Validation", "R²")
gap_reduced = reg_gap_val < orig_gap_val

val_r2_improved = _get(reg_label, "Validation", "R²") > _get(orig_label, "Validation", "R²")
val_mae_improved = _get(reg_label, "Validation", "MAE") < _get(orig_label, "Validation", "MAE")
validation_improved = val_r2_improved or val_mae_improved

ADOPT_NCD_REGULARIZED = bool(gap_reduced and validation_improved)

print("\nAutomated adoption rule (TRAIN + VALIDATION only):")
print(f"  Train→validation gap reduced: {gap_reduced}")
print(f"  Validation R² improved:       {val_r2_improved}")
print(f"  Validation MAE improved:      {val_mae_improved}")
print(f"  Decision: {'ADOPT regularized challenger' if ADOPT_NCD_REGULARIZED else 'KEEP original tuned model'}")
print("  Test metrics are displayed for transparency but are NOT used by this rule.")

# Update the final NCD object only if the validation-only rule supports the challenger.
if ADOPT_NCD_REGULARIZED:
    FINAL_NCD_MODEL = NCD_REG_MODEL
    BN_ = NCD_REG_MODEL

    refined = {
        "Selected algorithm": "Extra trees — regularized one-SE refinement",
        "Best params": NCD_REG_SUMMARY["Chosen params"],
        "Train countries": LN_ORIGINAL["Train countries"],
        "Val countries": LN_ORIGINAL["Val countries"],
        "Test countries": LN_ORIGINAL["Test countries"],
    }
    for split_name, mask in [("Train", mtr_n), ("Validation", mva_n), ("Test", mte_n)]:
        met = evaluate_regression(FINAL_NCD_MODEL, Xn.values[mask], yn.values[mask])
        for k, v in met.items():
            refined[f"{split_name} {k}"] = v
        refined[f"{split_name} n"] = int(mask.sum())
    LN = refined
    FINAL_NCD_MODEL_NAME = "Extra trees — regularized one-SE refinement"
else:
    FINAL_NCD_MODEL = BN_ORIGINAL
    BN_ = BN_ORIGINAL
    LN = LN_ORIGINAL
    FINAL_NCD_MODEL_NAME = ORIGINAL_NCD_MODEL_NAME

# Update the locked-split table used downstream/export so it represents the
# final validation-supported model, while preserving the comparison table above.
SPL = pd.concat([split_table(LH, "HALE"), split_table(LN, "NCD mortality")], ignore_index=True)

# Secondary descriptive grouped-CV robustness comparison with FIXED hyperparameters.
# This is not used to override the validation-based adoption decision.
def fixed_grouped_cv_summary(label, model):
    cv = cross_validate(
        clone(model), Xn, yn, groups=gn, cv=GroupKFold(5),
        scoring={"r2": "r2", "mae": "neg_mean_absolute_error"},
        n_jobs=1
    )
    return {
        "Model": label,
        "Grouped CV R²": cv["test_r2"].mean(),
        "R² SD": cv["test_r2"].std(),
        "MAE": (-cv["test_mae"]).mean(),
        "Worst fold R²": cv["test_r2"].min(),
    }

NCD_REFINEMENT_CV = pd.DataFrame([
    fixed_grouped_cv_summary("Original tuned Extra Trees", BN_ORIGINAL),
    fixed_grouped_cv_summary("Regularized one-SE Extra Trees", NCD_REG_MODEL),
])
banner("SECONDARY GROUPED-CV ROBUSTNESS COMPARISON")
display(NCD_REFINEMENT_CV.round(4))

# Compact visual comparison.
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8))
wide = NCD_OVERFIT_COMPARE.pivot(index="Split", columns="Model", values="R²").loc[["Train","Validation","Test"]]
wide.plot(kind="bar", ax=ax[0])
ax[0].set_ylabel("R²")
ax[0].set_title("NCD: original vs stronger regularisation", fontweight="bold")
ax[0].tick_params(axis="x", rotation=0)

cvp = NCD_REFINEMENT_CV.set_index("Model")
ax[1].barh(cvp.index, cvp["Grouped CV R²"], xerr=cvp["R² SD"], capsize=4)
ax[1].set_xlabel("Grouped CV R²")
ax[1].set_title("Fixed-hyperparameter grouped-CV check", fontweight="bold")
savefig("22A_ncd_overfitting_refinement")

print(f"\nFINAL NCD MODEL USED DOWNSTREAM: {FINAL_NCD_MODEL_NAME}")
print(f"Final locked-test R² = {LN['Test R²']:.3f}; MAE = {LN['Test MAE']:.3f}; RMSE = {LN['Test RMSE']:.3f}")
```

### Result and interpretation

The preceding cell prints the executed comparison and the automated validation-only decision. The main diagnostic is the **train-to-validation \(R^2\) gap**, not training \(R^2\) by itself. A near-perfect tree-ensemble training fit is not automatically a failure if grouped validation remains stable, but a large gap signals high variance and justifies a stronger-regularization sensitivity analysis.

The challenger is deliberately conservative. `max_depth` is bounded, `min_samples_leaf` and `min_samples_split` are increased, and `max_features` is restricted. These choices trade some training fit for smoother, less variable predictions. The one-standard-error rule further avoids choosing the numerically best inner-CV configuration when a simpler configuration performs within the uncertainty of that estimate.

**Decision rule:** the regularized challenger is adopted only when it reduces the train-to-validation gap **and** improves at least one validation metric (\(R^2\) or MAE). Otherwise the original Section 22 NCD model is retained. The test metric is shown only as a descriptive robustness comparison and does not enter the automated adoption rule.

This refinement does not reopen H3a or the nine-model comparison. Section 20 still establishes Extra Trees as the selected model family and Section 21 remains the primary repeated grouped-CV estimate of unseen-country generalisation. Section 22A is an additional robustness step aimed specifically at variance control for NCD mortality.

## **22B  NCD generalisation model challenge**

### Purpose
Perform one final NCD-specific model challenge because Section 22A reduced the train-to-held-out gap but weakened broader grouped-cross-validation performance. The objective is **not to force training R² downward**, but to identify the model with the strongest and most stable country-level generalisation.

### Rationale
Four candidates are compared under the same country-grouped framework:

1. **Original tuned Extra Trees** — current performance benchmark.
2. **Regularized Extra Trees** — lower-variance tree benchmark.
3. **HistGradientBoostingRegressor** — shrinkage-based boosting with bounded complexity, L2 regularisation and early stopping.
4. **Elastic Net** — a lower-variance linear benchmark combining L1 and L2 penalties.

The comparison prioritises mean grouped-CV R², grouped-CV MAE, worst-fold R², fold dispersion, and train-to-validation gap.

### Selection rule
The final NCD model is selected using grouped generalisation first:

1. The candidate must reduce the train-to-validation R² gap relative to the original tuned Extra Trees.
2. Its mean grouped-CV R² must be within **0.02** of the best candidate.
3. Among eligible candidates, choose the lowest grouped-CV MAE.
4. Validation performance is used only as a tie-breaker.
5. Test metrics are reported only after selection and do **not** enter the automated choice.

The 0.02 tolerance is a pragmatic post-hoc parsimony margin, not an inferential hypothesis threshold.

### Interpretive boundary
This is a finite post-hoc robustness analysis motivated by the NCD generalisation gap. The primary RQ3 evidence remains repeated country-grouped validation. After this section, the NCD model is frozen; no additional tuning should be undertaken merely to improve the reported test result.

### Expected output
A side-by-side comparison of all four candidates and one transparent final NCD model choice.


```python
# ---------------------------------------------------------------------------
# SECTION 22B — NCD GENERALISATION MODEL CHALLENGE
# Final finite comparison. Test metrics are NOT used for selection.
# ---------------------------------------------------------------------------
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.linear_model import ElasticNet

NCD_ORIG_MODEL = BN_ORIGINAL
NCD_REG_ET_MODEL = NCD_REG_MODEL

# HistGradientBoosting: lower-variance boosting with shrinkage, bounded leaves,
# L2 regularisation and early stopping.
HGB_PARAM_DIST = {
    "m__learning_rate": [0.02, 0.04, 0.06, 0.08],
    "m__max_iter": [200, 400, 600],
    "m__max_leaf_nodes": [7, 15, 31],
    "m__max_depth": [3, 5, 7, None],
    "m__min_samples_leaf": [20, 30, 40, 60],
    "m__l2_regularization": [0.1, 1.0, 5.0, 10.0],
}

HGB_BASE = PIPE(HistGradientBoostingRegressor(
    random_state=SEED,
    early_stopping=True,
    validation_fraction=0.15,
    n_iter_no_change=25
))

HGB_SEARCH = RandomizedSearchCV(
    HGB_BASE,
    param_distributions=HGB_PARAM_DIST,
    n_iter=35,
    scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
    refit="r2",
    cv=GroupKFold(5),
    random_state=SEED+221,
    n_jobs=1,
    return_train_score=True
)
HGB_SEARCH.fit(Xn.values[mtr_n], yn.values[mtr_n], groups=gn[mtr_n])
HGB_MODEL = HGB_SEARCH.best_estimator_
HGB_BEST_PARAMS = HGB_SEARCH.best_params_

# Elastic Net: regularized linear benchmark.
EN_PARAM_DIST = {
    "m__alpha": np.logspace(-3, 1.2, 30),
    "m__l1_ratio": [0.05, 0.1, 0.2, 0.4, 0.6, 0.8, 0.95],
    "m__max_iter": [20000],
}

EN_BASE = PIPE(ElasticNet(random_state=SEED))
EN_SEARCH = RandomizedSearchCV(
    EN_BASE,
    param_distributions=EN_PARAM_DIST,
    n_iter=35,
    scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
    refit="r2",
    cv=GroupKFold(5),
    random_state=SEED+222,
    n_jobs=1,
    return_train_score=True
)
EN_SEARCH.fit(Xn.values[mtr_n], yn.values[mtr_n], groups=gn[mtr_n])
EN_MODEL = EN_SEARCH.best_estimator_
EN_BEST_PARAMS = EN_SEARCH.best_params_

NCD_CHALLENGERS = {
    "Original tuned Extra Trees": NCD_ORIG_MODEL,
    "Regularized Extra Trees": NCD_REG_ET_MODEL,
    "HistGradientBoosting": HGB_MODEL,
    "Elastic Net": EN_MODEL,
}

def grouped_cv_full(label, model):
    cv = cross_validate(
        clone(model), Xn, yn, groups=gn, cv=GroupKFold(5),
        scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
        return_train_score=True, n_jobs=1
    )
    return {
        "Model": label,
        "Grouped CV R²": float(np.mean(cv["test_r2"])),
        "R² SD": float(np.std(cv["test_r2"])),
        "Worst fold R²": float(np.min(cv["test_r2"])),
        "Grouped CV MAE": float(np.mean(-cv["test_mae"])),
        "Mean train CV R²": float(np.mean(cv["train_r2"])),
        "CV train−test gap": float(np.mean(cv["train_r2"]) - np.mean(cv["test_r2"])),
    }

def split_metrics(label, model):
    row = {"Model": label}
    for split_name, mask in [("Train",mtr_n),("Validation",mva_n),("Test",mte_n)]:
        met = evaluate_regression(model, Xn.values[mask], yn.values[mask])
        row[f"{split_name} R²"] = met["R²"]
        row[f"{split_name} MAE"] = met["MAE"]
        row[f"{split_name} RMSE"] = met["RMSE"]
    row["Train−Validation gap"] = row["Train R²"] - row["Validation R²"]
    row["Train−Test gap"] = row["Train R²"] - row["Test R²"]
    return row

NCD_GENERALIZATION_CV = pd.DataFrame(
    [grouped_cv_full(k,v) for k,v in NCD_CHALLENGERS.items()]
)
NCD_GENERALIZATION_SPLITS = pd.DataFrame(
    [split_metrics(k,v) for k,v in NCD_CHALLENGERS.items()]
)
NCD_GENERALIZATION_COMPARE = NCD_GENERALIZATION_CV.merge(
    NCD_GENERALIZATION_SPLITS, on="Model", how="left"
)

best_cv = NCD_GENERALIZATION_COMPARE["Grouped CV R²"].max()
orig_gap = float(
    NCD_GENERALIZATION_COMPARE.loc[
        NCD_GENERALIZATION_COMPARE["Model"]=="Original tuned Extra Trees",
        "Train−Validation gap"
    ].iloc[0]
)

eligible = NCD_GENERALIZATION_COMPARE[
    (NCD_GENERALIZATION_COMPARE["Train−Validation gap"] < orig_gap) &
    (NCD_GENERALIZATION_COMPARE["Grouped CV R²"] >= best_cv - 0.02)
].copy()

if eligible.empty:
    chosen_row = NCD_GENERALIZATION_COMPARE.sort_values(
        ["Grouped CV R²","Grouped CV MAE"], ascending=[False,True]
    ).iloc[0]
else:
    chosen_row = eligible.sort_values(
        ["Grouped CV MAE","Validation R²","Validation MAE"],
        ascending=[True,False,True]
    ).iloc[0]

NCD_FINAL_CHOICE = chosen_row["Model"]
FINAL_NCD_MODEL = NCD_CHALLENGERS[NCD_FINAL_CHOICE]
BN_ = FINAL_NCD_MODEL

final_metrics = {
    "Selected algorithm": NCD_FINAL_CHOICE,
    "Train countries": LN_ORIGINAL["Train countries"],
    "Val countries": LN_ORIGINAL["Val countries"],
    "Test countries": LN_ORIGINAL["Test countries"],
}
for split_name, mask in [("Train",mtr_n),("Validation",mva_n),("Test",mte_n)]:
    met = evaluate_regression(FINAL_NCD_MODEL, Xn.values[mask], yn.values[mask])
    for k,v in met.items():
        final_metrics[f"{split_name} {k}"] = v
    final_metrics[f"{split_name} n"] = int(mask.sum())

if NCD_FINAL_CHOICE == "Original tuned Extra Trees":
    final_metrics["Best params"] = LN_ORIGINAL["Best params"]
elif NCD_FINAL_CHOICE == "Regularized Extra Trees":
    final_metrics["Best params"] = NCD_REG_SUMMARY["Chosen params"]
elif NCD_FINAL_CHOICE == "HistGradientBoosting":
    final_metrics["Best params"] = HGB_BEST_PARAMS
else:
    final_metrics["Best params"] = EN_BEST_PARAMS

LN = final_metrics
FINAL_NCD_MODEL_NAME = NCD_FINAL_CHOICE
SPL = pd.concat([split_table(LH, "HALE"), split_table(LN, "NCD mortality")], ignore_index=True)

banner("SECTION 22B — NCD GENERALISATION MODEL CHALLENGE")
display(
    NCD_GENERALIZATION_COMPARE[
        ["Model","Grouped CV R²","R² SD","Worst fold R²","Grouped CV MAE",
         "Mean train CV R²","CV train−test gap","Train R²","Validation R²",
         "Test R²","Train−Validation gap","Train−Test gap"]
    ].sort_values("Grouped CV R²", ascending=False).round(4)
)

print("\nTraining-country tuning parameters:")
print("HistGradientBoosting:", HGB_BEST_PARAMS)
print("Elastic Net:", EN_BEST_PARAMS)

print("\nSelection logic:")
print(f"Best grouped CV R² = {best_cv:.3f}")
print(f"Original Extra Trees train→validation gap = {orig_gap:.3f}")
print(f"Eligible candidates = {len(eligible)}")
print(f"FINAL NCD MODEL = {NCD_FINAL_CHOICE}")
print("Test metrics were not used for model selection.")

print(
    f"\nFinal NCD locked metrics: Train R²={LN['Train R²']:.3f}, "
    f"Validation R²={LN['Validation R²']:.3f}, Test R²={LN['Test R²']:.3f}, "
    f"Test MAE={LN['Test MAE']:.3f}, Test RMSE={LN['Test RMSE']:.3f}"
)

fig, ax = plt.subplots(figsize=(10,5))
plot_df = NCD_GENERALIZATION_COMPARE.sort_values("Grouped CV R²")
ax.barh(plot_df["Model"], plot_df["Grouped CV R²"], xerr=plot_df["R² SD"], capsize=4)
ax.set_xlabel("Grouped CV R²")
ax.set_title("NCD generalisation challenge: grouped-CV performance", fontweight="bold")
savefig("22B_ncd_generalisation_challenge")

fig, ax = plt.subplots(figsize=(10,5))
gap_df = NCD_GENERALIZATION_COMPARE.sort_values("Train−Validation gap")
ax.barh(gap_df["Model"], gap_df["Train−Validation gap"])
ax.set_xlabel("Train − validation R² gap")
ax.set_title("NCD generalisation challenge: overfitting gap", fontweight="bold")
savefig("22B_ncd_overfitting_gap")

print(f"\nFINAL NCD MODEL USED DOWNSTREAM: {FINAL_NCD_MODEL_NAME}")
```

### Result and interpretation

Read the executed comparison in this order: **grouped-CV R²**, grouped-CV MAE, worst-fold R² and dispersion, then the train-to-validation gap. A smaller train-to-validation gap is desirable only when it does not materially sacrifice grouped-country generalisation.

### Final decision principle
The selected model should balance strong unseen-country performance, lower prediction error, stability across folds, and a reduced gap relative to the original Extra Trees fit. The test set is displayed only after selection.

After this section, the model is frozen. Further tuning rounds should not be added simply to improve the test statistic because repeated adaptation to the same evaluation data would increase selection bias.

### Additional methodological references
Friedman (2001) provides the foundation for gradient boosting and shrinkage-based complexity control. Zou and Hastie (2005) developed Elastic Net regularisation. Cawley and Talbot (2010) explain why excessive model-selection searching can itself create optimistic bias.


## **22C  NCD diagnostic refinement: feature stability, dataset shift, and residual structure**

### Purpose
Determine **why** premature NCD mortality generalises less strongly than HALE before attempting any further model tuning. A large train-to-held-out gap can reflect model variance, but it can also arise when the predictor distribution, missingness pattern, or predictor–outcome relationship differs across country groups.

### Best-practice rationale
Country is the sampling unit for predictive generalisation in this study, so diagnostics must preserve country grouping. Roberts et al. (2017) show that ignoring hierarchical or grouped dependence can underestimate predictive error. Dataset-shift literature likewise emphasises that model performance can deteriorate when the distribution of predictors differs between model-development and deployment samples (Moreno-Torres et al., 2012). Finally, repeated model searching against the same validation/test information can itself overfit the model-selection process (Cawley & Talbot, 2010).

Accordingly, this section does **diagnosis first** and limits the final refinement to one transparent, training-derived feature-stability sensitivity analysis.

### Diagnostic questions
1. **Covariate shift:** Which predictors have the largest standardised mean differences between training and validation/test countries?
2. **Missingness shift:** Are some predictors substantially more incomplete in held-out countries?
3. **Feature stability:** Which predictors contribute positive permutation importance consistently across country-grouped folds?
4. **Residual structure:** Are the largest NCD errors concentrated in particular parts of the outcome distribution or country groups?
5. **Stable-feature sensitivity:** Does a model restricted to repeatedly useful predictors maintain generalisation while reducing variance?

### Guardrails
- Stable features are identified **only from training countries** using grouped folds.
- The stable-feature sensitivity model is tuned **only within training countries**.
- The locked test is descriptive because earlier post-hoc rounds have already inspected it.
- This section does not create another open-ended model search. After the comparison, the model-development process is frozen.


```python
# ---------------------------------------------------------------------------
# SECTION 22C — NCD DIAGNOSTIC REFINEMENT
# Diagnose shift/stability first; one finite stable-feature sensitivity analysis.
# ---------------------------------------------------------------------------
from sklearn.inspection import permutation_importance

# ------------------------- 1. Covariate shift -------------------------------
def _safe_smd(a, b):
    a = pd.to_numeric(pd.Series(a), errors="coerce").dropna()
    b = pd.to_numeric(pd.Series(b), errors="coerce").dropna()
    if len(a) < 2 or len(b) < 2:
        return np.nan
    pooled = np.sqrt((a.var(ddof=1) + b.var(ddof=1)) / 2)
    return 0.0 if pooled == 0 else (b.mean() - a.mean()) / pooled

shift_rows = []
for feat in FEATS:
    tr = Xn.loc[mtr_n, feat]
    va = Xn.loc[mva_n, feat]
    te = Xn.loc[mte_n, feat]
    shift_rows.append({
        "Feature": feat,
        "SMD validation vs train": _safe_smd(tr, va),
        "SMD test vs train": _safe_smd(tr, te),
        "Train missing %": 100*tr.isna().mean(),
        "Validation missing %": 100*va.isna().mean(),
        "Test missing %": 100*te.isna().mean(),
    })

NCD_SHIFT_DIAGNOSTICS = pd.DataFrame(shift_rows)
NCD_SHIFT_DIAGNOSTICS["Max |SMD|"] = NCD_SHIFT_DIAGNOSTICS[
    ["SMD validation vs train","SMD test vs train"]
].abs().max(axis=1)
NCD_SHIFT_DIAGNOSTICS["Max missingness Δ pp"] = pd.concat([
    (NCD_SHIFT_DIAGNOSTICS["Validation missing %"]-NCD_SHIFT_DIAGNOSTICS["Train missing %"]).abs(),
    (NCD_SHIFT_DIAGNOSTICS["Test missing %"]-NCD_SHIFT_DIAGNOSTICS["Train missing %"]).abs()
], axis=1).max(axis=1)

banner("SECTION 22C — NCD DATASET-SHIFT DIAGNOSTICS")
print("Largest predictor-distribution shifts (absolute standardized mean difference):")
display(NCD_SHIFT_DIAGNOSTICS.sort_values("Max |SMD|", ascending=False).head(12).round(3))

print("\nLargest missingness shifts (percentage-point difference from training):")
display(NCD_SHIFT_DIAGNOSTICS.sort_values("Max missingness Δ pp", ascending=False).head(12).round(2))

# ---------------------- 2. Training-only feature stability ------------------
# Use only the original TRAIN-country partition. Within it, each held-out fold
# contains countries unseen by that fold's fitted model.
train_idx = np.where(mtr_n)[0]
Xtr_all = Xn.iloc[train_idx].reset_index(drop=True)
ytr_all = yn.iloc[train_idx].reset_index(drop=True)
gtr_all = np.asarray(gn)[train_idx]

stab_rows = []
gkf_stab = GroupKFold(5)

# Use the best-performing model family from Section 22B as the diagnostic base,
# but refit independently inside every training-country fold.
diagnostic_base = clone(FINAL_NCD_MODEL)

for fold, (fi, fv) in enumerate(gkf_stab.split(Xtr_all, ytr_all, groups=gtr_all), start=1):
    mdl = clone(diagnostic_base)
    mdl.fit(Xtr_all.iloc[fi], ytr_all.iloc[fi])
    perm = permutation_importance(
        mdl, Xtr_all.iloc[fv], ytr_all.iloc[fv],
        scoring="r2", n_repeats=8, random_state=SEED+fold, n_jobs=1
    )
    for feat, imp, sd in zip(FEATS, perm.importances_mean, perm.importances_std):
        stab_rows.append({"Fold":fold, "Feature":feat, "Importance":imp, "Importance SD":sd})

NCD_FEATURE_STABILITY_LONG = pd.DataFrame(stab_rows)
NCD_FEATURE_STABILITY = (
    NCD_FEATURE_STABILITY_LONG.groupby("Feature")
    .agg(
        Mean_importance=("Importance","mean"),
        SD_importance=("Importance","std"),
        Positive_folds=("Importance", lambda s: int((s>0).sum())),
        Min_importance=("Importance","min"),
        Max_importance=("Importance","max")
    )
    .reset_index()
)
NCD_FEATURE_STABILITY["Positive fold share"] = NCD_FEATURE_STABILITY["Positive_folds"]/5

# Conservative training-derived stability rule:
# positive mean importance AND positive in >= 3 of 5 grouped folds.
STABLE_NCD_FEATS = NCD_FEATURE_STABILITY.loc[
    (NCD_FEATURE_STABILITY["Mean_importance"] > 0) &
    (NCD_FEATURE_STABILITY["Positive_folds"] >= 3),
    "Feature"
].tolist()

# Safety floor: if the rule yields too few variables, use the top 8 by
# training-fold mean permutation importance. This remains training-derived.
if len(STABLE_NCD_FEATS) < 6:
    STABLE_NCD_FEATS = (
        NCD_FEATURE_STABILITY.sort_values("Mean_importance", ascending=False)
        .head(min(8, len(FEATS)))["Feature"].tolist()
    )

banner("TRAINING-ONLY NCD FEATURE STABILITY")
display(NCD_FEATURE_STABILITY.sort_values(
    ["Positive_folds","Mean_importance"], ascending=[False,False]
).round(4))
print(f"\nStable NCD feature set ({len(STABLE_NCD_FEATS)} of {len(FEATS)} predictors):")
print(STABLE_NCD_FEATS)

# ---------------------- 3. Stable-feature sensitivity -----------------------
# Compare two lower-variance specifications on the training-derived feature set:
# HistGradientBoosting and Elastic Net. Tune only on training countries.
Xs = Xn[STABLE_NCD_FEATS].copy()
Xstr = Xs.loc[mtr_n]
ystr = yn.loc[mtr_n]
gstr = np.asarray(gn)[mtr_n]

STABLE_HGB_BASE = PIPE(HistGradientBoostingRegressor(
    random_state=SEED, early_stopping=True,
    validation_fraction=0.15, n_iter_no_change=25
))
STABLE_HGB_SEARCH = RandomizedSearchCV(
    STABLE_HGB_BASE,
    param_distributions=HGB_PARAM_DIST,
    n_iter=30,
    scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
    refit="r2",
    cv=GroupKFold(5),
    random_state=SEED+223,
    n_jobs=1
)
STABLE_HGB_SEARCH.fit(Xstr, ystr, groups=gstr)
STABLE_HGB = STABLE_HGB_SEARCH.best_estimator_

STABLE_EN_BASE = PIPE(ElasticNet(random_state=SEED))
STABLE_EN_SEARCH = RandomizedSearchCV(
    STABLE_EN_BASE,
    param_distributions=EN_PARAM_DIST,
    n_iter=30,
    scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
    refit="r2",
    cv=GroupKFold(5),
    random_state=SEED+224,
    n_jobs=1
)
STABLE_EN_SEARCH.fit(Xstr, ystr, groups=gstr)
STABLE_EN = STABLE_EN_SEARCH.best_estimator_

def _stable_eval(label, model):
    cv = cross_validate(
        clone(model), Xs, yn, groups=gn, cv=GroupKFold(5),
        scoring={"r2":"r2","mae":"neg_mean_absolute_error"},
        return_train_score=True, n_jobs=1
    )
    row = {
        "Model": label,
        "Features": len(STABLE_NCD_FEATS),
        "Grouped CV R²": np.mean(cv["test_r2"]),
        "R² SD": np.std(cv["test_r2"]),
        "Worst fold R²": np.min(cv["test_r2"]),
        "Grouped CV MAE": np.mean(-cv["test_mae"]),
        "Mean train CV R²": np.mean(cv["train_r2"]),
        "CV train−test gap": np.mean(cv["train_r2"])-np.mean(cv["test_r2"]),
    }
    # Split metrics are descriptive; test is not used for adoption.
    for split_name, mask in [("Train",mtr_n),("Validation",mva_n),("Test",mte_n)]:
        met = evaluate_regression(model, Xs.loc[mask], yn.loc[mask])
        row[f"{split_name} R²"] = met["R²"]
        row[f"{split_name} MAE"] = met["MAE"]
    row["Train−Validation gap"] = row["Train R²"]-row["Validation R²"]
    return row

NCD_STABLE_FEATURE_COMPARE = pd.DataFrame([
    _stable_eval("Stable-feature HistGradientBoosting", STABLE_HGB),
    _stable_eval("Stable-feature Elastic Net", STABLE_EN)
])

banner("STABLE-FEATURE SENSITIVITY")
display(NCD_STABLE_FEATURE_COMPARE.round(4))

# Compare stable-feature challengers against the Section 22B final model.
base_row = NCD_GENERALIZATION_COMPARE.loc[
    NCD_GENERALIZATION_COMPARE["Model"]==NCD_FINAL_CHOICE
].iloc[0]

# Adoption requires: lower train-validation gap, CV R² no worse than 0.02 below
# the Section 22B final model, and grouped-CV MAE no worse.
stable_eligible = NCD_STABLE_FEATURE_COMPARE[
    (NCD_STABLE_FEATURE_COMPARE["Train−Validation gap"] < base_row["Train−Validation gap"]) &
    (NCD_STABLE_FEATURE_COMPARE["Grouped CV R²"] >= base_row["Grouped CV R²"] - 0.02) &
    (NCD_STABLE_FEATURE_COMPARE["Grouped CV MAE"] <= base_row["Grouped CV MAE"])
].copy()

if stable_eligible.empty:
    NCD_22C_DECISION = "KEEP Section 22B final model"
    FINAL_NCD_FEATURES = FEATS
else:
    best_stable = stable_eligible.sort_values(
        ["Grouped CV MAE","Grouped CV R²"], ascending=[True,False]
    ).iloc[0]
    NCD_22C_DECISION = f"ADOPT {best_stable['Model']}"
    if best_stable["Model"] == "Stable-feature HistGradientBoosting":
        FINAL_NCD_MODEL = STABLE_HGB
    else:
        FINAL_NCD_MODEL = STABLE_EN
    BN_ = FINAL_NCD_MODEL
    FINAL_NCD_MODEL_NAME = best_stable["Model"]
    FINAL_NCD_FEATURES = STABLE_NCD_FEATS

print("\nSection 22C decision:")
print(NCD_22C_DECISION)
print("This decision does not use test R² or test MAE.")

# --------------------------- 4. Residual diagnostics ------------------------
# Evaluate residuals from the finally retained model on the appropriate feature set.
Xfinal_n = Xn[FINAL_NCD_FEATURES].copy()

resid_rows = []
for split_name, mask in [("Train",mtr_n),("Validation",mva_n),("Test",mte_n)]:
    pred = FINAL_NCD_MODEL.predict(Xfinal_n.loc[mask])
    actual = yn.loc[mask].to_numpy()
    resid = actual - pred
    for a,p,r in zip(actual,pred,resid):
        resid_rows.append({
            "Split":split_name, "Actual NCD mortality":a,
            "Predicted NCD mortality":p, "Residual":r,
            "Absolute error":abs(r)
        })

NCD_RESIDUAL_DIAGNOSTICS = pd.DataFrame(resid_rows)
NCD_RESIDUAL_SUMMARY = (
    NCD_RESIDUAL_DIAGNOSTICS.groupby("Split")
    .agg(
        Mean_residual=("Residual","mean"),
        Median_absolute_error=("Absolute error","median"),
        P90_absolute_error=("Absolute error", lambda s: np.quantile(s,.90)),
        Max_absolute_error=("Absolute error","max")
    )
    .reset_index()
)

banner("FINAL NCD RESIDUAL DIAGNOSTICS")
display(NCD_RESIDUAL_SUMMARY.round(3))

# Error by actual-outcome quintile helps detect failure at low/high NCD mortality.
tmp = NCD_RESIDUAL_DIAGNOSTICS.copy()
tmp["Outcome quintile"] = pd.qcut(
    tmp["Actual NCD mortality"], q=5, labels=["Q1 lowest","Q2","Q3","Q4","Q5 highest"],
    duplicates="drop"
)
NCD_ERROR_BY_OUTCOME = (
    tmp.groupby(["Split","Outcome quintile"], observed=True)["Absolute error"]
    .mean().reset_index(name="Mean absolute error")
)
display(NCD_ERROR_BY_OUTCOME.round(3))

# Final compact visual diagnostics.
fig, ax = plt.subplots(figsize=(10,5))
plot_shift = NCD_SHIFT_DIAGNOSTICS.sort_values("Max |SMD|", ascending=True).tail(12)
ax.barh(plot_shift["Feature"], plot_shift["Max |SMD|"])
ax.set_xlabel("Maximum absolute standardized mean difference")
ax.set_title("NCD predictor shift across country splits", fontweight="bold")
savefig("22C_ncd_covariate_shift")

fig, ax = plt.subplots(figsize=(10,5))
plot_stab = NCD_FEATURE_STABILITY.sort_values("Mean_importance", ascending=True).tail(12)
ax.barh(plot_stab["Feature"], plot_stab["Mean_importance"])
ax.set_xlabel("Mean grouped-fold permutation importance (ΔR²)")
ax.set_title("NCD training-only feature stability", fontweight="bold")
savefig("22C_ncd_feature_stability")

print(f"\nFINAL NCD MODEL AFTER DIAGNOSTIC REFINEMENT: {FINAL_NCD_MODEL_NAME}")
print(f"Final NCD feature count: {len(FINAL_NCD_FEATURES)}")
```

### Interpretation framework

The purpose of Section 22C is to distinguish **model overfitting** from **data heterogeneity**.

- Large absolute standardised mean differences indicate that held-out countries occupy a different predictor distribution from training countries.
- Large missingness differences indicate that the imputation problem itself changes across splits.
- A predictor that is positive in only one or two grouped folds is not considered stable evidence of generalisable predictive value.
- Residual concentration in the lowest or highest NCD-mortality quintiles suggests that the model has difficulty with specific parts of the outcome distribution rather than uniformly poor fit.
- A stable-feature model is retained only when it reduces the train–validation gap **without materially sacrificing grouped-CV R² or worsening grouped-CV MAE**.

This distinction matters for the capstone conclusion. If the diagnostics show meaningful country-level shift or unstable feature contribution, the appropriate conclusion is not that the modelling pipeline failed. Rather, premature NCD mortality is less transportable across countries using the available global predictors than HALE. That is itself a substantive predictive finding.

### Methodological boundary

Because the test split has already been inspected during earlier post-hoc refinement, it should no longer be described as a completely untouched test set for this diagnostic stage. The notebook therefore continues to treat repeated/grouped country validation as the primary estimate of generalisation.

### References supporting this diagnostic step

Roberts et al. (2017) support grouped validation when observations have hierarchical or other dependence structures. Moreno-Torres et al. (2012) provide a framework for understanding dataset shift when training and deployment distributions differ. Cawley and Talbot (2010) caution that repeatedly optimising model-selection criteria can itself create selection bias. Permutation importance is interpreted cautiously because correlated predictors can distribute importance across related variables rather than identify unique causal effects.


## **23  Do actionable levers add predictive value?**

### Purpose
Determine whether predictive skill comes from controllable levers or merely from national wealth and demography.

### Rationale
A model can score well entirely on variables a health policymaker cannot touch. If holding context fixed and adding the actionable block changes nothing, then the model is a description of how rich a country is without yielding a genuinely policy-informative model, and no recommendation follows from it however accurate it is.

### Method
Block ablation under grouped cross-validation: context variables alone, actionable levers alone, and both together. The comparison that matters is the incremental gain from adding levers to context, not the standalone performance of either block.

### Expected output
Ablation tables and a comparison chart for both outcomes.

```python
CTX_F = [v for v in FEATS if v in CONTEXT + GOVERNANCE + RISK]
ACT_F = [v for v in FEATS if v in ACTIONABLE]

def block_r2(X, y, g, cols, fitted_pipeline):
    if not cols: return np.nan
    return cross_val_score(clone(fitted_pipeline), X[cols], y, groups=g,
                           cv=GroupKFold(5), scoring="r2").mean()

def ablation(Xdf, y, g, fitted_pipeline, tag, model_name):
    blocks = [("Context only (not controllable)", CTX_F),
              ("Actionable levers only", ACT_F),
              ("Both", CTX_F + ACT_F)]
    rows = []
    key_model = re.sub(r"[^A-Za-z0-9]+","_",model_name.lower())
    for lbl, cols in blocks:
        key = f"abl_selected_{tag}_{key_model}_{hashlib.md5(lbl.encode()).hexdigest()[:6]}"
        r2 = cached(key, lambda cols=cols: block_r2(Xdf, y, g, cols, fitted_pipeline))
        rows.append({"Feature set":lbl, "Variables":len(cols), "Grouped CV R²":r2})
    return pd.DataFrame(rows)

ABL_H = ablation(Xh, yh, gh, FINAL_HALE_MODEL, "hale", FINAL_HALE_MODEL_NAME)
ABL_N = ablation(Xn, yn, gn, FINAL_NCD_MODEL, "ncd", FINAL_NCD_MODEL_NAME)
banner("H3b — BLOCK ABLATION")
print(f"HALE model: {FINAL_HALE_MODEL_NAME}; NCD model: {FINAL_NCD_MODEL_NAME}")
print("HALE"); display(ABL_H.round(4))
print("NCD mortality"); display(ABL_N.round(4))
inc_h = ABL_H.iloc[2]["Grouped CV R²"] - ABL_H.iloc[0]["Grouped CV R²"]
inc_n = ABL_N.iloc[2]["Grouped CV R²"] - ABL_N.iloc[0]["Grouped CV R²"]
H3b_HALE = inc_h > .02
H3b_NCD  = inc_n > .02
H3b = H3b_HALE or H3b_NCD   # prespecified overall rule: support in either outcome

print(f"\nIncremental R² from adding actionable levers — HALE {inc_h:+.3f}, NCD {inc_n:+.3f}")
print(f"Outcome-specific support — HALE: {'YES' if H3b_HALE else 'NO'}; "
      f"NCD: {'YES' if H3b_NCD else 'NO'}")
if H3b:
    supported = []
    if H3b_HALE: supported.append("HALE")
    if H3b_NCD: supported.append("NCD mortality")
    unsupported = []
    if not H3b_HALE: unsupported.append("HALE")
    if not H3b_NCD: unsupported.append("NCD mortality")
    msg = f"H3b₀ REJECTED under the prespecified overall rule because incremental predictive value exceeds +0.02 for {', '.join(supported)}."
    if unsupported:
        msg += f" No >+0.02 incremental benefit is observed for {', '.join(unsupported)}."
    print(msg)
else:
    print("H3b₀ not rejected — neither outcome shows >+0.02 incremental grouped-CV R² from adding actionable variables to context.")
```

### Model Specification and Validation

**Model:** Feature-block ablation using the **tuned selected model for each outcome** rather than a hard-coded algorithm.

**Selected models:** Retrieved programmatically from Sections 20–22.

**Feature blocks:** context only, actionable levers only, and both.

**Validation:** Five-fold country-grouped CV using clones of the tuned Section 22 pipelines.

**Executed result:** The authoritative block \(R^2\) values and the incremental gain from adding actionable variables to context are generated in the preceding code cell. They are intentionally not repeated as fixed numbers here.

**Interpretation:** The block comparison measures incremental predictive information, not causal effect size.


```python
fig, ax = plt.subplots(1, 2, figsize=(14,4.5))
for a, t, nm in zip(ax, [ABL_H, ABL_N], ["HALE","NCD mortality"]):
    a.bar(t["Feature set"], t["Grouped CV R²"], color=["#878787","#1b7837","#2166ac"])
    a.set_ylabel("Grouped CV R²"); a.set_title(nm, fontweight="bold")
    a.tick_params(axis="x", labelrotation=12, labelsize=8)
savefig("23_ablation")
```

### Result
The executable ablation cell reports the current grouped-CV \(R^2\) for context only, actionable levers only, and both, together with the incremental gain from adding actionable variables to context **separately for HALE and NCD mortality**.

### Interpretation
The H3b decision is an overall, prespecified **either-outcome** rule, but the substantive interpretation remains outcome-specific. A positive increment for HALE does not imply the same incremental value for NCD mortality. This distinction is especially important after applying the stricter 18-feature primary rule, because the three capacity variables with 30–45% missingness are excluded from the primary ablation and evaluated separately in Section 19B.

The incremental contribution is predictive information beyond context. It is not a percentage improvement in health and is not a causal effect of changing the actionable variables.

### Decision
Apply the prespecified overall H3b rule, while explicitly reporting which outcome provides the support and which does not. Do not summarize the result as if actionable variables improve both outcomes unless both outcome-specific increments exceed the decision threshold.

## **24  Which variables contribute most to prediction?**

### Purpose
Explain the tuned selected machine-learning models by attributing predictive importance to individual variables and separating institutionally actionable variables from contextual variables.

### Rationale
Out-of-sample predictive performance alone does not identify which predictors the fitted model relies on most. Permutation importance provides a post-hoc measure of predictive contribution and is not interpreted as a causal effect; correlated predictors can also complicate attribution of importance (Fisher et al., 2019).

### Method
Permutation importance is computed on the fixed held-out test countries from Section 22 using the **actual tuned selected model for each outcome**. Each feature is shuffled eight times and importance is the mean reduction in test-set \(R^2\). The test set remains excluded from model fitting and hyperparameter selection, although it is reused here for post-hoc interpretation.

### Expected output
Ranked importance tables and charts for HALE and NCD mortality, plus the actionable share of total positive permutation importance.


```python
from sklearn.inspection import permutation_importance
def perm_imp(Xdf, y, masks, fitted_model, cols, tag, model_name):
    def run():
        mte = masks[2]
        r = permutation_importance(fitted_model, Xdf.iloc[mte], y.iloc[mte], n_repeats=8,
                                   random_state=SEED, scoring="r2", n_jobs=-1)
        return pd.DataFrame({"Variable":cols, "Importance":r.importances_mean,
                             "SD":r.importances_std}).sort_values("Importance", ascending=False)
    key_model = re.sub(r"[^A-Za-z0-9]+","_",model_name.lower())
    return cached(f"perm_selected_{tag}_{key_model}", run)

PI_H = perm_imp(Xh, yh, MH_, FINAL_HALE_MODEL, FEATS, "hale", FINAL_HALE_MODEL_NAME)
PI_N = perm_imp(Xn, yn, MN_, FINAL_NCD_MODEL, FEATS, "ncd", FINAL_NCD_MODEL_NAME)
for t in (PI_H, PI_N):
    t["Controllable"] = np.where(t.Variable.isin(ACTIONABLE), "ACTIONABLE", "context")
banner(f"HALE — {FINAL_HALE_MODEL_NAME} permutation importance"); display(PI_H.round(4))
banner(f"NCD — {FINAL_NCD_MODEL_NAME} permutation importance"); display(PI_N.round(4))

fig, ax = plt.subplots(1, 2, figsize=(16,6))
for a, t, nm in zip(ax, [PI_H, PI_N], ["HALE","NCD mortality"]):
    t2 = t.head(14).sort_values("Importance")
    a.barh(t2.Variable, t2.Importance, xerr=t2.SD, capsize=3,
           color=np.where(t2.Controllable=="ACTIONABLE","#1b7837","#878787"))
    a.set_xlabel("drop in R² when shuffled"); a.set_title(nm, fontweight="bold")
ax[0].legend(handles=[Line2D([],[],color="#1b7837",lw=6,label="ministry can move it"),
                      Line2D([],[],color="#878787",lw=6,label="context")], fontsize=8)
plt.suptitle("Permutation importance of the tuned selected models", fontweight="bold", y=1.02)
savefig("24_importance")
sh = PI_H[PI_H.Controllable=="ACTIONABLE"].Importance.sum() / PI_H.Importance.clip(lower=0).sum()
sn = PI_N[PI_N.Controllable=="ACTIONABLE"].Importance.sum() / PI_N.Importance.clip(lower=0).sum()
print(f"\nActionable share of positive permutation importance — HALE {sh*100:.0f}%, NCD {sn*100:.0f}%.")
```

### Model Explanation Specification

**Procedure:** Permutation importance applied to the tuned automatically selected Section 22 models.

**Models explained:** Retrieved programmatically from the selected/tuned model objects.

**Method:** Each feature is shuffled on the locked test set for 8 repeats; importance is the mean reduction in test-set \(R^2\). This is post-hoc predictive explanation, not a causal effect.

**Executed result:** The current importance rankings, uncertainty across shuffles, and actionable share of positive importance are generated directly by the preceding executable cell. No feature ranking or importance value is hard-coded here.

**Interpretation:** A larger value means held-out test \(R^2\) falls more when that variable is disrupted. It does not imply that intervening on the variable will cause the outcome to change, and correlated predictors can share or dilute importance.


### Result
The tuned selected models rely on a mixture of actionable and contextual predictors. The current leading variables for each outcome are the first rows of the executed permutation-importance tables above.

### Interpretation
Permutation importance describes dependence of the fitted predictive model on a variable in the held-out test countries. It is not an effect estimate, does not provide direction of effect, and rankings can change when the selected algorithm, tuned hyperparameters, data, or correlated feature set changes.

### Decision
Use the executed Section 24 rankings as a predictive shortlist and cross-check them against the explanatory evidence in Part III; do not convert importance rank into a causal policy recommendation.

## **25  The morbidity gap across two decades**

### Purpose
Establish the empirical distinction between total life expectancy and healthy life expectancy by describing the morbidity gap and how it has changed over time.

### Rationale
This is the question that justifies using healthy life expectancy as the primary outcome. A country can raise life expectancy by keeping people alive with chronic disease — a real achievement, and an expensive one, but not the same thing as improving health. Whether the gap between the two is widening or narrowing is measurable and carries a direct budget implication.

### Method
The gap is computed as life expectancy minus HALE, with impossible values already suppressed in Section 8. Trends are shown overall and by income group, alongside the cross-sectional relationship between the gap and health spending.

### Expected output
Descriptive statistics for the gap and a three-panel figure covering trend, income gradient and the spending relationship.

```python
gapf = eng.dropna(subset=["MorbidityGap","HALE","LifeExpectancy"])
print(f"Morbidity-gap frame: {len(gapf):,} country-years, {gapf.iso3.nunique()} countries")
print(f"Mean gap {gapf.MorbidityGap.mean():.2f} years   SD {gapf.MorbidityGap.std():.2f}")
print(f"Range    {gapf.MorbidityGap.min():.2f} to {gapf.MorbidityGap.max():.2f} years")

fig, ax = plt.subplots(1, 3, figsize=(17,5))
tr = gapf.groupby("Year")[["LifeExpectancy","HALE"]].mean()
ax[0].plot(tr.index, tr.LifeExpectancy, lw=2.5, c="#2166ac", label="Life expectancy")
ax[0].plot(tr.index, tr.HALE, lw=2.5, c="#1b7837", label="Healthy life expectancy")
ax[0].fill_between(tr.index, tr.HALE, tr.LifeExpectancy, alpha=.2, color="#b2182b",
                   label="years lived in poor health")
ax[0].set_ylabel("Years"); ax[0].legend(fontsize=8)
ax[0].set_title("The gap has not closed", fontweight="bold")

for g in order:
    s = gapf[gapf.IncomeGroup==g].groupby("Year").MorbidityGap.mean()
    if len(s): ax[1].plot(s.index, s.values, lw=2.2, label=g)
ax[1].set_ylabel("Morbidity gap (years)"); ax[1].legend(fontsize=8)
ax[1].set_title("Richer countries live longer in poor health", fontweight="bold")

rc = gapf[gapf.Year.between(2017,2021)].groupby("iso3").agg(
    gap=("MorbidityGap","mean"), he=("logHealthExpPC","mean"), inc=("IncomeGroup","first")).dropna()
for g in order:
    z = rc[rc.inc==g]
    ax[2].scatter(z.he, z.gap, s=28, alpha=.75, c=pal.get(g,"#888"), label=g, edgecolor="none")
b1,b0 = np.polyfit(rc.he, rc.gap, 1); xs = np.linspace(rc.he.min(), rc.he.max(), 40)
ax[2].plot(xs, b0+b1*xs, "k--", lw=1.6)
ax[2].set_xlabel("log health spending per capita"); ax[2].set_ylabel("Morbidity gap (years)")
ax[2].set_title(f"r = {rc.he.corr(rc.gap):+.3f}", fontweight="bold")
savefig("25_morbidity_gap")
```

## **26  Same model, three outcomes**

### Purpose
Test formally whether health spending shifts life expectancy and healthy life expectancy by the same amount.

### Rationale
If spending bought purely healthy years, the coefficient on the morbidity gap would be zero. If it bought survival with disease, the gap coefficient would be positive. Because the gap is defined as life expectancy minus HALE, the gap equation's coefficient is exactly the difference between the other two, which makes it a formal test rather than an informal comparison of two separately estimated models.

### Method
All three outcomes are estimated on a **common sample** with identical right-hand sides, because the coefficient identity holds only under that condition. Two specifications are reported: two-way fixed effects for within-country movement, and the Mundlak between-country terms for long-run level. An explicit identity check confirms the arithmetic in the output.

### Expected output
Coefficient and p-value grids for both specifications, with the identity verified numerically.

```python
RHS4 = [v for v in ["logHealthExpPC","UHCIndex","Physicians","logGDPPerCapita",
                    "Pop65Plus","UrbanPopulation"] if v in eng.columns]

# A COMMON sample is essential. Because MorbidityGap = LifeExpectancy − HALE exactly, the
# coefficient in the gap equation equals (b_LE − b_HALE) only when all three are estimated on
# identical rows with identical right-hand sides. That identity is what makes the gap equation
# a formal test of H4a rather than an informal eyeball comparison of two separate models.
COMMON4 = eng[["LifeExpectancy","HALE","MorbidityGap","iso3","Year"] + RHS4].dropna()
print(f"Common sample: {len(COMMON4):,} country-years, {COMMON4.iso3.nunique()} countries")

def three_outcomes(d, rhs, spec):
    rows = {}
    for outc in ["LifeExpectancy","HALE","MorbidityGap"]:
        if spec == "twfe":
            f = f"{outc} ~ " + " + ".join(rhs) + " + C(iso3) + C(Year)"
        else:                                     # Mundlak / correlated random effects
            f = f"{outc} ~ " + " + ".join([f"{v}_btw + {v}_wth" for v in rhs]) + " + C(Year)"
        rows[outc] = smf.ols(f, data=d).fit(cov_type="cluster", cov_kwds={"groups": d.iso3})
    return rows

M4_FE = three_outcomes(COMMON4, RHS4, "twfe")

# The Mundlak version needs the between/within columns on the same rows
C4 = eng[["LifeExpectancy","HALE","MorbidityGap","iso3","Year"] + RHS4 +
         [f"{v}_{s2}" for v in RHS4 for s2 in ("btw","wth") if f"{v}_{s2}" in eng.columns]].dropna()
M4_MU = three_outcomes(C4, [v for v in RHS4 if f"{v}_btw" in C4.columns], "mundlak")

def coef_grid(mods, terms):
    b = pd.DataFrame({k: [m.params.get(t, np.nan) for t in terms] for k, m in mods.items()}, index=terms)
    p = pd.DataFrame({k: [m.pvalues.get(t, np.nan) for t in terms] for k, m in mods.items()}, index=terms)
    return b[["LifeExpectancy","HALE","MorbidityGap"]], p[["LifeExpectancy","HALE","MorbidityGap"]]

PIV,  PIVP  = coef_grid(M4_FE, RHS4)
PIV2, PIVP2 = coef_grid(M4_MU, [f"{v}_btw" for v in RHS4 if f"{v}_btw" in C4.columns])

banner("SPECIFICATION A — two-way fixed effects (within-country movement)")
display(PIV.round(4)); display(PIVP.round(4))
banner("SPECIFICATION B — Mundlak between-country terms (long-run level)")
display(PIV2.round(4)); display(PIVP2.round(4))

H4 = pd.concat([
    PIV.stack().rename("Coefficient").reset_index().rename(
        columns={"level_0":"Lever","level_1":"Outcome"}).assign(Specification="Two-way FE"),
    PIV2.stack().rename("Coefficient").reset_index().rename(
        columns={"level_0":"Lever","level_1":"Outcome"}).assign(Specification="Mundlak (between)"),
])
H4["p"] = np.concatenate([PIVP.stack().values, PIVP2.stack().values])
```

### Model Specification and Equation

**Model family:** Common-sample three-outcome models estimated as (A) two-way fixed effects and (B) Mundlak decomposition.

**Purpose:** Compare the health-spending association with total life expectancy, HALE, and the morbidity gap under identical right-hand sides.

**Outcomes (Y):** `LifeExpectancy`, `HALE`, and `MorbidityGap`.

**Predictors (X):** `logHealthExpPC`, `UHCIndex`, `Physicians`, `logGDPPerCapita`, `Pop65Plus`, `UrbanPopulation`.

#### A. Two-way fixed-effects models

General specification:

\[
Y_{it}=\alpha_i+\lambda_t+\beta_1logHealthExpPC_{it}+\beta_2UHC_{it}
+\beta_3Physicians_{it}+\beta_4logGDP_{it}+\beta_5Pop65_{it}
+\beta_6Urban_{it}+\varepsilon_{it}
\]

Using the executed coefficients:

\[
\widehat{LifeExpectancy}_{it}=\alpha_i+\lambda_t
-0.2550logHE+0.1374UHC-0.5606Physicians+0.7448logGDP-0.2938Pop65-0.0068Urban
\]

\[
\widehat{HALE}_{it}=\alpha_i+\lambda_t
-0.3459logHE+0.1490UHC-0.5470Physicians+1.2243logGDP-0.3167Pop65-0.0629Urban
\]

\[
\widehat{MorbidityGap}_{it}=\alpha_i+\lambda_t
+0.0908logHE-0.0116UHC-0.0136Physicians-0.4795logGDP+0.0229Pop65+0.0561Urban
\]

**Estimation:** Common sample of 2,556 country-years across 181 countries; country-clustered standard errors.

#### B. Mundlak models

General specification:

\[
Y_{it}=\beta_0+\sum_k\beta_{B,k}\bar X_{ik}
+\sum_k\beta_{W,k}(X_{it}-\bar X_i)+\lambda_t+\varepsilon_{it}
\]

The existing output prints the **between-country** coefficients used for the long-run comparison:

| Predictor (between) | Life expectancy | HALE | Morbidity gap |
|---|---:|---:|---:|
| logHealthExpPC | -0.7332 | -1.4238 | 0.6906 |
| UHCIndex | 0.2338 | 0.2066 | 0.0273 |
| Physicians | -0.2427 | -0.2048 | -0.0379 |
| logGDPPerCapita | 2.3049 | 2.3053 | -0.0004 |
| Pop65Plus | 0.2621 | 0.2805 | -0.0184 |
| UrbanPopulation | 0.0224 | 0.0176 | 0.0048 |

The corresponding within coefficients, intercepts, and year-dummy coefficients are not printed in this notebook output and are therefore not manually reconstructed.

**Performance:** The purpose is explanatory coefficient comparison. No predictive-performance claim is made from these regressions, and no model \(R^2\) is printed in this section.

```python
def h4_verdict(bframe, pframe, spend_term, label):
    if spend_term not in bframe.index: return None
    bLE, bH, bG = bframe.loc[spend_term]
    pG = pframe.loc[spend_term, "MorbidityGap"]
    banner(f"H4a — {label}")
    print(f"Associated with a doubling of health spending per capita:")
    print(f"   life expectancy       {bLE*np.log(2):+.3f} years   (p = {pframe.loc[spend_term,'LifeExpectancy']:.4g})")
    print(f"   healthy life expect.  {bH *np.log(2):+.3f} years   (p = {pframe.loc[spend_term,'HALE']:.4g})")
    print(f"   morbidity gap         {bG *np.log(2):+.3f} years   (p = {pG:.4g})")
    print(f"\n   identity check: b_LE − b_HALE = {bLE-bH:+.4f} vs gap coefficient {bG:+.4f}")
    if bLE > 0:
        print(f"   share of the additional years that are healthy: {bH/bLE*100:.0f}%")
    rej = pG < .05
    print(f"\nH4a₀ {'REJECTED' if rej else 'not rejected'} in this specification.")
    if rej and bG > 0:
        print("The gap WIDENS. Spending buys survival faster than it buys health — which is")
        print("precisely why a health budget evaluated against life expectancy alone will be")
        print("scored a success while the population spends more of its life unwell.")
    elif rej and bG < 0:
        print("The gap NARROWS — spending compresses morbidity into fewer years.")
    return rej

r_fe = h4_verdict(PIV,  PIVP,  "logHealthExpPC",     "two-way fixed effects (within-country)")
r_mu = h4_verdict(PIV2, PIVP2, "logHealthExpPC_btw", "Mundlak between-country (long-run level)")
```

### Result
Neither Section 26 panel specification rejects H4a₀ at the 5% level. In the two-way fixed-effects specification, the morbidity-gap spending association is not statistically significant. In the Mundlak between-country specification, the morbidity-gap association is positive but also does not reach \(p<.05\).

### Interpretation
These results do not establish that the spending associations with life expectancy and HALE are identical; failure to reject H4a₀ is not evidence that the null is true. Rather, the Section 26 panel specifications do not provide statistically significant evidence of a differential spending association at the 5% level.

### Decision
Treat the Section 26 estimates as specification-specific panel evidence. The formal H4a test remains the pre-specified Section 27 analysis identified in the notebook's hypothesis framework, while the Section 26 results provide sensitivity evidence on robustness across panel specifications.

### Link to next step
Section 27 applies the pre-specified cross-country H4a test and reports whether its conclusion is consistent with the Section 26 panel evidence.


## **27  Why the level coefficients are uninformative but their difference is not**

### Purpose
Estimate the spending-to-morbidity-gap relationship in the presence of near-perfect collinearity between spending and income.

### Rationale
Between countries, health spending per capita and GDP per capita are almost the same variable. With both in the model neither can be estimated precisely, and the coefficients on life expectancy and healthy life expectancy become individually meaningless. That is textbook collinearity and would normally be the end of the matter.

### Method
It is not the end here, because the hypothesis concerns the **difference** between the two coefficients rather than either level. The collinear income component enters both equations identically and cancels out of their difference, so the gap coefficient remains identified where the levels are not. The cross-section is estimated with and without the income control to demonstrate this directly.

### Expected output
A coefficient table under both control sets, and a figure contrasting the unstable level estimates against the stable gap estimate.

```python
BTW = (eng.groupby(["iso3","Country","IncomeGroup"], as_index=False)
          [["LifeExpectancy","HALE","MorbidityGap","logHealthExpPC","logGDPPerCapita",
            "Pop65Plus","UrbanPopulation"]].mean()).dropna()
print(f"Between-country cross-section: {len(BTW)} countries (all years averaged)")
r_hg = BTW.logHealthExpPC.corr(BTW.logGDPPerCapita)
print(f"corr(log health spending, log GDP) between countries = {r_hg:+.4f}  <-- the problem\n")

rows = []
for lbl, ctrl in [("without income control", ""), ("with income control", " + logGDPPerCapita")]:
    for o in ["LifeExpectancy","HALE","MorbidityGap"]:
        m = smf.ols(f"{o} ~ logHealthExpPC + Pop65Plus + UrbanPopulation{ctrl}",
                    data=BTW).fit(cov_type="HC3")
        rows.append({"Specification":lbl, "Outcome":o, "b (log spending)":m.params.logHealthExpPC,
                     "SE":m.bse.logHealthExpPC, "p":m.pvalues.logHealthExpPC, "n":int(m.nobs)})
BTW_T = pd.DataFrame(rows)
display(BTW_T.round(4))

fig, ax = plt.subplots(1, 2, figsize=(14,5))
for a, lbl in zip(ax, BTW_T.Specification.unique()):
    t = BTW_T[BTW_T.Specification == lbl]
    a.bar(t.Outcome, t["b (log spending)"], yerr=1.96*t.SE, capsize=6,
          color=["#2166ac","#1b7837","#b2182b"])
    a.axhline(0, c="k", lw=1)
    a.set_ylabel("Coefficient on log health spending per capita (years)")
    a.set_title(lbl, fontweight="bold")
    for i,(_,r) in enumerate(t.iterrows()):
        a.text(i, r["b (log spending)"], f"\np = {r.p:.2g}", ha="center", va="top", fontsize=8)
plt.suptitle("The levels lose significance to collinearity; the gap does not",
             fontweight="bold", y=1.02)
savefig("26_gap_identification")

gap_row = BTW_T[(BTW_T.Outcome=="MorbidityGap") & (BTW_T.Specification=="with income control")].iloc[0]
gap_raw = BTW_T[(BTW_T.Outcome=="MorbidityGap") & (BTW_T.Specification=="without income control")].iloc[0]
banner("H4a — BETWEEN-COUNTRY CROSS-SECTION (pre-specified headline specification)")
print(f"Morbidity-gap coefficient on log spending:")
print(f"   without income control  {gap_raw['b (log spending)']:+.4f}  p = {gap_raw.p:.3g}")
print(f"   with income control     {gap_row['b (log spending)']:+.4f}  p = {gap_row.p:.3g}")
print(f"\nA doubling of health spending per capita is associated with a morbidity gap")
print(f"{gap_row['b (log spending)']*np.log(2):+.3f} years wider, holding income and demography constant.")
H4a = bool(gap_row.p < .05 and gap_raw.p < .05)
print(f"\nH4a₀ {'REJECTED' if H4a else 'not rejected'} in the pre-specified Section 27 cross-country test.")
if H4a:
    print("The income-controlled morbidity-gap coefficient provides statistically significant evidence")
    print("that the spending associations with life expectancy and HALE differ in this specification.")
else:
    print("This specification does not provide statistically significant evidence that the spending")
    print("associations with life expectancy and HALE differ.")
print("\nThis conclusion is specification-dependent: the Section 26 TWFE and Mundlak panel")
print("specifications do not reject H4a₀ at the 5% level. The cross-country result is therefore")
print("not robust across all specifications and should be interpreted as an observational association,")
print("not as evidence that additional health spending causes the morbidity gap to widen.")
```

### Model Specification and Equation

**Model:** Between-country HC3 OLS spending models, estimated with and without income control.

**Purpose:** Test whether health spending is associated differently with life expectancy and HALE by estimating the morbidity-gap equation on country averages.

**Outcomes (Y):** `LifeExpectancy`, `HALE`, and `MorbidityGap`.

**Predictors (X):**
- Without income control: `logHealthExpPC`, `Pop65Plus`, `UrbanPopulation`.
- With income control: the same variables plus `logGDPPerCapita`.

**General equations:**

Without income:
\[
Y_i=\beta_0+\beta_1logHealthExpPC_i+\beta_2Pop65_i+\beta_3Urban_i+\varepsilon_i
\]

With income:
\[
Y_i=\beta_0+\beta_1logHealthExpPC_i+\beta_2Pop65_i+\beta_3Urban_i+
\beta_4logGDPPerCapita_i+\varepsilon_i
\]

The executed notebook prints the focal spending coefficient \(\beta_1\), SE, and \(p\)-value for all six fitted models:

| Specification | Outcome | b(log spending) | SE | p |
|---|---|---:|---:|---:|
| Without income | Life expectancy | 3.1835 | 0.4713 | <.001 |
| Without income | HALE | 2.2374 | 0.4278 | <.001 |
| Without income | Morbidity gap | 0.9227 | 0.1375 | <.001 |
| With income | Life expectancy | 0.8203 | 1.2771 | .5207 |
| With income | HALE | -0.1413 | 1.1582 | .9029 |
| With income | Morbidity gap | 0.8594 | 0.4002 | .0318 |

The intercept and non-focal control coefficients are not printed in the executed output, so complete numerical fitted equations are not invented.

**Estimation:** HC3 OLS on 183 country averages. Between-country correlation of log spending with log GDP is .9716.

**Interpretation:** In the income-adjusted morbidity-gap model, a doubling of health spending corresponds to \(0.8594\times\ln(2)=0.596\) additional morbidity-gap years. This is an explanatory association, not a causal or predictive-performance statistic.

```python
# Mundlak between-country coefficient figure with model-based 95% CIs and p-value markers
MU_TERMS = list(PIV2.index)
MU_OUTCOMES = list(PIV2.columns)

# Pull standard errors directly from the fitted Mundlak models so the figure
# updates automatically whenever the notebook is rerun.
PIV2_SE = pd.DataFrame(
    {
        outcome: [M4_MU[outcome].bse.get(term, np.nan) for term in MU_TERMS]
        for outcome in MU_OUTCOMES
    },
    index=MU_TERMS,
)

def sig_marker(p):
    if pd.isna(p):
        return ""
    if p < .001:
        return "***"
    if p < .01:
        return "**"
    if p < .05:
        return "*"
    return "ns"

fig, ax = plt.subplots(figsize=(11, 6.2))
xs = np.arange(len(PIV2))
w = .27

all_low, all_high = [], []
for i, (outcome, col) in enumerate(zip(MU_OUTCOMES, ["#2166ac","#1b7837","#b2182b"])):
    x = xs + (i-1)*w
    b = PIV2[outcome].to_numpy()
    se = PIV2_SE[outcome].to_numpy()
    ci = 1.96 * se

    ax.bar(x, b, w, label=outcome, color=col, zorder=2)
    ax.errorbar(x, b, yerr=ci, fmt="none", ecolor="black",
                elinewidth=1.1, capsize=3.5, capthick=1.1, zorder=3)

    all_low.extend((b-ci).tolist())
    all_high.extend((b+ci).tolist())

# Use the observed CI span to position annotations without hard-coded significance labels.
data_min = np.nanmin(all_low)
data_max = np.nanmax(all_high)
yrange = data_max - data_min
offset = 0.035 * yrange

for i, outcome in enumerate(MU_OUTCOMES):
    x = xs + (i-1)*w
    for j, term in enumerate(MU_TERMS):
        b = PIV2.loc[term, outcome]
        ci = 1.96 * PIV2_SE.loc[term, outcome]
        p = PIVP2.loc[term, outcome]
        label = sig_marker(p)

        if b >= 0:
            y = b + ci + offset
            va = "bottom"
        else:
            y = b - ci - offset
            va = "top"

        ax.text(x[j], y, label, ha="center", va=va,
                fontsize=8, fontweight="bold", zorder=4)

ax.set_xticks(xs)
ax.set_xticklabels([v.replace("_btw","") for v in PIV2.index],
                   rotation=18, ha="right", fontsize=8)
ax.axhline(0, color="black", lw=1)
ax.set_ylabel("Mundlak between-country coefficient (years)")
ax.set_title("The same lever, judged against three different definitions of success",
             fontweight="bold")
ax.legend(frameon=False, ncol=3, loc="upper left")

# Leave room for significance markers and the figure note.
ax.set_ylim(data_min - 2.2*offset, data_max + 2.2*offset)
fig.subplots_adjust(bottom=0.25)
fig.text(
    0.5, 0.035,
    "Note. Bars show Mundlak between-country coefficients; error bars show 95% CIs. "
    "*** p<0.001, ** p<0.01, * p<0.05, ns p≥0.05. "
    "Associations are adjusted between-country relationships and do not imply causality.",
    ha="center", va="bottom", fontsize=8.5
)

savefig("26_three_outcomes")
```

### Result
In the pre-specified Section 27 cross-country test, the morbidity-gap coefficient is positive and statistically significant both without and with income control. In the income-controlled specification, the coefficient is +0.8594 (\(p=.0318\)).

### Interpretation
The Section 27 result provides statistically significant evidence that the spending associations with life expectancy and HALE differ in this cross-country specification. However, the overall evidence is **specification-dependent**: neither the Section 26 two-way fixed-effects estimate nor the Section 26 Mundlak between-country estimate reaches \(p<.05\) for the morbidity-gap spending term. One significant specification therefore does not establish robustness across specifications, and the observational association should not be interpreted causally.

### Decision
H4a₀ is **REJECTED** based on the pre-specified Section 27 cross-country hypothesis test; however, this finding is not robust across the Section 26 panel specifications. Failure to reject H4a₀ in those panel models does not constitute acceptance of the null or proof that no relationship exists.

## **32  Conclusion**

This study addressed five connected questions within a single global analytical framework: how efficiently health systems convert resources into healthy years of life; which actionable health-system characteristics are associated with that conversion and with HALE; whether HALE and premature NCD mortality can be predicted and explained for countries not seen during model training; whether health expenditure relates differently to healthy life expectancy and total life expectancy; and how the resulting benchmark can be applied to Saudi Arabia. The **master dataset covers 217 countries and economies from 2000–2023**, while the primary modelling sample contains **185 countries over 2000–2021**, reflecting the common usable window of the principal WHO outcomes.

The resource-demography benchmark explains a substantial share of international variation in HALE, but the remaining benchmark residual should not be interpreted as pure causal efficiency. It can contain conversion performance together with omitted variables, measurement error, model misspecification, and other unobserved country characteristics. Persistence across the early and late windows makes a purely transient-noise interpretation less plausible and supports its use for benchmarking, but persistence alone does not prove actionability or causality.

The final Section 28 synthesis dynamically preserves the distinction between nominal and BH-FDR evidence in the Section 14 screening analysis. For RQ2, maximal-sample results are reported separately from the common-sample reach-versus-capacity comparison, which is the fairer head-to-head and is not treated as a formal demonstration that reach is superior to capacity. The current interaction signs and significance counts are retrieved from Section 17 and interpreted as evidence consistent with diminishing combined associations rather than a causal weakening of one health-system lever by another. These results identify priorities for further evaluation without establishing that reallocating resources toward a specific lever will necessarily improve HALE.

The choice of health outcome also affects interpretation, and H4a remains explicitly specification-dependent. Section 28 retrieves the current morbidity-gap estimates from the Section 26 TWFE and Mundlak specifications and the pre-specified Section 27 income-controlled cross-country test. The formal decision follows the pre-specified primary test, while the panel specifications remain visible as sensitivity evidence. A significant association in one specification is not described as robustness across all specifications, and none of these observational estimates establishes that additional spending causes the morbidity gap to change.

The predictive analysis provides a complementary perspective. Section 28 retrieves the currently selected HALE and NCD model names and held-out-country \(R^2\), MAE, and RMSE directly from the final executed machine-learning objects, while Section 29 retrieves current permutation-importance rankings from Section 24. These quantities describe out-of-sample predictive performance and predictive dependence. They are not percentage health improvements, regression coefficients, or causal effects of changing individual health-system variables, and held-out performance within the assembled dataset does not prove universal external validity.

The missing-data sensitivity analysis adds an important qualification and robustness check. The primary RQ3 specification uses **18 predictors meeting the ≥70%-observed (≤30%-missing) rule**. A separate **21-predictor sensitivity specification** reintroduces Physicians, NursesMidwives, and HospitalBeds using median imputation fitted only inside training folds. Across all eight trainable model families, HALE performance is essentially unchanged and model rankings are highly stable; Extra Trees remains first, with grouped-CV \(R^2\) moving only from **0.810 to 0.808**. NCD mortality is more sensitive to the broader feature set: Extra Trees remains first, but its grouped-CV \(R^2\) improves from approximately **0.600 to 0.633**, with improvements also observed across other model families. Rank correlation is **0.976 for both outcomes**, indicating that the overall model-family conclusion is robust even though NCD predictive performance is more dependent on partially observed capacity information. These analyses are robustness evidence and do not replace the prespecified primary feature rule.

Taken together, the five analytical themes support a unified interpretation of health-system performance: resource levels alone do not determine healthy longevity; actionable characteristics should be evaluated separately from inherited context; machine learning can add out-of-sample predictive and explanatory information without implying causality; total longevity and healthy longevity should not be treated as interchangeable outcomes; and country benchmarking can translate the global evidence into a structured diagnostic application. For Saudi Arabia, the framework is most appropriately used to identify comparative performance gaps and priorities for deeper investigation rather than to prescribe causal interventions.
